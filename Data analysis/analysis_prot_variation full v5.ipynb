```python
# LOAD DATA
# =========================
#csv_path1 = r"Z:\\EPFL\\Biology\bachelor\output\05_sextile_exact_predictable_if.csv"
csv_path1 = r"Z:\\EPFL\\Biology\\bachelor\\Rep1-TEF1-h2b.csv"
csv_path2 = r"Z:\\EPFL\\Biology\\bachelor\\3T3-rep1-H2B-H3_3-rep2-TEAD1-GAPDH-PolII-BRG1 data.csv"
csv_path3 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\3T3_H3_3_rep2 data.csv"
csv_path4 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\3T3_GAPDH_PolII_BRG1_rep1 data.csv"
#csv_path5 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\mESCs Rep1 all proteins true version.csv"
csv_path6 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\mESCs Rep2 all proteins true version.csv"
#csv_path6 = r"Z:\\EPFL\Biology\bachelor\\mock_protein_data_predictable.csv"
csv_path5 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\mESCs Rep1 all proteins true version v2.csv"
csv_path7 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\hESCs Rep1 all proteins 2.csv"
csv_path8 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\hESCs Rep1 all proteins.csv"
csv_path9 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\hESCs Rep2 all proteins and CHX.csv"
csv_path10 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\3T3 GAPDH cyto rep1.csv"
csv_path_11 = r"Z:\\EPFL\\Biology\\bachelor\\260417-all-csv-20260417T130751Z-3-001\\260417-all-csv\\3T3 GAPDH cyto rep2.csv"
df_1 = pd.read_csv(csv_path1)
df_2 = pd.read_csv(csv_path2)
df_3 = pd.read_csv(csv_path3)
df_4 = pd.read_csv(csv_path4)
df_5 = pd.read_csv(csv_path5)
df_6 = pd.read_csv(csv_path6)
df_7 = pd.read_csv(csv_path7)
df_8 = pd.read_csv(csv_path8)
df_9 = pd.read_csv(csv_path9)
df_10 = pd.read_csv(csv_path10)
df_11 = pd.read_csv(csv_path_11)

dataset_dfs = [df_1, df_2, df_3, df_4, df_5, df_6, df_7, df_8, df_9, df_10, df_11]

# Filter by r and/or c (set wanted_c* = None to ignore c filtering)
# 3T3s:
# TEAD 3t3:
tead_3t3_id = [1, 2]  # List of dataset indices to keep
r1_3t3_tead = [1, 2]      # df_1: keep r01, r02, ...
c1_3t3_tead = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_3t3_tead = [1, 2]      # df_2
c2_3t3_tead = 2           # can be None, a single int, or a list
# H3_3 3t3:
h3_3t3_id = [2, 3]  # List of dataset indices to keep
r1_3t3_h3_3 = [3, 4]      # df_1: keep r01, r02, ...
c1_3t3_h3_3 = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_3t3_h3_3 = [1, 2]      # df_2
c2_3t3_h3_3 = 1           # can be None, a single int, or a list
# PolII 3t3:
polii_3t3_id = [4, 2]  # List of dataset indices to keep
r1_3t3_polii = [3, 4]      # df_1: keep r01, r02, ...
c1_3t3_polii = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_3t3_polii = [5, 6]      # df_2
c2_3t3_polii = 2           # can be None, a single int, or a list
# BRG1 3t3:
brg1_3t3_id = [4, 2]  # List of dataset indices to keep
r1_3t3_brg1 = [5, 6]     # df_1: keep r01, r02, ...
c1_3t3_brg1 = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_3t3_brg1 = [7, 8]      # df_2
c2_3t3_brg1 = 2           # can be None, a single int, or a list


# mESCs:
nanog_mescs_id = [5, 6]  # List of dataset indices to keep
r1_mescs_nanog = [1, 2]      # df_1: keep r01, r02, ...
c1_mescs_nanog = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_mescs_nanog = [1, 2]      # df_2
c2_mescs_nanog = 1           # can be None, a single int, or a list

# H3_3 mESCs:
h3_mescs_id = [5, 6]  # List of dataset indices to keep
r1_mescs_h3_3 = [1, 2]      # df_1: keep r01, r02, ...
c1_mescs_h3_3 = 2           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_mescs_h3_3 = [1, 2]      # df_2
c2_mescs_h3_3 = 2           # can be None, a single int, or a list
# SOX2 mESCs:
sox2_mescs_id = [5, 6]  # List of dataset indices to keep
r1_mescs_sox2 = [3, 4]      # df_1: keep r01, r02, ...
c1_mescs_sox2 = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_mescs_sox2 = [3, 4]      # df_2
c2_mescs_sox2 = 1           # can be None, a single int, or a list
# PolII mESCs:
polii_mescs_id = [5, 6]  # List of dataset indices to keep
r1_mescs_polii = [5, 6]      # df_1: keep r01, r02, ...
c1_mescs_polii = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_mescs_polii = [5, 6]      # df_2
c2_mescs_polii = 1           # can be None, a single int, or a list
# BRG1 mESCs:
brg1_mescs_id = [5, 6]  # List of dataset indices to keep
r1_mescs_brg1 = [7, 8]     # df_1: keep r01, r02, ...
c1_mescs_brg1 = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_mescs_brg1 = [7, 8]      # df_2
c2_mescs_brg1 = 1          # can be None, a single int, or a list

#hESCs
'''
nanog_hescs_id = [7, 7]  # List of dataset indices to keep
r1_hescs_nanog = [1, 2]      # df_1: keep r01, r02, ...
c1_hescs_nanog = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_hescs_nanog = [1, 2]      # df_2
c2_hescs_nanog = 1           # can be None, a single int, or a list
'''
# H3_3 hESCs:
pol2_hescs_id = [8, 9]  # List of dataset indices to keep
r1_hescs_pol2 = [1, 2]      # df_1: keep r01, r02, ...
c1_hescs_pol2 = 2           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_hescs_pol2 = [7, 8]      # df_2
c2_hescs_pol2 = 1           # can be None, a single int, or a list
# SOX2 hESCs:
sox2_hescs_id = [8, 9]  # List of dataset indices to keep
r1_hescs_sox2 = [3, 4]      # df_1: keep r01, r02, ...
c1_hescs_sox2 = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_hescs_sox2 = [1, 2]      # df_2
c2_hescs_sox2 = 1           # can be None, a single int, or a list
# OCT4 hESCs:
oct4_hescs_id = [8, 9]  # List of dataset indices to keep
r1_hescs_oct4 = [5, 6]      # df_1: keep r01, r02, ...
c1_hescs_oct4 = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_hescs_oct4 = [3, 4]      # df_2
c2_hescs_oct4 = 1           # can be None, a single int, or a list
# H3.3 hESCs:
h3_hescs_id = [8, 9]  # List of dataset indices to keep
r1_hescs_h3 = [7, 8]     # df_1: keep r01, r02, ...
c1_hescs_h3 = 1           # can be None, a single int (e.g. 1), or a list (e.g. [1,3,6])
r2_hescs_h3 = [5, 6]      # df_2
c2_hescs_h3 = 1          # can be None, a single int, or a list

id_list = [tead_3t3_id, h3_3t3_id, polii_3t3_id, brg1_3t3_id,
           nanog_mescs_id, h3_mescs_id, sox2_mescs_id, polii_mescs_id, brg1_mescs_id,
            oct4_hescs_id, h3_hescs_id, sox2_hescs_id, pol2_hescs_id]

wanted_list = [
    (r1_3t3_tead, c1_3t3_tead, r2_3t3_tead, c2_3t3_tead),
    (r1_3t3_h3_3, c1_3t3_h3_3, r2_3t3_h3_3, c2_3t3_h3_3),
    (r1_3t3_polii, c1_3t3_polii, r2_3t3_polii, c2_3t3_polii),
    (r1_3t3_brg1, c1_3t3_brg1, r2_3t3_brg1, c2_3t3_brg1),
    (r1_mescs_nanog, c1_mescs_nanog, r2_mescs_nanog, c2_mescs_nanog),
    (r1_mescs_h3_3, c1_mescs_h3_3, r2_mescs_h3_3, c2_mescs_h3_3),
    (r1_mescs_sox2, c1_mescs_sox2, r2_mescs_sox2, c2_mescs_sox2),
    (r1_mescs_polii, c1_mescs_polii, r2_mescs_polii, c2_mescs_polii),
    (r1_mescs_brg1, c1_mescs_brg1, r2_mescs_brg1, c2_mescs_brg1),
    (r1_hescs_oct4, c1_hescs_oct4, r2_hescs_oct4, c2_hescs_oct4),
    (r1_hescs_h3, c1_hescs_h3, r2_hescs_h3, c2_hescs_h3),
    (r1_hescs_sox2, c1_hescs_sox2, r2_hescs_sox2, c2_hescs_sox2),
    (r1_hescs_pol2, c1_hescs_pol2, r2_hescs_pol2, c2_hescs_pol2)
]

# =========================
# FILTER SETTINGS
# =========================
nonneg_cols = ["mean_IF_DR", "k", "s"]
trim_cols = ["mean_IF_DR", "k", "s"]
trim_q_low, trim_q_high = 0.02, 0.98

# =========================
# HELPER FUNCTIONS
# =========================
def _as_list_or_none(x):
    if x is None:
        return None
    if isinstance(x, (list, tuple, set, np.ndarray, pd.Series)):
        return list(x)
    return [x]

def add_rc_numbers(df, root_col="if_root"):
    root_id = df[root_col].astype(str).map(lambda s: os.path.basename(os.path.normpath(s)))
    extracted = root_id.str.extract(r"^r(\d+)c(\d+)")
    out = df.copy()
    out["r_num"] = extracted[0].astype("Int64")
    out["c_num"] = extracted[1].astype("Int64")
    return out

def drop_any_negative(df, cols):
    cols_present = [c for c in (cols or []) if c in df.columns]
    if not cols_present:
        return df.copy()
    vals = df[cols_present].apply(pd.to_numeric, errors="coerce")
    return df[~(vals.lt(0).any(axis=1))].copy()

def drop_outside_quantiles(df, cols, q_low, q_high):
    cols_present = [c for c in (cols or []) if c in df.columns]
    if not cols_present:
        return df.copy()
    vals = df[cols_present].apply(pd.to_numeric, errors="coerce")
    lows = vals.quantile(q_low)
    highs = vals.quantile(q_high)
    keep = ((vals >= lows) & (vals <= highs) | vals.isna()).all(axis=1)
    return df[keep].copy()

def return_filt_dfs(df_1, df_2, r1, c1, r2, c2):

    df_1x = add_rc_numbers(df_1)
    df_2x = add_rc_numbers(df_2)

    c1 = _as_list_or_none(c1)
    c2 = _as_list_or_none(c2)

    mask1 = df_1x["r_num"].isin(r1)
    if c1 is not None:
        mask1 &= df_1x["c_num"].isin(c1)

    mask2 = df_2x["r_num"].isin(r2)
    if c2 is not None:
        mask2 &= df_2x["c_num"].isin(c2)

    df_1_filt = drop_outside_quantiles(drop_any_negative(df_1x[mask1], nonneg_cols), trim_cols, trim_q_low, trim_q_high)
    df_2_filt = drop_outside_quantiles(drop_any_negative(df_2x[mask2], nonneg_cols), trim_cols, trim_q_low, trim_q_high)

    # split into wells
    df_1_w1 = df_1_filt[df_1_filt["r_num"] == r1[0]].copy()
    df_1_w2 = df_1_filt[df_1_filt["r_num"] == r1[1]].copy()
    df_2_w1 = df_2_filt[df_2_filt["r_num"] == r2[0]].copy()
    df_2_w2 = df_2_filt[df_2_filt["r_num"] == r2[1]].copy()

    return df_1_w1, df_1_w2, df_2_w1, df_2_w2


# =========================
# DATASET MAPPING (CRITICAL FIX)
# =========================
protein_map = [
    ("3T3", "tead1"),
    ("3T3", "h3_3"),
    ("3T3", "polii"),
    ("3T3", "brg1"),
    ("mESCs", "nanog"),
    ("mESCs", "h3_3"),
    ("mESCs", "sox2"),
    ("mESCs", "polii"),
    ("mESCs", "brg1"),
    ("hESCs", "oct4"),
    ("hESCs", "h3_3"),
    ("hESCs", "sox2"),
    ("hESCs", "polii")
]

# =========================
# INITIALIZE CLEAN STRUCTURE
# =========================
def init_empty():
    return {
        "before_threshold": None,
        "high": None,
        "low": None
    }

full_dataset = {
    "3T3": {p: {"rep1": {"well1": init_empty(), "well2": init_empty()},
                "rep2": {"well1": init_empty(), "well2": init_empty()}}
            for p in ["tead1", "h3_3", "polii", "brg1"]},
    "mESCs": {p: {"rep1": {"well1": init_empty(), "well2": init_empty()},
                  "rep2": {"well1": init_empty(), "well2": init_empty()}}
              for p in ["nanog", "h3_3", "sox2", "polii", "brg1"]},
    "hESCs": {p: {"rep1": {"well1": init_empty(), "well2": init_empty()},
                  "rep2": {"well1": init_empty(), "well2": init_empty()}}
              for p in ["oct4","h3_3","sox2","polii"]}
}

# =========================
# MAIN LOOP (FIXED)
# =========================
for i, (cell_line, protein) in enumerate(protein_map):

    df_1_w1, df_1_w2, df_2_w1, df_2_w2 = return_filt_dfs(
        dataset_dfs[id_list[i][0] - 1],
        dataset_dfs[id_list[i][1] - 1],
        wanted_list[i][0],
        wanted_list[i][1],
        wanted_list[i][2],
        wanted_list[i][3]
    )

    # Debug check
    if any(len(df) == 0 for df in [df_1_w1, df_1_w2, df_2_w1, df_2_w2]):
        print(f"⚠️ EMPTY: {cell_line} | {protein}")

    # Assign
    full_dataset[cell_line][protein]["rep1"]["well1"]["before_threshold"] = df_1_w1
    full_dataset[cell_line][protein]["rep1"]["well2"]["before_threshold"] = df_1_w2
    full_dataset[cell_line][protein]["rep2"]["well1"]["before_threshold"] = df_2_w1
    full_dataset[cell_line][protein]["rep2"]["well2"]["before_threshold"] = df_2_w2


# =========================
# FINAL SANITY CHECK
# =========================
for cl in full_dataset:
    for prot in full_dataset[cl]:
        for rep in full_dataset[cl][prot]:
            for well in full_dataset[cl][prot][rep]:
                df = full_dataset[cl][prot][rep][well]["before_threshold"]
                if df is None or len(df) == 0:
                    print("❌ STILL EMPTY:", cl, prot, rep, well)
```

```python
# SPLIT INTO HIGH / LOW (FIXED)
# =========================

def split_k_high_low_quantiles(df, k_col="k", q_low=1/6, q_high=5/6):
    """
    Returns (df_high, df_low, k_q_high, k_q_low)
    """

    if df is None or len(df) == 0:
        empty = pd.DataFrame(columns=df.columns if df is not None else [])
        return empty, empty, np.nan, np.nan

    if k_col not in df.columns:
        empty = df.iloc[0:0].copy()
        return empty, empty, np.nan, np.nan

    k = pd.to_numeric(df[k_col], errors="coerce")
    k_valid = k.dropna()

    if len(k_valid) == 0:
        empty = df.iloc[0:0].copy()
        return empty, empty, np.nan, np.nan

    k_q_low = float(k_valid.quantile(q_low))
    k_q_high = float(k_valid.quantile(q_high))
    print(k_q_low, k_q_high)
    high = df[k >= k_q_high].copy()
    low  = df[k <= k_q_low].copy()

    return high, low, k_q_high, k_q_low


# =========================
# APPLY TO FULL DATASET
# =========================

k_q_low = 1/6
k_q_high = 5/6

for cell_line in full_dataset:
    for protein in full_dataset[cell_line]:
        for rep in full_dataset[cell_line][protein]:
            for well in full_dataset[cell_line][protein][rep]:

                df = full_dataset[cell_line][protein][rep][well]["before_threshold"]

                high, low, q_high_val, q_low_val = split_k_high_low_quantiles(
                    df, q_low=k_q_low, q_high=k_q_high
                )

                full_dataset[cell_line][protein][rep][well]["high"] = high
                full_dataset[cell_line][protein][rep][well]["low"] = low

                # 🔍 DEBUG (very useful)
                if len(high) == 0 or len(low) == 0:
                    print(f"⚠️ EMPTY SPLIT: {cell_line} | {protein} | {rep} | {well}")
                    print(f"   total={len(df)} | high={len(high)} | low={len(low)} | q_low={q_low_val:.3g} | q_high={q_high_val:.3g}")

'''
# =========================
# PRINT LENGTHS OF ALL HIGH/LOW AND BALANCE SAMPLES
# =========================

print("\n" + "=" * 100)
print("BEFORE BALANCING: Lengths of all high/low splits")
print("=" * 100)

all_lengths = {}
for cell_line in full_dataset:
    for protein in full_dataset[cell_line]:
        for rep in full_dataset[cell_line][protein]:
            for well in full_dataset[cell_line][protein][rep]:
                high = full_dataset[cell_line][protein][rep][well]["high"]
                low = full_dataset[cell_line][protein][rep][well]["low"]
                
                key = f"{cell_line} | {protein} | {rep} | {well}"
                all_lengths[key] = {
                    "high": len(high) if high is not None else 0,
                    "low": len(low) if low is not None else 0
                }
                print(f"{key:50s} | high: {len(high) if high is not None else 0:6d} | low: {len(low) if low is not None else 0:6d}")

# Find minimum lengths (excluding zero-length entries)
high_lengths = [v["high"] for v in all_lengths.values() if v["high"] > 0]
low_lengths = [v["low"] for v in all_lengths.values() if v["low"] > 0]

if high_lengths:
    min_high = min(high_lengths)
else:
    min_high = 0

if low_lengths:
    min_low = min(low_lengths)
else:
    min_low = 0

print(f"\nMinimum high length: {min_high}")
print(f"Minimum low length: {min_low}")

# Resample all to minimum lengths
print("\n" + "=" * 100)
print("RESAMPLING to balanced amounts (random selection)...")
print("=" * 100)

for cell_line in full_dataset:
    for protein in full_dataset[cell_line]:
        for rep in full_dataset[cell_line][protein]:
            for well in full_dataset[cell_line][protein][rep]:
                high = full_dataset[cell_line][protein][rep][well]["high"]
                low = full_dataset[cell_line][protein][rep][well]["low"]
                
                if high is not None and len(high) > 0 and min_high > 0:
                    high_balanced = high.sample(n=min_high, random_state=42)
                    full_dataset[cell_line][protein][rep][well]["high"] = high_balanced
                
                if low is not None and len(low) > 0 and min_low > 0:
                    low_balanced = low.sample(n=min_low, random_state=42)
                    full_dataset[cell_line][protein][rep][well]["low"] = low_balanced

print("\n" + "=" * 100)
print("AFTER BALANCING: Lengths of all high/low splits")
print("=" * 100)

for cell_line in full_dataset:
    for protein in full_dataset[cell_line]:
        for rep in full_dataset[cell_line][protein]:
            for well in full_dataset[cell_line][protein][rep]:
                high = full_dataset[cell_line][protein][rep][well]["high"]
                low = full_dataset[cell_line][protein][rep][well]["low"]
                
                key = f"{cell_line} | {protein} | {rep} | {well}"
                print(f"{key:50s} | high: {len(high) if high is not None else 0:6d} | low: {len(low) if low is not None else 0:6d}")
'''
```

```python
# Plot DR variability (std/mean) for min/max groups, for all proteins and both cell lines.
# Uses full_dataset[...] structure created in the earlier cells.

import re


def _as_df(x):
    # before_threshold is stored as a list of DataFrames (because you append); high/low are DataFrames.
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()

def dr_cv_std_over_mean(df, col="mean_IF_DR"):
    if col not in df.columns:
        return np.nan
    v = pd.to_numeric(df[col], errors="coerce").dropna()
    if len(v) == 0:
        return np.nan
    mu = float(v.mean())
    if mu == 0 or np.isnan(mu):
        return np.nan
    sd = float(v.std(ddof=1)) if len(v) >= 2 else 0.0
    return sd / mu

def get_group_df(full_dataset, cell_line, protein, rep, well, group_key):
    node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep, {}).get(well, {})
    return _as_df(node.get(group_key, []))

def plot_dr_variability_for_dataset(cell_line, protein):
    # X layout: MIN first (left), then MAX (right), per rep
    xpos = {
        ("low", "rep1"): 0, ("low", "rep2"): 1,
        ("high", "rep1"): 3, ("high", "rep2"): 4,
    }

    sns.set_style("whitegrid")
    plt.rcParams.update({"font.size": 11})
    fig, ax = plt.subplots(figsize=(9.6, 4.8), dpi=200)

    c_well1, c_well2 = sns.color_palette(n_colors=2)
    c_avg = "black"
    marker_well1, marker_well2, marker_avg = "o", "o", "_"
    s_point = 85
    s_avg = 110

    any_points = False
    for group, group_key in (("low", "low"), ("high", "high")):
        for rep in ("rep1", "rep2"):
            df_w1 = get_group_df(full_dataset, cell_line, protein, rep, "well1", group_key)
            df_w2 = get_group_df(full_dataset, cell_line, protein, rep, "well2", group_key)
            y1 = dr_cv_std_over_mean(df_w1)
            y2 = dr_cv_std_over_mean(df_w2)
            yavg = float(np.nanmean([y1, y2]))
            x = xpos[(group, rep)]

            if not (np.isnan(y1) and np.isnan(y2) and np.isnan(yavg)):
                any_points = True

            ax.scatter(x, y1, color=c_well1, marker=marker_well1, s=s_point, edgecolor="white", linewidth=0.8, zorder=3)
            ax.scatter(x, y2, color=c_well2, marker=marker_well2, s=s_point, edgecolor="white", linewidth=0.8, zorder=3)
            ax.scatter(x, yavg, color=c_avg, marker=marker_avg, s=s_avg, zorder=4)

            if not np.isnan(yavg):
                ax.annotate(f"{yavg:.3g}", (x, yavg), textcoords="offset points", xytext=(0, 9), ha="center", fontsize=9)

    ax.set_ylabel("CV")
    ax.set_xlabel("Group")
    ax.set_xticks([0, 1, 3, 4])
    ax.set_xticklabels(["rep1", "rep2", "rep1", "rep2"])
    ax.axvline(2, color="0.85", linewidth=1)
    ax.text(0.5, -0.14, "low k (5/6 percentile)", transform=ax.get_xaxis_transform(), ha="center", va="top", fontsize=11)
    ax.text(3.5, -0.14, "high k (1/6 percentile)", transform=ax.get_xaxis_transform(), ha="center", va="top", fontsize=11)
    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax.set_xlim(-0.5, 4.5)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    from matplotlib.lines import Line2D
    legend_items = [
        Line2D([0], [0], marker=marker_well1, color="none", label="well1", markerfacecolor=c_well1, markeredgecolor="white", markersize=9),
        Line2D([0], [0], marker=marker_well2, color="none", label="well2", markerfacecolor=c_well2, markeredgecolor="white", markersize=9),
        Line2D([0], [0], marker=marker_avg,   color="none", label="average(well1,well2)", markerfacecolor=c_avg, markersize=9),
    ]
    ax.legend(handles=legend_items, frameon=True, loc="upper right")

    ax.set_title(f"{cell_line} | {protein} variability for low vs high k")
    plt.tight_layout()

    
    # 1. Get title
    title = plt.gca().get_title()

    # 2. Make it filename-safe
    safe_title = re.sub(r'[^\w\-_. ]', '', title).replace(" ", "_")

    # 3. Create folder if it doesn't exist
    folder = "plots"
    os.makedirs(folder, exist_ok=True)

    # 4. Save plot
    filepath = os.path.join(folder, f"{safe_title}.png")
    plt.savefig(filepath)

    if any_points:
        plt.show()
    else:
        plt.close(fig)


# Iterate over the whole dataset
for cell_line in ("3T3", "mESCs", "hESCs"):
    proteins = list(full_dataset.get(cell_line, {}).keys())
    for protein in proteins:
        plot_dr_variability_for_dataset(cell_line, protein)
```

```python
# All in one figure
# Plots DR variability as CV = std(mean_IF_DR) / mean(mean_IF_DR) for low vs high k,
# with points for well1, well2, and their average, for rep1 & rep2.

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
N_COLS = 5  # Set this to the number of columns you want in the grid (e.g. max proteins per cell line)

def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def dr_cv_std_over_mean(df, col="mean_IF_DR"):
    if df is None or df.empty or col not in df.columns:
        return np.nan
    v = pd.to_numeric(df[col], errors="coerce").dropna()
    if len(v) == 0:
        return np.nan
    mu = float(v.mean())
    if not np.isfinite(mu) or mu == 0:
        return np.nan
    sd = float(v.std(ddof=1)) if len(v) >= 2 else 0.0
    return sd / mu


def get_group_df(full_dataset, cell_line, protein, rep, well, group_key):
    node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep, {}).get(well, {})
    return _as_df(node.get(group_key, []))


def plot_dr_variability_groups_ax(
    full_dataset,
    cell_line,
    protein,
    *,
    ax,
    col="mean_IF_DR",
    annotate=False,
):
    xpos = {
        ("low", "rep1"): 0, ("low", "rep2"): 1,
        ("high", "rep1"): 3, ("high", "rep2"): 4,
    }

    c_well1, c_well2 = sns.color_palette(n_colors=2)
    c_avg = "black"
    marker_well1, marker_well2, marker_avg = "o", "o", "_"
    s_point = 55
    s_avg = 75

    any_points = False
    for group, group_key in (("low", "low"), ("high", "high")):
        for rep in ("rep1", "rep2"):
            df_w1 = get_group_df(full_dataset, cell_line, protein, rep, "well1", group_key)
            df_w2 = get_group_df(full_dataset, cell_line, protein, rep, "well2", group_key)

            y1 = dr_cv_std_over_mean(df_w1, col=col)
            y2 = dr_cv_std_over_mean(df_w2, col=col)
            yavg = float(np.nanmean([y1, y2]))
            x = xpos[(group, rep)]

            if not (np.isnan(y1) and np.isnan(y2) and np.isnan(yavg)):
                any_points = True

            ax.scatter(x, y1, color=c_well1, marker=marker_well1, s=s_point,
                       edgecolor="white", linewidth=0.6, zorder=3)
            ax.scatter(x, y2, color=c_well2, marker=marker_well2, s=s_point,
                       edgecolor="white", linewidth=0.6, zorder=3)
            ax.scatter(x, yavg, color=c_avg, marker=marker_avg, s=s_avg, zorder=4)

            if annotate and np.isfinite(yavg):
                ax.annotate(f"{yavg:.2g}", (x, yavg),
                            textcoords="offset points", xytext=(0, 6),
                            ha="center", fontsize=7)

    if not any_points:
        ax.axis("off")
        return False

    ax.set_xlim(-0.5, 4.5)
    ax.set_xticks([0, 1, 3, 4])
    ax.axvline(2, color="0.85", linewidth=1)
    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.set_title(f"{cell_line} | {protein}", fontsize=10)

    return True


# ---- Compose the grid (like your sliding-k example) ----
sns.set_style("whitegrid")

cell_lines = ["3T3", "mESCs", "hESCs"]
proteins_3t3 = list(full_dataset.get("3T3", {}).keys())
proteins_mescs = list(full_dataset.get("mESCs", {}).keys())
proteins_hescs = list(full_dataset.get("hESCs", {}).keys())

# choose number of columns (use N_COLS if you already have it; else use max available)
try:
    ncols = int(N_COLS)
except Exception:
    ncols = max(len(proteins_3t3), len(proteins_mescs), len(proteins_hescs))

fig, axes = plt.subplots(
    nrows=3,
    ncols=ncols,
    figsize=(4.8 * ncols, 9.0),
    dpi=300,
    sharey=False,
    sharex=False,
)

# Track y-limits to make all subplots comparable
all_y = []

def _collect_y(cell_line, protein, col="mean_IF_DR"):
    ys = []
    for group_key in ("low", "high"):
        for rep in ("rep1", "rep2"):
            df_w1 = get_group_df(full_dataset, cell_line, protein, rep, "well1", group_key)
            df_w2 = get_group_df(full_dataset, cell_line, protein, rep, "well2", group_key)
            y1 = dr_cv_std_over_mean(df_w1, col=col)
            y2 = dr_cv_std_over_mean(df_w2, col=col)
            yavg = float(np.nanmean([y1, y2]))
            ys.extend([y1, y2, yavg])
    return [float(y) for y in ys if np.isfinite(y)]

for col_i in range(ncols):
    # Row 0: 3T3
    if col_i < len(proteins_3t3):
        prot = proteins_3t3[col_i]
        all_y += _collect_y("3T3", prot)
    # Row 1: mESCs
    if col_i < len(proteins_mescs):
        prot = proteins_mescs[col_i]
        all_y += _collect_y("mESCs", prot)
    # Row 2: hESCs
    if col_i < len(proteins_hescs):
        prot = proteins_hescs[col_i]
        all_y += _collect_y("hESCs", prot)

if len(all_y) > 0:
    y_max = max(all_y)
    y_pad = 0.08 * y_max if y_max > 0 else 0.1
    y_lim = (0, y_max + y_pad)
else:
    y_lim = None

for col_i in range(ncols):
    # 3T3 row
    if col_i < len(proteins_3t3):
        prot = proteins_3t3[col_i]
        ok = plot_dr_variability_groups_ax(full_dataset, "3T3", prot, ax=axes[0, col_i], annotate=False)
        if ok and y_lim is not None:
            axes[0, col_i].set_ylim(*y_lim)
    else:
        axes[0, col_i].axis("off")

    # mESCs row
    if col_i < len(proteins_mescs):
        prot = proteins_mescs[col_i]
        ok = plot_dr_variability_groups_ax(full_dataset, "mESCs", prot, ax=axes[1, col_i], annotate=False)
        if ok and y_lim is not None:
            axes[1, col_i].set_ylim(*y_lim)
    else:
        axes[1, col_i].axis("off")

    # hESCs row
    if col_i < len(proteins_hescs):
        prot = proteins_hescs[col_i]
        ok = plot_dr_variability_groups_ax(full_dataset, "hESCs", prot, ax=axes[2, col_i], annotate=False)
        if ok and y_lim is not None:
            axes[2, col_i].set_ylim(*y_lim)
    else:
        axes[2, col_i].axis("off")

# Clean up x labels: only show on bottom row
for r in range(3):
    for c in range(ncols):
        ax = axes[r, c]
        if not ax.has_data():
            continue
        ax.set_xticks([0, 1, 3, 4])
        if r == 2:
            ax.set_xticklabels(["rep1", "rep2", "rep1", "rep2"], fontsize=9)
            ax.set_xlabel("Group", fontsize=10)
            ax.text(0.5, -0.22, "low k (5/6 percentile)", transform=ax.get_xaxis_transform(),
                    ha="center", va="top", fontsize=9)
            ax.text(3.5, -0.22, "high k (1/6 percentile)", transform=ax.get_xaxis_transform(),
                    ha="center", va="top", fontsize=9)
        else:
            ax.set_xticklabels([])
            ax.set_xlabel("")

# One legend for the whole figure
from matplotlib.lines import Line2D
c_well1, c_well2 = sns.color_palette(n_colors=2)
legend_items = [
    Line2D([0], [0], marker="o", color="none", label="well1",
           markerfacecolor=c_well1, markeredgecolor="white", markersize=8),
    Line2D([0], [0], marker="o", color="none", label="well2",
           markerfacecolor=c_well2, markeredgecolor="white", markersize=8),
    Line2D([0], [0], marker="_", color="none", label="average(well1,well2)",
           markerfacecolor="black", markersize=10),
]
fig.legend(handles=legend_items, loc="upper right", frameon=True)

fig.suptitle("DR variability (CV=std/mean) | low vs high k", y=1.02, fontsize=14)
plt.tight_layout()
plt.show()

# Optional save (uses your SAVE_* vars if they exist)
save_plots = globals().get("SAVE_PLOTS", True)
if save_plots:
    folder = globals().get("SAVE_FOLDER", "plots")
    dpi = globals().get("SAVE_DPI", 300)
    os.makedirs(folder, exist_ok=True)
    title = "DR variability (CV=std_mean) grid"
    safe_title = re.sub(r"[^\w\-_. ]", "", title).replace(" ", "_")
    out_path = os.path.join(folder, f"{safe_title}.png")
    fig.savefig(out_path, dpi=dpi, bbox_inches="tight")
    print(f"Saved: {out_path}")
```

```python
# One figure per cell type (3T3: 3x2, mESCs: 3x2, hESCs: 2x2)
# X per replicate uses 3 slots: low, (middle slot for Δ bracket), high
# Rep1 uses left y-axis; Rep2 uses right y-axis (independent scales)
# Dots (well1, well2, avg) are aligned at the SAME x-position
# All subplots: primary (left) y-axis labeled
# Remove "k group" label
# Put bold "rep1"/"rep2" labels at the TOP (inside axes), slightly below title
# Slightly larger protein title font + more space between subplots

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def get_group_df(full_dataset, cell_line, protein, rep, well, group_key):
    node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep, {}).get(well, {})
    return _as_df(node.get(group_key, []))


def dr_cv_std_over_mean(df, col="mean_IF_DR"):
    if df is None or df.empty or col not in df.columns:
        return np.nan
    v = pd.to_numeric(df[col], errors="coerce").dropna()
    if len(v) == 0:
        return np.nan
    mu = float(v.mean())
    if not np.isfinite(mu) or mu == 0:
        return np.nan
    sd = float(v.std(ddof=1)) if len(v) >= 2 else 0.0
    return sd / mu


def _ylim_from_values(vals, pad_frac=0.12):
    vals = [float(v) for v in vals if np.isfinite(v)]
    if len(vals) == 0:
        return None
    y_min = min(vals)
    y_max = max(vals)
    rng = y_max - y_min
    pad = pad_frac * rng if rng > 0 else pad_frac * (abs(y_max) if y_max != 0 else 1.0)
    return (y_min - pad, y_max + pad)


def plot_dr_variability_groups_ax(
    full_dataset,
    cell_line,
    protein,
    *,
    ax,
    col="mean_IF_DR",
    annotate_delta=True,
    xgap=4.0,            # rep2 block starts at xgap (rep1 block starts at 0)
    slot=1.0,            # spacing between low / mid / high slots
    cap_halfwidth=0.18,  # cap size for Δ bracket
    title_fs=12,
    rep_label_fs=10,
):
    ax_r = ax.twinx()

    rep_base = {"rep1": 0.0, "rep2": float(xgap)}

    def x_low(rep):  return rep_base[rep] + 0.0 * slot
    def x_mid(rep):  return rep_base[rep] + 1.0 * slot
    def x_high(rep): return rep_base[rep] + 2.0 * slot

    # styling
    c_well1, c_well2 = sns.color_palette(n_colors=2)
    c_avg = "black"
    marker_well1, marker_well2, marker_avg = "o", "o", "_"
    s_point = 55
    s_avg = 75

    any_points = False
    rep_vals = {"rep1": [], "rep2": []}
    rep_avg = {"rep1": {"low": np.nan, "high": np.nan}, "rep2": {"low": np.nan, "high": np.nan}}

    for rep in ("rep1", "rep2"):
        rep_axis = ax if rep == "rep1" else ax_r

        for group, group_key in (("low", "low"), ("high", "high")):
            df_w1 = get_group_df(full_dataset, cell_line, protein, rep, "well1", group_key)
            df_w2 = get_group_df(full_dataset, cell_line, protein, rep, "well2", group_key)

            y1 = dr_cv_std_over_mean(df_w1, col=col)
            y2 = dr_cv_std_over_mean(df_w2, col=col)
            yavg = float(np.nanmean([y1, y2]))
            rep_avg[rep][group] = yavg

            xg = x_low(rep) if group == "low" else x_high(rep)

            if not (np.isnan(y1) and np.isnan(y2) and np.isnan(yavg)):
                any_points = True

            # ALL aligned on the same x position:
            rep_axis.scatter(
                xg, y1,
                color=c_well1, marker=marker_well1, s=s_point,
                edgecolor="white", linewidth=0.6, zorder=3
            )
            rep_axis.scatter(
                xg, y2,
                color=c_well2, marker=marker_well2, s=s_point,
                edgecolor="white", linewidth=0.6, zorder=3
            )
            rep_axis.scatter(xg, yavg, color=c_avg, marker=marker_avg, s=s_avg, zorder=4)

            for y in (y1, y2, yavg):
                if np.isfinite(y):
                    rep_vals[rep].append(float(y))

    if not any_points:
        ax.axis("off")
        ax_r.axis("off")
        return False

    # Δ bracket at the middle slot
    if annotate_delta:
        for rep in ("rep1", "rep2"):
            rep_axis = ax if rep == "rep1" else ax_r
            y_low = rep_avg[rep]["low"]
            y_high = rep_avg[rep]["high"]

            if np.isfinite(y_low) and np.isfinite(y_high):
                delta = float(y_high - y_low)
                xm = x_mid(rep)
                ym = 0.5 * (y_low + y_high)

                rep_axis.plot([xm, xm], [y_low, y_high], ls="--", lw=1.1, color="0.25", zorder=2)
                rep_axis.plot([xm - cap_halfwidth, xm + cap_halfwidth], [y_low, y_low],
                              lw=1.1, color="0.25", zorder=2)
                rep_axis.plot([xm - cap_halfwidth, xm + cap_halfwidth], [y_high, y_high],
                              lw=1.1, color="0.25", zorder=2)

                rep_axis.text(
                    xm, ym, f"{delta:.3g}",
                    ha="center", va="center", fontsize=8,
                    bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85),
                    zorder=6,
                )

    # independent y-lims (NOT forced to start at 0)
    y1_lim = _ylim_from_values(rep_vals["rep1"])
    y2_lim = _ylim_from_values(rep_vals["rep2"])
    if y1_lim is not None:
        ax.set_ylim(*y1_lim)
    if y2_lim is not None:
        ax_r.set_ylim(*y2_lim)

    # x formatting
    ticks = [
        x_low("rep1"), x_mid("rep1"), x_high("rep1"),
        x_low("rep2"), x_mid("rep2"), x_high("rep2"),
    ]
    ax.set_xticks(ticks)

    # separator between replicate blocks
    sep = 0.5 * (x_high("rep1") + x_low("rep2"))
    ax.axvline(sep, color="0.85", linewidth=1)

    ax.set_xlim(x_low("rep1") - 0.7, x_high("rep2") + 0.7)

    # protein title + rep labels near top (inside axes, below title)
    ax.set_title(str(protein), fontsize=title_fs, pad=18)

    y_rep = 1.02  # in axes coords (0..1), inside plot near top
    ax.text(
        x_mid("rep1"), y_rep, "rep1",
        transform=ax.get_xaxis_transform(),
        ha="center", va="top",
        fontsize=rep_label_fs, fontweight="bold"
    )
    ax.text(
        x_mid("rep2"), y_rep, "rep2",
        transform=ax.get_xaxis_transform(),
        ha="center", va="top",
        fontsize=rep_label_fs, fontweight="bold"
    )

    # cosmetics
    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax_r.grid(False)
    ax.spines["top"].set_visible(False)
    ax_r.spines["top"].set_visible(False)

    return True


def plot_one_celltype_figure(
    full_dataset,
    cell_line,
    *,
    nrows,
    ncols,
    col="mean_IF_DR",
    annotate_delta=True,
    xgap=4.0,
    save=True,
    show=True,
    out_folder=None,
    dpi=300,
):
    sns.set_style("whitegrid")

    proteins = list(full_dataset.get(cell_line, {}).keys())
    nslots = nrows * ncols
    proteins_to_plot = proteins[:nslots]

    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(6.6 * ncols, 3.6 * nrows),
        dpi=dpi,
        sharex=False,
        sharey=False,
        squeeze=False,
    )

    axes_flat = axes.ravel()
    for i, ax in enumerate(axes_flat):
        if i < len(proteins_to_plot):
            prot = proteins_to_plot[i]
            plot_dr_variability_groups_ax(
                full_dataset, cell_line, prot,
                ax=ax, col=col, annotate_delta=annotate_delta, xgap=xgap,
                title_fs=12, rep_label_fs=10
            )
        else:
            ax.axis("off")

    # X labels: only bottom row; NO "k group" xlabel anywhere
    # Also: label the PRIMARY (left) y-axis for every subplot
    # Replace ONLY this part in plot_one_celltype_figure():
    # (the loop that currently says "# X labels: only bottom row ...")

    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            if not ax.axison:
                continue

            ax.set_ylabel("CV")  # primary axis labeled on every subplot

            # show low/high on EVERY subplot (not only bottom row)
            ax.set_xticks([0, 1, 2, xgap + 0, xgap + 1, xgap + 2])
            ax.set_xticklabels(["low", "", "high", "low", "", "high"], fontsize=9)

            ax.set_xlabel("")  # remove "k group"

    # One legend for the whole figure
    from matplotlib.lines import Line2D
    c_well1, c_well2 = sns.color_palette(n_colors=2)
    legend_items = [
        Line2D([0], [0], marker="o", color="none", label="well1",
               markerfacecolor=c_well1, markeredgecolor="white", markersize=8),
        Line2D([0], [0], marker="o", color="none", label="well2",
               markerfacecolor=c_well2, markeredgecolor="white", markersize=8),
        Line2D([0], [0], marker="_", color="none", label="average(well1,well2)",
               markerfacecolor="black", markersize=10),
    ]
    fig.legend(handles=legend_items, loc="upper right", frameon=True)

    fig.suptitle(f"{cell_line} | Protein variability for low vs high degradation rate (k)", y=0.995, fontsize=16, fontweight="bold")

    # more space between subplots
    fig.tight_layout(rect=[0, 0, 1, 0.97], pad=1, w_pad=1, h_pad=1)
    fig.subplots_adjust(wspace=0.25, hspace=0.35)

    if save:
        if out_folder is None:
            out_folder = globals().get("SAVE_FOLDER", "plots")
        os.makedirs(out_folder, exist_ok=True)
        safe = re.sub(r"[^\w\-_. ]", "", f"{cell_line}_DR_variability_low_vs_high_k_withDelta_dualY").replace(" ", "_")
        out_path = os.path.join(out_folder, f"{safe}.png")
        fig.savefig(out_path, dpi=dpi, bbox_inches="tight")
        print(f"Saved: {out_path}")

    if show:
        plt.show()
    else:
        plt.close(fig)

    return fig, axes


# --- Generate one figure per cell type (requested layouts) ---
plot_one_celltype_figure(full_dataset, "3T3",  nrows=3, ncols=2, annotate_delta=True, xgap=4.0, save=True, show=True)
plot_one_celltype_figure(full_dataset, "mESCs", nrows=3, ncols=2, annotate_delta=True, xgap=4.0, save=True, show=True)
plot_one_celltype_figure(full_dataset, "hESCs", nrows=2, ncols=2, annotate_delta=True, xgap=4.0, save=True, show=True)
```

```python
# TAIL GROUPS: CV(low k-tail) and CV(high k-tail) vs percentile threshold x
#
# Definition for each x (percentile):
#   high group = all points with k >= percentile_x(k)
#   low group  = all points with k <= percentile_(100-x)(k)
# (symmetric tails; variable #cells per group)
#
# For each protein subplot:
# - compute CV(mean_IF_DR) in low and high groups
# - separately for rep1 and rep2
# - within each rep, average the CV between well1 and well2 (only if both wells valid)
#
# One figure per cell type, same layouts as before:
# - 3T3: 2x3
# - mESCs: 2x3
# - hESCs: 2x2
# Legend once per figure:
# - for 2x3: put legend in the 6th (empty) subplot if available
# - otherwise: figure-level legend (upper right)
#
# Low/high are encoded as LIGHT/DARK variants (no dashed lines).

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
from matplotlib.colors import to_rgb


# -----------------------
# Controls
# -----------------------
K_COL = "k"
Y_COL = "mean_IF_DR"

# Recommended range (tails become noisy near 100)
PCT_MIN = 50           # start x (high percentile)
PCT_MAX = 95           # end x
PCT_STEP = 2           # step size in percentile units

MIN_GROUP_N = 50       # require at least this many cells in EACH WELL for a tail CV to be valid

SAVE = True
SAVE_DPI = 300
SAVE_FOLDER = r"DR variability continuous plots\DR_vs_K_tailPercentile"


# -----------------------
# Color helpers
# -----------------------
def _tint(color, t=0.55):
    """Blend toward white; t in [0..1]. Higher = lighter."""
    r, g, b = to_rgb(color)
    return (r + (1 - r) * t, g + (1 - g) * t, b + (1 - b) * t)

def _shade(color, s=0.12):
    """Blend toward black; s in [0..1]. Higher = darker."""
    r, g, b = to_rgb(color)
    return (r * (1 - s), g * (1 - s), b * (1 - s))


# -----------------------
# Dataset helpers (matches your full_dataset structure)
# -----------------------
def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()

def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []
    reps = (rep,) if rep is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")

    for rep_ in reps:
        for well_ in wells:
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep_, {}).get(well_, {})
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


# -----------------------
# Stats
# -----------------------
def _cv(arr):
    arr = np.asarray(arr, dtype=float)
    arr = arr[np.isfinite(arr)]
    if arr.size < 2:
        return np.nan
    mu = float(np.mean(arr))
    if not np.isfinite(mu) or mu == 0:
        return np.nan
    sd = float(np.std(arr, ddof=1))
    return sd / mu

def _avg_if_both(a, b):
    return 0.5 * (a + b) if (np.isfinite(a) and np.isfinite(b)) else np.nan


def _tail_cv_vs_percentile_for_rep(
    df_w1,
    df_w2,
    *,
    k_col=K_COL,
    y_col=Y_COL,
    percentiles=None,
    min_group_n=MIN_GROUP_N,
):
    """
    Returns DataFrame columns: x_percentile, cv_low, cv_high
    for ONE rep (wells averaged).
    Thresholds are defined using pooled k across the two wells (within the rep).
    """
    if percentiles is None:
        percentiles = np.arange(PCT_MIN, PCT_MAX + 0.5, PCT_STEP, dtype=float)

    if df_w1 is None:
        df_w1 = pd.DataFrame()
    if df_w2 is None:
        df_w2 = pd.DataFrame()

    # pooled k to define thresholds
    k_pool = []
    for df in (df_w1, df_w2):
        if (df is not None) and (not df.empty) and (k_col in df.columns):
            k_pool.append(pd.to_numeric(df[k_col], errors="coerce"))

    if not k_pool:
        return pd.DataFrame(columns=["x_percentile", "cv_low", "cv_high"])

    k_pool = pd.concat(k_pool, ignore_index=True).dropna().to_numpy(dtype=float)
    if k_pool.size < 2:
        return pd.DataFrame(columns=["x_percentile", "cv_low", "cv_high"])

    def _group_cv(df, *, q_low, q_high, which):
        if df is None or df.empty:
            return np.nan
        if not ({k_col, y_col} <= set(df.columns)):
            return np.nan

        k = pd.to_numeric(df[k_col], errors="coerce").to_numpy(dtype=float)
        y = pd.to_numeric(df[y_col], errors="coerce").to_numpy(dtype=float)
        m = np.isfinite(k) & np.isfinite(y)
        k = k[m]
        y = y[m]
        if y.size == 0:
            return np.nan

        if which == "low":
            y = y[k <= q_low]
        else:  # "high"
            y = y[k >= q_high]

        if y.size < int(min_group_n):
            return np.nan
        return _cv(y)

    rows = []
    for x in percentiles:
        x = float(x)
        if not (0.0 <= x <= 100.0):
            continue

        q_high = float(np.percentile(k_pool, x))
        q_low = float(np.percentile(k_pool, 100.0 - x))

        low1 = _group_cv(df_w1, q_low=q_low, q_high=q_high, which="low")
        low2 = _group_cv(df_w2, q_low=q_low, q_high=q_high, which="low")
        high1 = _group_cv(df_w1, q_low=q_low, q_high=q_high, which="high")
        high2 = _group_cv(df_w2, q_low=q_low, q_high=q_high, which="high")

        rows.append((x, _avg_if_both(low1, low2), _avg_if_both(high1, high2)))

    return pd.DataFrame(rows, columns=["x_percentile", "cv_low", "cv_high"])


# -----------------------
# Plotting (per protein axis)
# -----------------------
def plot_tail_percentile_ax(
    full_dataset,
    cell_line,
    protein,
    *,
    ax,
    rep_colors,          # expects nested dict: rep_colors[rep]["low"/"high"]
    percentiles=None,
    min_group_n=MIN_GROUP_N,
):
    if percentiles is None:
        percentiles = np.arange(PCT_MIN, PCT_MAX + 0.5, PCT_STEP, dtype=float)

    any_plotted = False

    for rep in ("rep1", "rep2"):
        df_w1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
        df_w2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")

        curve = _tail_cv_vs_percentile_for_rep(
            df_w1,
            df_w2,
            percentiles=percentiles,
            min_group_n=min_group_n,
        )
        if curve.empty:
            continue

        x = curve["x_percentile"].to_numpy(dtype=float)
        y_low = curve["cv_low"].to_numpy(dtype=float)
        y_high = curve["cv_high"].to_numpy(dtype=float)

        if np.isfinite(y_low).any():
            ax.plot(x, y_low, color=rep_colors[rep]["low"], lw=2.2, alpha=0.95)
            any_plotted = True

        if np.isfinite(y_high).any():
            ax.plot(x, y_high, color=rep_colors[rep]["high"], lw=2.2, alpha=0.95)
            any_plotted = True

    if not any_plotted:
        ax.axis("off")
        return

    # LOW curve (make solid)
    ax.plot(
        x, y_low,
        color=rep_colors[rep]["low"],
        lw=2.2,
        alpha=0.95,
        ls="-",          # or just delete this line
    )

    # HIGH curve (make solid)
    ax.plot(
        x, y_high,
        color=rep_colors[rep]["high"],
        lw=2.2,
        alpha=0.95,
        ls="-",          # or just delete this line
    )

    ax.set_title(f"{protein}", fontsize=12, pad=6)
    ax.set_xlim(float(np.min(percentiles)), float(np.max(percentiles)))
    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)


# -----------------------
# One figure per cell type (same composition as before)
# -----------------------
def plot_celltype_tail_percentile_fig(
    full_dataset,
    cell_line,
    *,
    nrows,
    ncols,
    save=SAVE,
    percentiles=None,
    min_group_n=MIN_GROUP_N,
):
    proteins = list(full_dataset.get(cell_line, {}).keys())
    nslots = nrows * ncols
    proteins_to_plot = proteins[:nslots]

    if percentiles is None:
        percentiles = np.arange(PCT_MIN, PCT_MAX + 0.5, PCT_STEP, dtype=float)

    sns.set_style("whitegrid")
    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(4.7 * ncols, 3.5 * nrows),
        dpi=SAVE_DPI,
        sharex=True,
        sharey=False,
        squeeze=False,
    )

    # Rep colors (base) + light/dark variants for low/high
    pal = sns.color_palette("colorblind", 2)
    rep_colors = {
        "rep1": {"high": _shade(pal[0], 0.12), "low": _tint(pal[0], 0.55)},
        "rep2": {"high": _shade(pal[1], 0.12), "low": _tint(pal[1], 0.55)},
    }

    for i, ax in enumerate(axes.ravel()):
        if i < len(proteins_to_plot):
            prot = proteins_to_plot[i]
            plot_tail_percentile_ax(
                full_dataset,
                cell_line,
                prot,
                ax=ax,
                rep_colors=rep_colors,
                percentiles=percentiles,
                min_group_n=min_group_n,
            )
        else:
            ax.axis("off")

    # axis labels (like before: only bottom row x-label, only first col y-label)
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            if not ax.axison:
                continue

            if r == nrows - 1:
                ax.set_xlabel("x percentile  (high: k ≥ x, low: k ≤ 100−x)")
            else:
                ax.set_xlabel("")

            if c == 0:
                ax.set_ylabel("CV(mean_IF_DR)")
            else:
                ax.set_ylabel("")

    # Legend once per figure (no dashed lines)
    legend_handles = [
        Line2D([0], [0], color=rep_colors["rep1"]["low"],  lw=2.2, label="rep1 low (k ≤ 100−x)"),
        Line2D([0], [0], color=rep_colors["rep1"]["high"], lw=2.2, label="rep1 high (k ≥ x)"),
        Line2D([0], [0], color=rep_colors["rep2"]["low"],  lw=2.2, label="rep2 low (k ≤ 100−x)"),
        Line2D([0], [0], color=rep_colors["rep2"]["high"], lw=2.2, label="rep2 high (k ≥ x)"),
    ]

    # If we truly have an extra slot, use it for the legend
    if len(proteins_to_plot) < nslots:
        ax_leg = axes.ravel()[len(proteins_to_plot)]
        ax_leg.axis("off")
        ax_leg.legend(
            handles=legend_handles,
            loc="center",
            frameon=True,
            fontsize=9,
            title="Tail CV vs percentile",
            title_fontsize=10,
        )
    else:
        fig.legend(
            handles=legend_handles,
            loc="upper right",
            frameon=True,
            fontsize=9,
            title="Tail CV vs percentile",
            title_fontsize=10,
        )

    fig.suptitle(
        f"{cell_line} | CV vs percentile",
        y=0.995,
        fontsize=14,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.96], pad=1.0, w_pad=0.8, h_pad=0.9)
    fig.subplots_adjust(wspace=0.25, hspace=0.35)

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out_path = os.path.join(SAVE_FOLDER, f"{cell_line}_tailCV_vs_percentile_avgWells.png")
        fig.savefig(out_path, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out_path}")

    plt.show()


# -----------------------
# Run: one figure per cell type with requested layouts
# -----------------------
plot_celltype_tail_percentile_fig(full_dataset, "3T3",  nrows=2, ncols=3, save=True)
plot_celltype_tail_percentile_fig(full_dataset, "mESCs", nrows=2, ncols=3, save=True)
plot_celltype_tail_percentile_fig(full_dataset, "hESCs", nrows=2, ncols=2, save=True)
```

```python
# k vs s linear fit, for all proteins and both cell lines (uses before_threshold data).
import re
def _concat_before_threshold(full_dataset, cell_line, protein):
    parts = []
    for rep in ("rep1", "rep2"):
        for well in ("well1", "well2"):
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep, {}).get(well, {})
            x = node.get("before_threshold", [])
            if isinstance(x, list):
                parts.extend([d for d in x if isinstance(d, pd.DataFrame)])
            elif isinstance(x, pd.DataFrame):
                parts.append(x)
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()

def plot_k_vs_s_fit(df, title):
    if not ({"s", "k"} <= set(df.columns)):
        return
    x = pd.to_numeric(df["s"], errors="coerce")
    y = pd.to_numeric(df["k"], errors="coerce")
    mask = x.notna() & y.notna()
    x = x[mask].to_numpy()
    y = y[mask].to_numpy()
    if len(x) < 2:
        return
    m, b = np.polyfit(x, y, 1)
    y_hat = m * x + b
    ss_res = np.sum((y - y_hat) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)
    r2 = 1 - ss_res / ss_tot if ss_tot != 0 else np.nan

    plt.figure(figsize=(6.5, 4.5), dpi=200)
    plt.scatter(x, y, s=8, alpha=0.7)
    xs = np.linspace(x.min(), x.max(), 200)
    plt.plot(xs, m * xs + b, linewidth=2, color="tab:orange",
             label=f"fit: k={m:.3g}s+{b:.3g}, $R^2$={r2:.3f}")
    plt.xlabel("s")
    plt.ylabel("k")
    plt.title(title)
    plt.legend()
    plt.tight_layout()

        # 1. Get title
    title = plt.gca().get_title()

    # 2. Make it filename-safe
    safe_title = re.sub(r'[^\w\-_. ]', '', title).replace(" ", "_")

    # 3. Create folder if it doesn't exist
    folder = "k vs s plots"
    os.makedirs(folder, exist_ok=True)

    # 4. Save plot
    filepath = os.path.join(folder, f"{safe_title}.png")
    plt.savefig(filepath)

    plt.show()

for cell_line in ("3T3", "mESCs", "hESCs"):
    for protein in full_dataset.get(cell_line, {}).keys():
        df = _concat_before_threshold(full_dataset, cell_line, protein)
        plot_k_vs_s_fit(df, title=f"{cell_line} | {protein} — k vs s")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# helper: concatenate per cell line
# ----------------------------
def concat_cell_line(full_dataset, cell_line):
    parts = []
    for protein in full_dataset[cell_line].keys():
        for rep in ("rep1", "rep2"):
            for well in ("well1", "well2"):
                df = full_dataset[cell_line][protein][rep][well]["before_threshold"]
                if isinstance(df, pd.DataFrame) and len(df) > 0:
                    tmp = df[["s", "k"]].copy()
                    tmp["protein"] = protein
                    parts.append(tmp)
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


# ----------------------------
# helper: trim extremes globally
# ----------------------------
def trim_outliers(df, cols=("s", "k"), q_low=0.02, q_high=0.98):
    df = df.copy()
    for c in cols:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    mask = np.ones(len(df), dtype=bool)
    for c in cols:
        low, high = df[c].quantile(q_low), df[c].quantile(q_high)
        mask &= df[c].between(low, high)

    return df[mask]


# ----------------------------
# build datasets
# ----------------------------
cell_lines = ["3T3", "mESCs", "hESCs"]

data = {}
for cl in cell_lines:
    df = concat_cell_line(full_dataset, cl)
    df = df.dropna(subset=["s", "k"])
    df = trim_outliers(df)
    data[cl] = df


# ----------------------------
# plotting
# ----------------------------
plt.figure(figsize=(7, 5), dpi=200)

alpha = 0.25
point_size = 6

# plot sequentially (important for visual layering)
for cl in ["3T3", "mESCs", "hESCs"]:
    df = data[cl]
    plt.scatter(df["s"], df["k"], s=point_size, alpha=alpha, label=cl)

plt.xlabel("s")
plt.ylabel("k")
plt.title("k vs s (all cell types combined)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
from matplotlib.colors import LinearSegmentedColormap

# --- proper black-to-viridis colormap ---
cmap = LinearSegmentedColormap.from_list(
    "black_viridis",
    ["black", plt.cm.viridis(0.3), plt.cm.viridis(0.6), plt.cm.viridis(1.0)]
)

fig, axes = plt.subplots(
    1, 3,
    figsize=(12, 4),
    dpi=200,
    facecolor="black"
)

for ax, cl in zip(axes, ["3T3", "mESCs", "hESCs"]):

    ax.set_facecolor("black")

    x = data[cl]["s"].values
    y = data[cl]["k"].values

    kde = gaussian_kde(np.vstack([x, y]), bw_method=1.0)

    xi, yi = np.mgrid[
        x.min():x.max():300j,
        y.min():y.max():300j
    ]

    zi = kde(np.vstack([xi.ravel(), yi.ravel()])).reshape(xi.shape)
    zi = zi / zi.max()

    # 🚨 THIS is the real fix for the purple halo
    zi = np.clip(zi, 0.03, 1.0)   # hard cutoff instead of "soft cmap tricks"

    im = ax.imshow(
        zi.T,
        origin="lower",
        extent=[x.min(), x.max(), y.min(), y.max()],
        cmap=cmap,
        interpolation="bilinear",
        aspect="auto"
    )

    ax.set_title(cl, color="white")
    ax.set_xlabel("s", color="white")
    ax.set_ylabel("k", color="white")

    ax.tick_params(colors="white")
    for spine in ax.spines.values():
        spine.set_color("white")

plt.tight_layout()
plt.show()
```

```python
# All-pairs scatter + linear fit V1
# (uses before_threshold data). Rows = cell type, Cols = proteins *of that cell type*
# i.e. each row can have a different protein list (like your sliding plots).

import itertools

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


def _concat_before_threshold_with_rep(full_dataset, cell_line, protein, *, rep_col="_rep", well_col="_well"):
    parts = []
    for rep in ("rep1", "rep2"):
        for well in ("well1", "well2"):
            node = (
                full_dataset.get(cell_line, {})
                .get(protein, {})
                .get(rep, {})
                .get(well, {})
            )
            x = node.get("before_threshold", [])
            if isinstance(x, list):
                dfs = [d for d in x if isinstance(d, pd.DataFrame)]
            elif isinstance(x, pd.DataFrame):
                dfs = [x]
            else:
                dfs = []

            for d in dfs:
                tmp = d.copy()
                tmp[rep_col] = rep
                tmp[well_col] = well
                parts.append(tmp)

    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib.ticker import MaxNLocator
from matplotlib.colors import LinearSegmentedColormap

# Optional global style (usually nicer than default)
sns.set_theme(style="ticks", context="notebook")

def set_clean_plot_style():
    # Clean, minimal style (call once)
    sns.set_theme(style="ticks", context="notebook")
    # Optional: slightly nicer defaults
    import matplotlib as mpl
    mpl.rcParams.update({
        "axes.titlesize": 9,
        "axes.labelsize": 9,
        "xtick.labelsize": 8,
        "ytick.labelsize": 8,
        "axes.linewidth": 0.8,
    })


def plot_x_vs_y_fit_by_rep_clean(
    ax,
    df,
    xcol,
    ycol,
    *,
    rep_col="_rep",
    palette=None,              # {"rep1": color, "rep2": color}
    alpha=0.65,                # applies to marker edges
    s=14,                      # marker area
    edge_lw=0.6,               # marker edge linewidth
    downsample=2500,           # per rep; None to disable
    fit_lw=2.4,
    fit_alpha=0.95,
    min_points=3,
    clip_quantiles=None,       # e.g. (0.01, 0.99) if you want nicer limits
    show_legend=False,
):
    if df is None or df.empty or not ({xcol, ycol} <= set(df.columns)) or rep_col not in df.columns:
        ax.set_axis_off()
        return

    if palette is None:
        c1, c2 = sns.color_palette("colorblind", 2)
        palette = {"rep1": c1, "rep2": c2}

    # Clean axes
    ax.set_facecolor("white")
    sns.despine(ax=ax, top=True, right=True)
    ax.grid(True, which="major", linewidth=0.8, alpha=0.15)
    ax.set_axisbelow(True)

    # Optional axis clipping (helps when a few outliers stretch everything)
    if clip_quantiles is not None:
        x_all = pd.to_numeric(df[xcol], errors="coerce")
        y_all = pd.to_numeric(df[ycol], errors="coerce")
        ok = x_all.notna() & y_all.notna()
        if ok.any():
            qlo, qhi = clip_quantiles
            xlo, xhi = np.nanquantile(x_all[ok].to_numpy(), [qlo, qhi])
            ylo, yhi = np.nanquantile(y_all[ok].to_numpy(), [qlo, qhi])
            if np.isfinite(xlo) and np.isfinite(xhi) and xlo < xhi:
                ax.set_xlim(xlo, xhi)
            if np.isfinite(ylo) and np.isfinite(yhi) and ylo < yhi:
                ax.set_ylim(ylo, yhi)

    rng = np.random.default_rng(0)

    for rep in ("rep1", "rep2"):
        sub = df[df[rep_col] == rep]
        if sub.empty:
            continue

        x = pd.to_numeric(sub[xcol], errors="coerce")
        y = pd.to_numeric(sub[ycol], errors="coerce")
        mask = x.notna() & y.notna()
        x = x[mask].to_numpy()
        y = y[mask].to_numpy()
        if x.size == 0:
            continue

        # Downsample for display (keeps the plot readable)
        if downsample is not None and x.size > downsample:
            idx = rng.choice(x.size, size=downsample, replace=False)
            xs, ys = x[idx], y[idx]
        else:
            xs, ys = x, y

        color = palette.get(rep)

        # Key: hollow markers => overlap doesn’t make dark ugly blobs
        ax.scatter(
            xs, ys,
            s=s,
            facecolors="none",
            edgecolors=color,
            linewidths=edge_lw,
            alpha=alpha,
            label=rep,
            rasterized=True,
        )

        # Fit line (computed on all points)
        if x.size >= min_points and not np.all(x == x[0]):
            m, b = np.polyfit(x, y, 1)
            xline = np.linspace(np.nanmin(x), np.nanmax(x), 200)
            ax.plot(
                xline, m * xline + b,
                color=color,
                linewidth=fit_lw,
                alpha=fit_alpha,
                solid_capstyle="round",
            )

    if show_legend:
        ax.legend(frameon=False, fontsize=7, loc="best")

def _concat_before_threshold(full_dataset, cell_line, protein):
    parts = []
    for rep in ("rep1", "rep2"):
        for well in ("well1", "well2"):
            node = (
                full_dataset.get(cell_line, {})
                .get(protein, {})
                .get(rep, {})
                .get(well, {})
            )
            x = node.get("before_threshold", [])
            if isinstance(x, list):
                parts.extend([d for d in x if isinstance(d, pd.DataFrame)])
            elif isinstance(x, pd.DataFrame):
                parts.append(x)
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def plot_x_vs_y_fit(ax, df, xcol, ycol):
    """Scatter plot (xcol vs ycol) with simple linear fit, drawn into ax."""
    if df is None or df.empty or not ({xcol, ycol} <= set(df.columns)):
        ax.set_axis_off()
        return

    x = pd.to_numeric(df[xcol], errors="coerce")
    y = pd.to_numeric(df[ycol], errors="coerce")
    mask = x.notna() & y.notna()
    x = x[mask].to_numpy()
    y = y[mask].to_numpy()

    if x.size == 0:
        ax.set_axis_off()
        return

    ax.scatter(x, y, s=8, alpha=0.7)

    # Fit only if we have enough non-degenerate points
    if x.size < 2 or np.all(x == x[0]):
        return

    m, b = np.polyfit(x, y, 1)
    xs = np.linspace(x.min(), x.max(), 200)
    ax.plot(xs, m * xs + b, linewidth=2, color="tab:orange")

    y_hat = m * x + b
    ss_res = np.sum((y - y_hat) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)
    r2 = 1 - ss_res / ss_tot if ss_tot != 0 else np.nan

    ax.text(
        0.03,
        0.97,
        f"m={m:.3g}\nR²={r2:.2f}",
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=7,
    )


# --- Controls ---
CELL_LINES = ["3T3", "mESCs", "hESCs"]

# Keep 5 columns per figure (like the sliding plots).
PROTEINS_PER_FIG = 5

# Variables to pairwise-scatter
'''
lst = [
    "area_TL", "k", "s",
    "mean_TL_G", "sum_TL_G", "mean_TL_R", "sum_TL_R",
    "area_IF", "mean_IF_G", "sum_IF_G", "mean_IF_R", "sum_IF_R",
    "mean_IF_DR", "sum_IF_DR",
    "mean_IF_B", "sum_IF_B",
    "ratio", "sum_BG", "P_tot",
]
'''
lst = [
    "area_TL", "k", "s",
    "mean_IF_DR", "sum_IF_DR",
    "mean_IF_B", "sum_IF_B",
    "ratio", "P_tot",
]


sns.set_style("whitegrid")

proteins_by_line = {cl: list(full_dataset.get(cl, {}).keys()) for cl in CELL_LINES}
max_proteins = max((len(v) for v in proteins_by_line.values()), default=0)
set_clean_plot_style()
# --- Plot all pairs ---
for xcol, ycol in itertools.combinations(lst, 2):
    for start in range(0, max_proteins, PROTEINS_PER_FIG):
        ncols = min(PROTEINS_PER_FIG, max_proteins - start)
        if ncols <= 0:
            continue

        fig, axes = plt.subplots(
            nrows=len(CELL_LINES),
            ncols=ncols,
            figsize=(5.0 * ncols, 3.2 * len(CELL_LINES)),
            dpi=300,
            sharex=False,
            sharey=False,
            squeeze=False,
        )

        for row, cell_line in enumerate(CELL_LINES):
            proteins = proteins_by_line[cell_line]
            for col in range(ncols):
                idx = start + col
                ax = axes[row, col]

                if idx >= len(proteins):
                    ax.set_axis_off()
                    continue

                protein = proteins[idx]
                df = _concat_before_threshold_with_rep(full_dataset, cell_line, protein)
                plot_x_vs_y_fit_by_rep_clean(ax, df, xcol, ycol)

                ax.set_title(protein, fontsize=9)

                if col == 0:
                    ax.set_ylabel(f"{cell_line}\n{ycol}", fontsize=9)
                if row == len(CELL_LINES) - 1:
                    ax.set_xlabel(xcol, fontsize=9)

        end = min(start + ncols, max_proteins)
        fig.suptitle(
            f"{xcol} vs {ycol} (before_threshold) — protein index {start + 1}-{end}",
            fontsize=12,
        )
        plt.tight_layout()
        plt.show()
        plt.close(fig)
```

```python
# Distributions (mean_IF_DR, k, s), for all proteins and both cell lines.

def _concat_before_threshold(full_dataset, cell_line, protein):
    parts = []
    for rep in ("rep1", "rep2"):
        for well in ("well1", "well2"):
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep, {}).get(well, {})
            x = node.get("before_threshold", [])
            if isinstance(x, list):
                parts.extend([d for d in x if isinstance(d, pd.DataFrame)])
            elif isinstance(x, pd.DataFrame):
                parts.append(x)
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()

def plot_distributions(df, title, cols=("mean_IF_DR", "k", "s", "mean_IF_B", "sum_IF_B", "ratio", "P_tot")):
    present = [c for c in cols if c in df.columns]
    if not present:
        return
    fig, axes = plt.subplots(1, len(present), figsize=(5 * len(present), 4), dpi=200)
    if len(present) == 1:
        axes = [axes]
    for ax, col in zip(axes, present):
        x = pd.to_numeric(df[col], errors="coerce").dropna()
        if len(x) == 0:
            ax.set_title(f"{col} (no data)")
            ax.set_xlabel(col)
            ax.set_ylabel("probability density")
            continue
        sns.histplot(x, bins=50, stat="density", kde=True, ax=ax)
        ax.set_xlabel(col)
        ax.set_ylabel("probability density")
        ax.set_title(f"{col} (n={len(x)})")
    fig.suptitle(title)
    fig.subplots_adjust(top=0.80)
    plt.tight_layout()
    plt.show()

for cell_line in ("3T3", "mESCs", "hESCs"):
    for protein in full_dataset.get(cell_line, {}).keys():
        df = _concat_before_threshold(full_dataset, cell_line, protein)
        plot_distributions(df, title=f"{cell_line} | {protein} — distributions")
```

```python
# Compare distributions: well1 vs well2 (hist + KDE), for all proteins, both cell lines, each replicate.

def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()

def _before_threshold_df(full_dataset, cell_line, protein, rep, well):
    node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep, {}).get(well, {})
    return _as_df(node.get("before_threshold", []))

def plot_filter_comparison(
    df_filt1,
    df_filt2,
    *,
    cols=("mean_IF_DR", "k", "s"),
    label1="well1",
    label2="well2",
    bins=50,
    title=None,
    figsize=None,
    colors=None,
    alpha=0.35,
    kde=True,
    common_norm=False,
    clip=None,
    xlims=None,
    ):
    if figsize is None:
        figsize = (5 * len(cols), 4)
    if colors is None:
        colors = sns.color_palette(n_colors=2)
    fig, axes = plt.subplots(1, len(cols), figsize=figsize, dpi=200)
    if len(cols) == 1:
        axes = [axes]
    for ax, col in zip(axes, cols):
        if col not in df_filt1.columns or col not in df_filt2.columns:
            # still try to plot if one side has it
            x1 = pd.to_numeric(df_filt1[col], errors="coerce").dropna() if col in df_filt1.columns else pd.Series([], dtype=float)
            x2 = pd.to_numeric(df_filt2[col], errors="coerce").dropna() if col in df_filt2.columns else pd.Series([], dtype=float)
        else:
            x1 = pd.to_numeric(df_filt1[col], errors="coerce").dropna()
            x2 = pd.to_numeric(df_filt2[col], errors="coerce").dropna()
        if len(x1) == 0 and len(x2) == 0:
            ax.set_title(f"{col} (no data)")
            ax.set_xlabel(col)
            ax.set_ylabel("probability density")
            continue
        if len(x1) > 0:
            sns.histplot(
                x1, bins=bins, stat="density", kde=False, ax=ax,
                color=colors[0], alpha=alpha, element="step", fill=True, linewidth=1,
                label=f"{label1} (n={len(x1)})", common_norm=common_norm,
            )
        if len(x2) > 0:
            sns.histplot(
                x2, bins=bins, stat="density", kde=False, ax=ax,
                color=colors[1], alpha=alpha, element="step", fill=True, linewidth=1,
                label=f"{label2} (n={len(x2)})", common_norm=common_norm,
            )
        if kde:
            if len(x1) > 1:
                sns.kdeplot(x1, ax=ax, color=colors[0], linewidth=2, clip=clip)
            if len(x2) > 1:
                sns.kdeplot(x2, ax=ax, color=colors[1], linewidth=2, clip=clip)
        ax.set_xlabel(col)
        ax.set_ylabel("probability density")
        ax.set_title(col)
        if xlims and col in xlims:
            ax.set_xlim(*xlims[col])
    if title:
        fig.suptitle(title)
        fig.subplots_adjust(top=0.85)
    handles, labels = axes[0].get_legend_handles_labels()
    if handles:
        fig.legend(handles, labels, loc="upper right", frameon=True)
    plt.tight_layout()
    '''
        # 1. Get title
    #title = plt.gca().get_title()

    # 2. Make it filename-safe
    safe_title = re.sub(r'[^\w\-_. ]', '', title).replace(" ", "_")

    # 3. Create folder if it doesn't exist
    folder = "distributions 2 plots"
    #os.makedirs(folder, exist_ok=True)

    # 4. Save plot
    filepath = os.path.join(folder, f"{safe_title}.png")
    plt.savefig(filepath)
    '''
    plt.show()

for cell_line in ("3T3", "mESCs", "hESCs"):
    for protein in full_dataset.get(cell_line, {}).keys():
        for rep in ("rep1", "rep2"):
            df_w1 = _before_threshold_df(full_dataset, cell_line, protein, rep, "well1")
            df_w2 = _before_threshold_df(full_dataset, cell_line, protein, rep, "well2")
            # Skip empty
            if df_w1.empty and df_w2.empty:
                continue
            plot_filter_comparison(
                df_w1, df_w2,
                title=f"{cell_line} | {protein} — {rep}: well1 vs well2 distributions",
                label1=f"{rep} well1",
                label2=f"{rep} well2",
            )
```

```python
# Sliding-window (continuous) DR variability along k-quantiles, using the UNSPLIT data (before_threshold), ALL prots, not good
#
# Idea: sort cells by k, take a fixed-mass window (fixed number of cells == fixed area under the empirical k distribution),
# compute CV = std(mean_IF_DR) / mean(mean_IF_DR) in that window, then slide the window across k.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Controls ---
# Choose ONE of these window definitions:
WINDOW_N = None        # e.g. 1000  -> fixed number of cells per window (recommended if you want “same #cells”)
WINDOW_FRAC = 0.05     # e.g. 0.15  -> fixed fraction of the dataset per window (ignored if WINDOW_N is set)

# How far to slide the window each step:
STEP_N = None          # e.g. 100   -> fixed step in cells (ignored if None)
STEP_FRAC = 0.01      # e.g. 0.02  -> step as fraction of dataset (ignored if STEP_N is set)

MIN_WINDOW_N = 50


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein):
    parts = []
    for rep in ("rep1", "rep2"):
        for well in ("well1", "well2"):
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep, {}).get(well, {})
            parts.append(_as_df(node.get("before_threshold", [])))
    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(
    df,
    *,
    k_col="k",
    y_col="mean_IF_DR",
    window_frac=0.15,
    step_frac=0.02,
    window_n=None,
    step_n=None,
    min_window_n=50,
):
    """Return a curve of CV(y)=sd(y)/mean(y) over sliding windows along sorted k."""
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)

    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))

    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy()

        mu = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu

        q_center = (start + 0.5 * (window_n - 1)) / (n - 1)  # in [0,1]
        k_median = float(np.median(sub[k_col].to_numpy()))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def plot_sliding_cv_for_dataset(cell_line, protein, *, window_frac, step_frac, window_n, step_n):
    df_all = _concat_before_threshold(full_dataset, cell_line, protein)
    curve = _sliding_k_quantile_cv(
        df_all,
        window_frac=window_frac,
        step_frac=step_frac,
        window_n=window_n,
        step_n=step_n,
        min_window_n=MIN_WINDOW_N,
    )
    if curve.empty:
        return

    sns.set_style("whitegrid")
    plt.figure(figsize=(7.2, 4.6), dpi=200)
    plt.plot(curve["q_center"], curve["cv"], linewidth=2)
    plt.xlabel("k quantile (sliding window center)")
    plt.ylabel("DR variability: std(mean_IF_DR) / mean(mean_IF_DR)")

    win_txt = f"window={int(curve['n_window'].iloc[0])} cells"
    plt.title(f"{cell_line} | {protein} — sliding DR variability vs k-quantile ({win_txt})")
    plt.ylim(bottom=0)
    plt.tight_layout()
    plt.show()


# Plot for all proteins and both cell lines
for cell_line in ("3T3", "mESCs", "hESCs"):
    for protein in full_dataset.get(cell_line, {}).keys():
        plot_sliding_cv_for_dataset(
            cell_line,
            protein,
            window_frac=WINDOW_FRAC,
            step_frac=STEP_FRAC,
            window_n=WINDOW_N,
            step_n=STEP_N,
        )
```

```python
# SLIDING Protein CV vs k
# Sliding-window (continuous) DR variability along k-quantiles, using the UNSPLIT data (before_threshold).
#
# Idea: sort cells by k, take a fixed-mass window (fixed number of cells == fixed area under the empirical k distribution),
# compute CV = std(mean_IF_DR) / mean(mean_IF_DR) in that window, then slide the window across k.
#
# What this version does (per your split):
# - Plots 4 curves: rep1_well1, rep1_well2, rep2_well1, rep2_well2
# - Adds 2 averages: rep1_avg = mean(rep1_well1, rep1_well2), rep2_avg = mean(rep2_well1, rep2_well2)
# - Adds y-axis padding (min - margin, max + margin)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re

# --- Controls ---
# Choose ONE of these window definitions:
WINDOW_N = None        # e.g. 1000  -> fixed number of cells per window (recommended if you want “same #cells”)
WINDOW_FRAC = 1/6     # e.g. 0.15  -> fixed fraction of the dataset per window (ignored if WINDOW_N is set)

# How far to slide the window each step:
STEP_N = None          # e.g. 100   -> fixed step in cells (ignored if None)
STEP_FRAC = 0.01      # e.g. 0.02  -> step as fraction of dataset (ignored if STEP_N is set)

MIN_WINDOW_N = 50

# Common x-grid for plotting/averaging (0..1 corresponds to k-quantiles)
X_GRID_N = 250


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []

    reps = (rep,) if rep is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")

    for rep_ in reps:
        for well_ in wells:
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep_, {}).get(well_, {})
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(
    df,
    *,
    k_col="k",
    col2="s",
    y_col="mean_IF_DR",
    window_frac=0.20,
    step_frac=0.01,
    window_n=None,
    step_n=None,
    min_window_n=50,
):
    """Return a curve of CV(y)=sd(y)/mean(y) over sliding windows along sorted k.

    Output columns:
    - q_center: window center position in quantile space (0..1)
    - k_median: median k in the window
    - cv: sd(y)/mean(y)
    - n_window: number of points in each window
    """
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)

    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work2 = df[[k_col, col2]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work2[k_col] = pd.to_numeric(work2[k_col], errors="coerce")
    work2[col2] = pd.to_numeric(work2[col2], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))
    
    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))
    print(f"DEBUG: n={n}, window_n={window_n}, step_n={step_n}")
    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    denom = max(1, n - 1)
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy()
        sub2 = work2.iloc[start : start + window_n]
        y2 = sub2[col2].to_numpy()
        mu2 = float(np.mean(y2))
        mu = float(np.mean(y))
        mu3 = mu * mu2
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu

        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[k_col].to_numpy()))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="cv"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    # De-duplicate x (rare but can happen); keep first occurrence
    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]

    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def plot_sliding_cv_groups(cell_line, protein, *, window_frac, step_frac, window_n, step_n):
    curves = {
        "rep1_well1": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep1", well="well1"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
        ),
        "rep1_well2": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep1", well="well2"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
        ),
        "rep2_well1": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep2", well="well1"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
        ),
        "rep2_well2": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep2", well="well2"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
        ),
    }

    if all(c.empty for c in curves.values()):
        return

    x_grid = np.linspace(0.0, 1.0, X_GRID_N)
    y_grid = {name: _interp_to_grid(curve, x_grid) for name, curve in curves.items()}

    # Averages within replicate (pointwise in quantile space)
    '''
    y_grid["rep1_avg"] = np.nanmean(np.vstack([y_grid["rep1_well1"], y_grid["rep1_well2"]]), axis=0)
    y_grid["rep2_avg"] = np.nanmean(np.vstack([y_grid["rep2_well1"], y_grid["rep2_well2"]]), axis=0)
    '''
    # Only display average if both wells have data
    a = y_grid["rep1_well1"]
    b = y_grid["rep1_well2"]
    stack = np.vstack([a, b])
    avg = np.nanmean(stack, axis=0)
    count = np.isfinite(stack).sum(axis=0)
    avg[count < 2] = np.nan
    y_grid["rep1_avg"] = avg

    a = y_grid["rep2_well1"]
    b = y_grid["rep2_well2"]
    stack = np.vstack([a, b])
    avg = np.nanmean(stack, axis=0)
    count = np.isfinite(stack).sum(axis=0)
    avg[count < 2] = np.nan
    y_grid["rep2_avg"] = avg

    sns.set_style("whitegrid")
    fig, ax = plt.subplots(figsize=(7.8, 4.9), dpi=300)

    style = {
        "rep1_well1": dict(color="C0", lw=2.0),
        "rep1_well2": dict(color="C1", lw=2.0),
        "rep2_well1": dict(color="C2", lw=2.0),
        "rep2_well2": dict(color="C3", lw=2.0),
        "rep1_avg": dict(color="black", lw=2.4, ls="--"),
        "rep2_avg": dict(color="dimgray", lw=2.4, ls="--"),
    }

    for key in ("rep1_well1", "rep1_well2", "rep2_well1", "rep2_well2", "rep1_avg", "rep2_avg"):
        y = y_grid.get(key)
        if y is None or np.all(np.isnan(y)):
            continue
        ax.plot(x_grid, y, label=key, **style.get(key, {}))

    ax.set_xlabel("k")
    ax.set_ylabel("CV")

    if window_n is not None:
        win_txt = f"window={int(window_n)} cells"
    else:
        win_txt = f"window={window_frac:.0%} of cells"

    if step_n is not None:
        step_txt = f"step={int(step_n)} cells"
    else:
        step_txt = f"step={step_frac:.0%}"

    ax.set_title(f"{cell_line}  |  {protein}")

    # Y-limits with padding
    finite_chunks = []
    for v in y_grid.values():
        if v is None:
            continue
        vv = v[np.isfinite(v)]
        if vv.size:
            finite_chunks.append(vv)

    if finite_chunks:
        all_y = np.concatenate(finite_chunks)
        y_min = float(np.min(all_y))
        y_max = float(np.max(all_y))
        y_rng = y_max - y_min
        pad = 0.05 * y_rng if y_rng > 0 else (0.05 * abs(y_max) if y_max != 0 else 0.1)
        ax.set_ylim(y_min - pad, y_max + pad)

    ax.legend(frameon=True)
    plt.tight_layout()
    
    
    # 1. Get title
    title = plt.gca().get_title()

    # 2. Make it filename-safe
    safe_title = re.sub(r'[^\w\-_. ]', '', title).replace(" ", "_")

    # 3. Create folder if it doesn't exist
    folder = "DR variability continuous plots\\DR_vs_S"
    os.makedirs(folder, exist_ok=True)

    # 4. Save plot
    filepath = os.path.join(folder, f"{safe_title}.png")
    plt.savefig(filepath)
    
    plt.show()


# Plot for all proteins and both cell lines

for cell_line in ("3T3", "mESCs", "hESCs"):
    for protein in full_dataset.get(cell_line, {}).keys():
        plot_sliding_cv_groups(
            cell_line,
            protein,
            window_frac=WINDOW_FRAC,
            step_frac=STEP_FRAC,
            window_n=WINDOW_N,
            step_n=STEP_N,
        )
```

```python
# SLIDING Protein CV vs k V2 (rep1 left axis, rep2 right axis)
# One figure per cell type:
# - 3T3: 2x3
# - mESCs: 2x3
# - hESCs: 2x2
#
# For each protein subplot: plot ONLY the average curve between well1 and well2
# (separately for rep1 and rep2), with nice colors.
# Legend is added ONCE per figure:
# - For 3T3/mESCs: placed in the empty 6th subplot
# - For hESCs: placed at figure level (upper right)
#
# NEW:
# - rep1 plotted on LEFT y-axis
# - rep2 plotted on RIGHT y-axis (twinx)
# - left/right axis spine + ticks colored to match rep colors

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D


# --- Controls ---
WINDOW_N = None        # e.g. 1000 -> fixed #cells per window (overrides WINDOW_FRAC)
WINDOW_FRAC = 1 / 6    # fixed fraction of cells per window

STEP_N = None          # e.g. 100  -> fixed step (overrides STEP_FRAC)
STEP_FRAC = 0.01       # step as fraction of dataset

MIN_WINDOW_N = 50
X_GRID_N = 250

K_COL = "k"
Y_COL = "mean_IF_DR"

SAVE = True
SAVE_FOLDER = r"DR variability continuous plots\DR_vs_K_CHXstyle"
SAVE_DPI = 300


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []
    reps = (rep,) if rep is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")

    for rep_ in reps:
        for well_ in wells:
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep_, {}).get(well_, {})
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(
    df,
    *,
    k_col=K_COL,
    y_col=Y_COL,
    window_frac=WINDOW_FRAC,
    step_frac=STEP_FRAC,
    window_n=WINDOW_N,
    step_n=STEP_N,
    min_window_n=MIN_WINDOW_N,
):
    """CV(y)=std(y)/mean(y) over sliding windows along sorted k.
    Returns columns: q_center (0..1), k_median, cv, n_window
    """
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))

    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    denom = max(1, n - 1)

    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy(dtype=float)

        mu = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu

        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[k_col].to_numpy(dtype=float)))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="cv"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]
    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def _avg_two_wells_on_grid(y_w1, y_w2):
    """Pointwise average, only where both wells finite (no warnings)."""
    y_w1 = np.asarray(y_w1, dtype=float)
    y_w2 = np.asarray(y_w2, dtype=float)

    avg = np.full_like(y_w1, np.nan, dtype=float)
    mask = np.isfinite(y_w1) & np.isfinite(y_w2)
    avg[mask] = 0.5 * (y_w1[mask] + y_w2[mask])
    return avg


def _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid):
    df_w1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
    df_w2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")

    c1 = _sliding_k_quantile_cv(df_w1)
    c2 = _sliding_k_quantile_cv(df_w2)

    y1 = _interp_to_grid(c1, x_grid)
    y2 = _interp_to_grid(c2, x_grid)

    return _avg_two_wells_on_grid(y1, y2)


def plot_sliding_avg_only_ax(full_dataset, cell_line, protein, *, ax, rep_colors):
    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    y_rep1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y_rep2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)

    # right axis for rep2
    ax2 = ax.twinx()
    ax2.grid(False)

    plotted1 = False
    plotted2 = False

    if not np.all(np.isnan(y_rep1)):
        ax.plot(x_grid, y_rep1, color=rep_colors["rep1"], lw=2.4, label="rep1 avg(well1,well2)")
        plotted1 = True

    if not np.all(np.isnan(y_rep2)):
        ax2.plot(x_grid, y_rep2, color=rep_colors["rep2"], lw=2.4, label="rep2 avg(well1,well2)")
        plotted2 = True

    if not (plotted1 or plotted2):
        ax.axis("off")
        ax2.axis("off")
        return None  # returning ax2 not needed by caller if axis is off

    ax.set_title(f"{protein}", fontsize=12, pad=6)
    ax.set_xlim(0, 1)

    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax.spines["top"].set_visible(False)
    ax2.spines["top"].set_visible(False)

    # clean overlap spines
    ax.spines["right"].set_visible(False)
    ax2.spines["left"].set_visible(False)

    # color-code y axes to match reps
    if plotted1:
        ax.spines["left"].set_color(rep_colors["rep1"])
        ax.tick_params(axis="y", colors=rep_colors["rep1"])
    else:
        ax.tick_params(axis="y", left=False, labelleft=False)
        ax.spines["left"].set_visible(False)

    if plotted2:
        ax2.spines["right"].set_color(rep_colors["rep2"])
        ax2.tick_params(axis="y", colors=rep_colors["rep2"])
    else:
        ax2.tick_params(axis="y", right=False, labelright=False)
        ax2.spines["right"].set_visible(False)

    return ax2


def plot_celltype_sliding_fig(full_dataset, cell_line, *, nrows, ncols, save=SAVE):
    proteins = list(full_dataset.get(cell_line, {}).keys())
    nslots = nrows * ncols
    proteins_to_plot = proteins[:nslots]

    sns.set_style("whitegrid")
    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(4.7 * ncols, 3.5 * nrows),
        dpi=SAVE_DPI,
        sharex=False,
        sharey=False,
        squeeze=False,
    )

    pal = sns.color_palette("colorblind", 2)
    rep_colors = {"rep1": pal[0], "rep2": pal[1]}

    # store twinx axes (same grid shape)
    axes2 = [[None for _ in range(ncols)] for _ in range(nrows)]

    for i, ax in enumerate(axes.ravel()):
        r, c = divmod(i, ncols)
        if i < len(proteins_to_plot):
            prot = proteins_to_plot[i]
            ax2 = plot_sliding_avg_only_ax(full_dataset, cell_line, prot, ax=ax, rep_colors=rep_colors)
            axes2[r][c] = ax2
        else:
            ax.axis("off")
            axes2[r][c] = None

    # labels: left label only first col, right label only last col
    # labels: show x label on every subplot + show rep2 tick labels on every subplot
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            ax2 = axes2[r][c]

            if not ax.axison:
                continue

            # X label on each plot
            ax.set_xlabel("k")

            # Left y-label only on first column (rep1)
            if c == 0:
                ax.set_ylabel("CV", color="black")#rep_colors["rep1"])
            else:
                ax.set_ylabel("")

            # Right axis: ALWAYS show tick labels (rep2 values) on every column
            if ax2 is not None:
                ax2.tick_params(labelright=True)

                # Keep the right-side y-label text only on last column (less clutter)
                if c == ncols - 1:
                    ax2.set_ylabel("", color=rep_colors["rep2"])
                else:
                    ax2.set_ylabel("")

    legend_handles = [
        Line2D([0], [0], color=rep_colors["rep1"], lw=2.4, label="Biological replicate 1"),
        Line2D([0], [0], color=rep_colors["rep2"], lw=2.4, label="Biological replicate 2"),
    ]

    # legend placement:
    # - use empty subplot when available (typically 3T3/mESCs in 2x3 with 5 proteins)
    # - otherwise use figure-level legend (hESCs)
    empty_axes = []
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            ax2 = axes2[r][c]
            if not ax.axison:
                continue
            has_any_data = ax.has_data() or (ax2 is not None and ax2.has_data())
            if not has_any_data:
                empty_axes.append(ax)

    if len(empty_axes) > 0:
        ax_leg = empty_axes[0]
        ax_leg.axis("off")
        ax_leg.legend(
            handles=legend_handles,
            loc="center",
            frameon=True,
            fontsize=9,
            title="Averages",
            title_fontsize=10,
        )
    else:
        fig.legend(
            handles=legend_handles,
            loc="upper right",
            frameon=True,
            fontsize=9,
            title="Averages",
            title_fontsize=10,
        )

    fig.suptitle(
        f"{cell_line} | Protein CV vs k (16.67% of cells/CV value)",
        y=0.995,
        fontsize=14,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.96], pad=1.0, w_pad=0.8, h_pad=0.9)
    fig.subplots_adjust(wspace=0.35, hspace=0.35)

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out_path = os.path.join(
            SAVE_FOLDER,
            f"{cell_line}_sliding_CV_vs_kquantile_avgWells_only_dualAxis.png",
        )
        fig.savefig(out_path, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out_path}")

    plt.show()


# --- Run: one figure per cell type with requested layouts ---
plot_celltype_sliding_fig(full_dataset, "3T3",  nrows=2, ncols=2, save=True)
plot_celltype_sliding_fig(full_dataset, "mESCs", nrows=2, ncols=3, save=True)
plot_celltype_sliding_fig(full_dataset, "hESCs", nrows=2, ncols=2, save=True)
```

```python
# SLIDING Protein CV vs k V2 (rep1 left axis, rep2 right axis) with fixed span
# One figure per cell type:
# - 3T3: 2x3
# - mESCs: 2x3
# - hESCs: 2x2
#
# For each protein subplot: plot ONLY the average curve between well1 and well2
# (separately for rep1 and rep2), with nice colors.
# Legend is added ONCE per figure:
# - For 3T3/mESCs: placed in the empty 6th subplot
# - For hESCs: placed at figure level (upper right)
#
# NEW:
# - rep1 plotted on LEFT y-axis
# - rep2 plotted on RIGHT y-axis (twinx)
# - left/right axis spine + ticks colored to match rep colors

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
from matplotlib import ticker as mticker  # add this import


# --- Controls ---
WINDOW_N = None        # e.g. 1000 -> fixed #cells per window (overrides WINDOW_FRAC)
WINDOW_FRAC = 1 / 5    # fixed fraction of cells per window

STEP_N = None          # e.g. 100  -> fixed step (overrides STEP_FRAC)
STEP_FRAC = 0.01       # step as fraction of dataset

MIN_WINDOW_N = 50
X_GRID_N = 250

K_COL = "s"
Y_COL = "mean_IF_DR"

SAVE = True
SAVE_FOLDER = r"DR variability continuous plots\DR_vs_S2"
SAVE_DPI = 300

# --- Controls ---
CV_YSPAN = 0.08             # <- set this to whatever span you want
CV_ENSURE_DATA_VISIBLE = True # if False, span is forced exactly and may clip data
CV_CLAMP_MIN = 0.0           # set None to allow negative (usually CV>=0 so 0.0 is fine)
CV_N_TICKS = 5  
CV_DECIMALS = 2   # <- number of decimals on BOTH y-axes
CV_DECIMALS = 2   # <- number of decimals on BOTH y-axes

def _apply_y_ticks(ax, *, n_ticks=CV_N_TICKS, decimals=CV_DECIMALS):
    ax.yaxis.set_major_locator(mticker.LinearLocator(n_ticks))

    # fixed decimals, and avoid showing "-0.00"
    eps = 0.5 * 10 ** (-decimals)
    ax.yaxis.set_major_formatter(
        mticker.FuncFormatter(lambda v, pos: f"{0.0:.{decimals}f}" if abs(v) < eps else f"{v:.{decimals}f}")
    )

def _set_fixed_yspan(ax, y, *, span=CV_YSPAN, ensure_visible=CV_ENSURE_DATA_VISIBLE, clamp_min=CV_CLAMP_MIN):
    """Set ax y-limits so (ymax - ymin) == span (or >= span if ensure_visible)."""
    y = np.asarray(y, dtype=float)
    finite = y[np.isfinite(y)]
    if finite.size == 0:
        return

    y_min = float(finite.min())
    y_max = float(finite.max())
    data_span = y_max - y_min

    span_use = max(float(span), data_span) if ensure_visible else float(span)
    center = 0.5 * (y_min + y_max)

    lo = center - 0.5 * span_use
    hi = center + 0.5 * span_use

    if clamp_min is not None and lo < clamp_min:
        shift = clamp_min - lo
        lo += shift
        hi += shift

    ax.set_ylim(lo, hi)

def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []
    reps = (rep,) if rep is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")

    for rep_ in reps:
        for well_ in wells:
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep_, {}).get(well_, {})
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(
    df,
    *,
    k_col=K_COL,
    y_col=Y_COL,
    window_frac=WINDOW_FRAC,
    step_frac=STEP_FRAC,
    window_n=WINDOW_N,
    step_n=STEP_N,
    min_window_n=MIN_WINDOW_N,
):
    """CV(y)=std(y)/mean(y) over sliding windows along sorted k.
    Returns columns: q_center (0..1), k_median, cv, n_window
    """
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))

    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    denom = max(1, n - 1)

    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy(dtype=float)

        mu = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu

        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[k_col].to_numpy(dtype=float)))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="cv"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]
    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def _avg_two_wells_on_grid(y_w1, y_w2):
    """Pointwise average, only where both wells finite (no warnings)."""
    y_w1 = np.asarray(y_w1, dtype=float)
    y_w2 = np.asarray(y_w2, dtype=float)

    avg = np.full_like(y_w1, np.nan, dtype=float)
    mask = np.isfinite(y_w1) & np.isfinite(y_w2)
    avg[mask] = 0.5 * (y_w1[mask] + y_w2[mask])
    return avg


def _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid):
    df_w1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
    df_w2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")

    c1 = _sliding_k_quantile_cv(df_w1)
    c2 = _sliding_k_quantile_cv(df_w2)

    y1 = _interp_to_grid(c1, x_grid)
    y2 = _interp_to_grid(c2, x_grid)

    return _avg_two_wells_on_grid(y1, y2)


def plot_sliding_avg_only_ax(full_dataset, cell_line, protein, *, ax, rep_colors):
    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    y_rep1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y_rep2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)

    # right axis for rep2
    ax2 = ax.twinx()
    ax2.grid(False)

    plotted1 = False
    plotted2 = False

    if not np.all(np.isnan(y_rep1)):
        ax.plot(x_grid, y_rep1, color=rep_colors["rep1"], lw=2.4, label="rep1 avg(well1,well2)")
        plotted1 = True

    if not np.all(np.isnan(y_rep2)):
        ax2.plot(x_grid, y_rep2, color=rep_colors["rep2"], lw=2.4, label="rep2 avg(well1,well2)")
        plotted2 = True

    if not (plotted1 or plotted2):
        ax.axis("off")
        ax2.axis("off")
        return None  # returning ax2 not needed by caller if axis is off

    if plotted1:
        _set_fixed_yspan(ax, y_rep1)
        _apply_y_ticks(ax)

    if plotted2:
        _set_fixed_yspan(ax2, y_rep2)
        _apply_y_ticks(ax2)

    ax.set_title(f"{protein}", fontsize=12, pad=6)
    ax.set_xlim(0, 1)

    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax.spines["top"].set_visible(False)
    ax2.spines["top"].set_visible(False)

    # clean overlap spines
    ax.spines["right"].set_visible(False)
    ax2.spines["left"].set_visible(False)

    # color-code y axes to match reps
    if plotted1:
        ax.spines["left"].set_color(rep_colors["rep1"])
        ax.tick_params(axis="y", colors=rep_colors["rep1"])
    else:
        ax.tick_params(axis="y", left=False, labelleft=False)
        ax.spines["left"].set_visible(False)

    if plotted2:
        ax2.spines["right"].set_color(rep_colors["rep2"])
        ax2.tick_params(axis="y", colors=rep_colors["rep2"])
    else:
        ax2.tick_params(axis="y", right=False, labelright=False)
        ax2.spines["right"].set_visible(False)

    return ax2


def plot_celltype_sliding_fig(full_dataset, cell_line, *, nrows, ncols, save=SAVE):
    proteins = list(full_dataset.get(cell_line, {}).keys())
    nslots = nrows * ncols
    proteins_to_plot = proteins[:nslots]

    sns.set_style("whitegrid")
    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(4.7 * ncols, 3.5 * nrows),
        dpi=SAVE_DPI,
        sharex=False,
        sharey=False,
        squeeze=False,
    )

    pal = sns.color_palette("colorblind", 2)
    rep_colors = {"rep1": pal[0], "rep2": pal[1]}

    # store twinx axes (same grid shape)
    axes2 = [[None for _ in range(ncols)] for _ in range(nrows)]

    for i, ax in enumerate(axes.ravel()):
        r, c = divmod(i, ncols)
        if i < len(proteins_to_plot):
            prot = proteins_to_plot[i]
            ax2 = plot_sliding_avg_only_ax(full_dataset, cell_line, prot, ax=ax, rep_colors=rep_colors)
            axes2[r][c] = ax2
        else:
            ax.axis("off")
            axes2[r][c] = None

    # labels: left label only first col, right label only last col
    # labels: show x label on every subplot + show rep2 tick labels on every subplot
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            ax2 = axes2[r][c]

            if not ax.axison:
                continue

            # X label on each plot
            ax.set_xlabel("s", fontsize=10)

            # Left y-label only on first column (rep1)
            if c == 0:
                ax.set_ylabel("CV", color="black")#rep_colors["rep1"])
            else:
                ax.set_ylabel("")

            # Right axis: ALWAYS show tick labels (rep2 values) on every column
            if ax2 is not None:
                ax2.tick_params(labelright=True)

                # Keep the right-side y-label text only on last column (less clutter)
                if c == ncols - 1:
                    ax2.set_ylabel("", color=rep_colors["rep2"])
                else:
                    ax2.set_ylabel("")

    legend_handles = [
        Line2D([0], [0], color=rep_colors["rep1"], lw=2.4, label="Biological replicate 1"),
        Line2D([0], [0], color=rep_colors["rep2"], lw=2.4, label="Biological replicate 2"),
    ]

    # legend placement:
    # - use empty subplot when available (typically 3T3/mESCs in 2x3 with 5 proteins)
    # - otherwise use figure-level legend (hESCs)
    empty_axes = []
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            ax2 = axes2[r][c]
            if not ax.axison:
                continue
            has_any_data = ax.has_data() or (ax2 is not None and ax2.has_data())
            if not has_any_data:
                empty_axes.append(ax)

    if len(empty_axes) > 0:
        ax_leg = empty_axes[0]
        ax_leg.axis("off")
        ax_leg.legend(
            handles=legend_handles,
            loc="center",
            frameon=True,
            fontsize=9,
            title=None,
            title_fontsize=10,
        )
    else:
        fig.legend(
            handles=legend_handles,
            loc="upper right",
            frameon=True,
            fontsize=9,
            title=None,
            title_fontsize=10,
        )

    fig.suptitle(
        f"{cell_line} | Protein CV vs Synthesis rate (s)",
        y=0.995,
        fontsize=14,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.96], pad=1.0, w_pad=0.8, h_pad=0.9)
    fig.subplots_adjust(wspace=0.35, hspace=0.35)

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out_path = os.path.join(
            SAVE_FOLDER,
            f"{cell_line}_sliding_CV_vs_kquantile_avgWells_only_dualAxis.png",
        )
        fig.savefig(out_path, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out_path}")

    plt.show()


# --- Run: one figure per cell type with requested layouts ---
plot_celltype_sliding_fig(full_dataset, "3T3",  nrows=2, ncols=2, save=True)
plot_celltype_sliding_fig(full_dataset, "mESCs", nrows=2, ncols=3, save=True)
plot_celltype_sliding_fig(full_dataset, "hESCs", nrows=2, ncols=2, save=True)



def _sliding_window_settings(
    df,
    *,
    k_col=K_COL,
    y_col=Y_COL,
    window_frac=WINDOW_FRAC,
    step_frac=STEP_FRAC,
    window_n=WINDOW_N,
    step_n=STEP_N,
    min_window_n=MIN_WINDOW_N,
):
    """Return effective n_total, window_n, step_n and n_windows for THIS df (after cleaning)."""
    if df is None or df.empty:
        return None
    if not ({k_col, y_col} <= set(df.columns)):
        return None

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return None

    wn = int(round(window_frac * n)) if window_n is None else int(window_n)
    wn = int(max(min_window_n, min(wn, n)))

    sn = int(round(step_frac * n)) if step_n is None else int(step_n)
    sn = int(max(1, sn))

    n_windows = len(range(0, n - wn + 1, sn))
    return {"n_total": n, "window_n": wn, "step_n": sn, "n_windows": n_windows}


def print_sliding_window_counts_combined_wells(full_dataset, cell_line, proteins=None):
    """
    Prints counts using the COMBINED dataset (well1 + well2) for each replicate.
    n_total is after dropping NaNs in K_COL and Y_COL (same cleaning as your CV code).
    """
    proteins = proteins or list(full_dataset.get(cell_line, {}).keys())

    rows = []
    for protein in proteins:
        for rep in ("rep1", "rep2"):
            # well=None => both wells are included inside _concat_before_threshold
            df_both = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well=None)
            s = _sliding_window_settings(df_both)

            rows.append({
                "cell_line": cell_line,
                "protein": protein,
                "rep": rep,
                "n_cells_w1+w2": (s["n_total"] if s else np.nan),
                "window_n": (s["window_n"] if s else np.nan),
                "step_n": (s["step_n"] if s else np.nan),
                "n_windows": (s["n_windows"] if s else np.nan),
            })

    out = pd.DataFrame(rows).sort_values(["cell_line", "protein", "rep"]).reset_index(drop=True)

    print(f"\n=== Sliding window cell counts (combined well1+well2): {cell_line} ===")
    with pd.option_context("display.max_rows", None, "display.max_columns", None):
        print(out.to_string(index=False))

    return out

print_sliding_window_counts_combined_wells(full_dataset, "3T3")
print_sliding_window_counts_combined_wells(full_dataset, "mESCs")
print_sliding_window_counts_combined_wells(full_dataset, "hESCs")
```

```python
# SLIDING Protein CV vs k/s V2 (rep1 left axis, rep2 right axis) with fixed span
# One figure per cell type:
# - 3T3: 2x3
# - mESCs: 2x3
# - hESCs: 2x2
#
# For each protein subplot: plot ONLY the average curve between well1 and well2
# (separately for rep1 and rep2), with nice colors.
# Legend is added ONCE per figure:
# - For 3T3/mESCs: placed in the empty 6th subplot
# - For hESCs: placed at figure level (upper right)
#
# NEW:
# - rep1 plotted on LEFT y-axis
# - rep2 plotted on RIGHT y-axis (twinx)
# - left/right axis spine + ticks colored to match rep colors
# - linear least-squares fit per replicate (slope + MSE), plotted + annotated

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
from matplotlib import ticker as mticker


# --- Controls (sliding window) ---
WINDOW_N = None        # e.g. 1000 -> fixed #cells per window (overrides WINDOW_FRAC)
WINDOW_FRAC = 1 / 5    # fixed fraction of cells per window

STEP_N = None          # e.g. 100  -> fixed step (overrides STEP_FRAC)
STEP_FRAC = 0.01       # step as fraction of dataset

MIN_WINDOW_N = 50
X_GRID_N = 250

K_COL = "s"
Y_COL = "mean_IF_DR"

SAVE = True
SAVE_FOLDER = r"DR variability continuous plots\DR_vs_S2"
SAVE_DPI = 300


# --- Axis formatting controls ---
CV_YSPAN = 0.08                 # y-range span for each axis (left and right independently)
CV_ENSURE_DATA_VISIBLE = True   # if False, span is forced exactly and may clip data/fit
CV_CLAMP_MIN = 0.0              # set None to allow negative
CV_N_TICKS = 5                  # same number of ticks on left/right axes
CV_DECIMALS = 2                 # decimals for tick labels on BOTH y-axes


# --- Linear fit controls (per replicate curve) ---
SHOW_LINEAR_FIT = True
FIT_PLOT_ONLY_WHERE_DATA = True  # if True: draw fit only where curve has finite values
FIT_LS = "--"
FIT_LW = 1.7
FIT_ALPHA = 0.85
FIT_FONT_SIZE = 8
FIT_DECIMALS_SLOPE = 3           # slope shown with fixed decimals
FIT_SIGFIGS_MSE = 2              # MSE shown with significant figures

# --- Spearman correlation (per plot + heatmaps) ---
from scipy.stats import spearmanr

SHOW_SPEARMAN_TEXT = True     # annotate each subplot (rep1/rep2) with Spearman ρ
SPEARMAN_DECIMALS = 2
SPEARMAN_MIN_POINTS = 8

# Heatmap x-variable choice:
# - "q_center": Spearman between CV and the plotted x-axis in your code (the 0..1 quantile grid)
# - "k_median": Spearman between CV and actual window-median s (uses sliding-window points, not the interpolated curve)
SPEARMAN_HEATMAP_X_MODE = "q_center"   # "q_center" or "k_median"
SPEARMAN_HEATMAP_ANNOT = True
SPEARMAN_CMAP = "vlag"

DELTA_YSPAN = 0.05   # zoom for ΔCV
RATIO_YSPAN = 0.2    # zoom for CV/CV0 (often near ~1)
DELTA_DECIMALS = 3   # optional: more decimals for small ΔCV

# ---------- One figure: 3 columns (cell types), ΔCV only, 0-line shown ----------

DELTA_YSPAN = 0.1      # zoom for ΔCV (tune if needed; you said ~0.05 range)
DELTA_DECIMALS = 3
DELTA_BASELINE_N = 1    # use first N finite points to define baseline (more stable than N=1)

def make_protein_color_map(full_dataset, *, palette="husl"):
    all_prots = sorted({prot for cl in full_dataset for prot in full_dataset.get(cl, {}).keys()})
    cols = sns.color_palette(palette, n_colors=max(1, len(all_prots)))
    return {p: cols[i] for i, p in enumerate(all_prots)}

def _apply_y_ticks_include_zero(ax, *, n_ticks=CV_N_TICKS, decimals=CV_DECIMALS):
    lo, hi = ax.get_ylim()
    lo = float(min(lo, 0.0))
    hi = float(max(hi, 0.0))
    if lo == hi:
        hi = lo + 1e-9
    ax.set_ylim(lo, hi)

    ticks = np.linspace(lo, hi, int(max(2, n_ticks)))
    # Replace the closest tick to 0 with exactly 0
    ticks[int(np.argmin(np.abs(ticks)))] = 0.0

    ax.yaxis.set_major_locator(mticker.FixedLocator(ticks))

    eps = 0.5 * 10 ** (-decimals)  # avoid "-0.00"
    ax.yaxis.set_major_formatter(
        mticker.FuncFormatter(
            lambda v, pos: f"{0.0:.{decimals}f}" if abs(v) < eps else f"{v:.{decimals}f}"
        )
    )

def _avg_two_arrays_strict(y1, y2):
    y1 = np.asarray(y1, dtype=float)
    y2 = np.asarray(y2, dtype=float)
    out = np.full_like(y1, np.nan, dtype=float)
    m = np.isfinite(y1) & np.isfinite(y2)
    out[m] = 0.5 * (y1[m] + y2[m])
    return out

def _avg_reps_curve_on_qgrid(full_dataset, cell_line, protein, x_grid):
    y1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)
    return _avg_two_arrays_strict(y1, y2)

def _baseline_from_left(y, *, n=DELTA_BASELINE_N):
    y = np.asarray(y, dtype=float)
    finite = y[np.isfinite(y)]
    if finite.size == 0:
        return np.nan
    n = int(max(1, min(n, finite.size)))
    return float(np.median(finite[:n]))  # median of first n windows

def _delta_cv(y, *, baseline_n=DELTA_BASELINE_N):
    y = np.asarray(y, dtype=float)
    cv0 = _baseline_from_left(y, n=baseline_n)
    if not np.isfinite(cv0):
        return np.full_like(y, np.nan, dtype=float)
    return y - cv0

def plot_all_celltypes_allproteins_deltaCV(
    full_dataset,
    *,
    cell_lines=("3T3", "mESCs", "hESCs"),
    protein_colors=None,
    save=SAVE,
):
    if protein_colors is None:
        protein_colors = make_protein_color_map(full_dataset)

    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    sns.set_style("whitegrid")
    fig, axes = plt.subplots(
        1, len(cell_lines),
        figsize=(5.2 * len(cell_lines), 4.2),
        dpi=SAVE_DPI,
        sharex=True,
        sharey=True,   # makes cell types comparable on the same ΔCV scale
        squeeze=True,
    )
    if len(cell_lines) == 1:
        axes = [axes]

    # Build one global legend (proteins across all plotted cell types)
    legend_handles = {}
    all_delta_for_span = []

    for ax, cell_line in zip(axes, cell_lines):
        prots = sorted(list(full_dataset.get(cell_line, {}).keys()))
        plotted_any = False

        for prot in prots:
            y_avg = _avg_reps_curve_on_qgrid(full_dataset, cell_line, prot, x_grid)
            if np.all(np.isnan(y_avg)):
                continue

            y_delta = _delta_cv(y_avg, baseline_n=DELTA_BASELINE_N)
            if np.all(np.isnan(y_delta)):
                continue

            col = protein_colors.get(prot, "0.25")
            h, = ax.plot(x_grid, y_delta, color=col, lw=2.0, alpha=0.95)
            plotted_any = True

            if prot not in legend_handles:
                legend_handles[prot] = h

            all_delta_for_span.append(y_delta)

        ax.set_title(cell_line, fontsize=12, pad=6)
        ax.set_xlim(0.1, 0.9)
        ax.set_xlabel("s quantile (low → high)", fontsize=10)
        ax.grid(True, axis="y", color="0.9", linewidth=1)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

        # Show 0 on y-axis (line)
        #ax.axhline(0, color="0.2", lw=1.2, alpha=0.75, zorder=0)

        if not plotted_any:
            ax.text(0.5, 0.5, "No data", transform=ax.transAxes, ha="center", va="center")
            ax.set_ylabel("ΔCV", fontsize=10)

    axes[0].set_ylabel("ΔCV = CV − CV(min window)", fontsize=10)

    y_all = np.concatenate([yy[np.isfinite(yy)] for yy in all_delta_for_span if np.any(np.isfinite(yy))])
    ymin, ymax = float(np.min(y_all)), float(np.max(y_all))

    ticks = _equal_spaced_ticks_that_include_zero(ymin, ymax, CV_N_TICKS)
    for ax in axes:
        _apply_fixed_ticks(ax, ticks, decimals=DELTA_DECIMALS)

    fig.suptitle("All proteins | avg(rep1,rep2) | ΔCV per cell type", y=1.02, fontsize=13)

    # Nice global legend (same protein colors across cell types)
    prot_names = sorted(legend_handles.keys())
    handles = [legend_handles[p] for p in prot_names]
    ncol = min(6, max(1, int(np.ceil(len(prot_names) / 8))))
    fig.legend(
        handles, prot_names,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.02),
        ncol=ncol,
        frameon=True,
        fontsize=9,
        title="Protein",
    )

    fig.tight_layout(rect=[0, 0.07, 1, 1])

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out_path = os.path.join(SAVE_FOLDER, "ALL_celltypes_allProteins_avgRep_deltaCV_1x3.png")
        fig.savefig(out_path, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out_path}")

    plt.show()

def _spearman_rho(x, y, *, min_points=SPEARMAN_MIN_POINTS):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    n = int(mask.sum())
    if n < min_points:
        return np.nan, np.nan, n
    rho, p = spearmanr(x[mask], y[mask])  # scipy handles ranking internally
    rho = float(rho) if np.isfinite(rho) else np.nan
    p = float(p) if np.isfinite(p) else np.nan
    return rho, p, n


def _fmt_rho(rho):
    return "NA" if not np.isfinite(rho) else f"{rho:.{SPEARMAN_DECIMALS}f}"

def _apply_y_ticks(ax, *, n_ticks=CV_N_TICKS, decimals=CV_DECIMALS):
    ax.yaxis.set_major_locator(mticker.LinearLocator(n_ticks))

    eps = 0.5 * 10 ** (-decimals)  # avoid showing "-0.00"
    ax.yaxis.set_major_formatter(
        mticker.FuncFormatter(
            lambda v, pos: f"{0.0:.{decimals}f}" if abs(v) < eps else f"{v:.{decimals}f}"
        )
    )


def _set_fixed_yspan(ax, y, *, span=CV_YSPAN, ensure_visible=CV_ENSURE_DATA_VISIBLE, clamp_min=CV_CLAMP_MIN):
    """Set ax y-limits so (ymax - ymin) == span (or >= span if ensure_visible)."""
    y = np.asarray(y, dtype=float)
    finite = y[np.isfinite(y)]
    if finite.size == 0:
        return

    y_min = float(finite.min())
    y_max = float(finite.max())
    data_span = y_max - y_min

    span_use = max(float(span), data_span) if ensure_visible else float(span)
    center = 0.5 * (y_min + y_max)

    lo = center - 0.5 * span_use
    hi = center + 0.5 * span_use

    if clamp_min is not None and lo < clamp_min:
        shift = clamp_min - lo
        lo += shift
        hi += shift

    ax.set_ylim(lo, hi)


def _linear_fit_mse(x, y):
    """Least-squares linear fit y = m*x + b. Returns (m, b, mse, yhat_full)."""
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return None

    xm = x[mask]
    ym = y[mask]

    A = np.column_stack([xm, np.ones_like(xm)])
    m, b = np.linalg.lstsq(A, ym, rcond=None)[0]

    yhat_m = m * xm + b
    mse = float(np.mean((ym - yhat_m) ** 2))

    yhat_full = m * x + b
    return float(m), float(b), mse, yhat_full


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []
    reps = (rep,) if rep is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")

    for rep_ in reps:
        for well_ in wells:
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep_, {}).get(well_, {})
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(
    df,
    *,
    k_col=K_COL,
    y_col=Y_COL,
    window_frac=WINDOW_FRAC,
    step_frac=STEP_FRAC,
    window_n=WINDOW_N,
    step_n=STEP_N,
    min_window_n=MIN_WINDOW_N,
):
    """CV(y)=std(y)/mean(y) over sliding windows along sorted k.
    Returns columns: q_center (0..1), k_median, cv, n_window
    """
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))

    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    denom = max(1, n - 1)

    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy(dtype=float)

        mu = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu

        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[k_col].to_numpy(dtype=float)))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="cv"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]
    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def _avg_two_wells_on_grid(y_w1, y_w2):
    """Pointwise average, only where both wells finite (no warnings)."""
    y_w1 = np.asarray(y_w1, dtype=float)
    y_w2 = np.asarray(y_w2, dtype=float)

    avg = np.full_like(y_w1, np.nan, dtype=float)
    mask = np.isfinite(y_w1) & np.isfinite(y_w2)
    avg[mask] = 0.5 * (y_w1[mask] + y_w2[mask])
    return avg


def _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid):
    df_w1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
    df_w2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")

    c1 = _sliding_k_quantile_cv(df_w1)
    c2 = _sliding_k_quantile_cv(df_w2)

    y1 = _interp_to_grid(c1, x_grid)
    y2 = _interp_to_grid(c2, x_grid)

    return _avg_two_wells_on_grid(y1, y2)

# ---------- All-proteins overlay plots (avg rep1+rep2) + two normalizations ----------

def make_protein_color_map(full_dataset, *, palette="husl"):
    """Consistent protein colors across all cell types."""
    all_prots = sorted({prot for cl in full_dataset for prot in full_dataset.get(cl, {}).keys()})
    cols = sns.color_palette(palette, n_colors=max(1, len(all_prots)))
    return {p: cols[i] for i, p in enumerate(all_prots)}


def _avg_two_arrays_strict(y1, y2):
    """Pointwise average ONLY where both finite; else NaN."""
    y1 = np.asarray(y1, dtype=float)
    y2 = np.asarray(y2, dtype=float)
    out = np.full_like(y1, np.nan, dtype=float)
    m = np.isfinite(y1) & np.isfinite(y2)
    out[m] = 0.5 * (y1[m] + y2[m])
    return out


def _avg_reps_curve_on_qgrid(full_dataset, cell_line, protein, x_grid):
    """Avg(rep1, rep2) on the same quantile grid; strict (needs both reps finite)."""
    y1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)
    return _avg_two_arrays_strict(y1, y2)


def _cv_at_min_window(y):
    """CV at the minimum-k/s sliding window ≈ first finite y on the (increasing) x grid."""
    y = np.asarray(y, dtype=float)
    m = np.isfinite(y)
    return float(y[m][0]) if m.any() else np.nan


def _normalize_cv(y, *, mode):
    """
    mode:
      - "delta": ΔCV = CV - CV0
      - "ratio": CV / CV0
    """
    y = np.asarray(y, dtype=float)
    cv0 = _cv_at_min_window(y)
    if not np.isfinite(cv0):
        return np.full_like(y, np.nan, dtype=float)

    if mode == "delta":
        return y - cv0

    if mode == "ratio":
        if cv0 == 0:
            return np.full_like(y, np.nan, dtype=float)
        return y / cv0

    raise ValueError(f"Unknown normalization mode: {mode!r}")

def _equal_spaced_ticks_that_include_zero(ymin, ymax, n_ticks):
    ymin = float(min(ymin, 0.0))
    ymax = float(max(ymax, 0.0))
    n_ticks = int(max(2, n_ticks))

    # choose which tick index will be 0 (j), and step size so range covers [ymin, ymax]
    best = None  # (span, step, j)
    for j in range(n_ticks):  # j = index where tick == 0
        left = j
        right = (n_ticks - 1 - j)

        if left == 0 and ymin < 0:
            continue
        if right == 0 and ymax > 0:
            continue

        step = 1e-12
        if left > 0:
            step = max(step, (-ymin) / left)
        if right > 0:
            step = max(step, (ymax) / right)

        span = step * (n_ticks - 1)
        if best is None or span < best[0]:
            best = (span, step, j)

    if best is None:
        # fallback (should be rare)
        return np.linspace(ymin, ymax, n_ticks)

    _, step, j = best
    lo = -j * step
    ticks = lo + step * np.arange(n_ticks)
    return ticks


def _apply_fixed_ticks(ax, ticks, *, decimals=CV_DECIMALS):
    ticks = np.asarray(ticks, dtype=float)
    ax.set_ylim(float(ticks[0]), float(ticks[-1]))
    ax.yaxis.set_major_locator(mticker.FixedLocator(ticks))

    eps = 0.5 * 10 ** (-decimals)  # avoid "-0.00"
    ax.yaxis.set_major_formatter(
        mticker.FuncFormatter(
            lambda v, pos: f"{0.0:.{decimals}f}" if abs(v) < eps else f"{v:.{decimals}f}"
        )
    )

def plot_celltype_allproteins_avgRep_norm_compare(
    full_dataset,
    cell_line,
    *,
    protein_colors,
    save=SAVE,
):
    """
    One FIGURE per cell type with two PANELS:
      left:  ΔCV = CV - CV(min window)
      right: CV / CV(min window)

    X-axis is the same as your current code: q_center grid in [0, 1].
    (So 'min window' means the lowest-s quantile window.)
    """
    proteins = sorted(list(full_dataset.get(cell_line, {}).keys()))
    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    sns.set_style("whitegrid")
    fig, (ax_delta, ax_ratio) = plt.subplots(
        1, 2,
        figsize=(12.5, 4.2),
        dpi=SAVE_DPI,
        sharex=True,
        sharey=False,
        squeeze=True,
    )

    plotted_any = False
    y_delta_all = []
    y_ratio_all = []
    legend_handles = []
    legend_labels = []

    for prot in proteins:
        y_avg = _avg_reps_curve_on_qgrid(full_dataset, cell_line, prot, x_grid)
        if np.all(np.isnan(y_avg)):
            continue

        col = protein_colors.get(prot, "0.25")

        y_delta = _normalize_cv(y_avg, mode="delta")
        y_ratio = _normalize_cv(y_avg, mode="ratio")

        h, = ax_delta.plot(x_grid, y_delta, color=col, lw=2.0, alpha=0.95, label=prot)
        ax_ratio.plot(x_grid, y_ratio, color=col, lw=2.0, alpha=0.95)

        legend_handles.append(h)
        legend_labels.append(prot)

        y_delta_all.append(y_delta)
        y_ratio_all.append(y_ratio)
        plotted_any = True

    if not plotted_any:
        plt.close(fig)
        print(f"No data to plot for {cell_line}")
        return

    # Formatting
    ax_delta.set_title("ΔCV = CV − CV(min window)", fontsize=11)
    ax_ratio.set_title("CV / CV(min window)", fontsize=11)

    for ax in (ax_delta, ax_ratio):
        ax.set_xlim(0, 1)
        ax.set_xlabel("s quantile (low → high)", fontsize=10)
        ax.grid(True, axis="y", color="0.9", linewidth=1)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

    ax_delta.set_ylabel("Normalized CV", fontsize=10)
    ax_ratio.set_ylabel("")

    # Keep your fixed-span + tick controls (span grows if needed because CV_ENSURE_DATA_VISIBLE=True)
    if len(y_delta_all):
        y = np.concatenate([yy[np.isfinite(yy)] for yy in y_delta_all if np.any(np.isfinite(yy))])
        _set_fixed_yspan(ax_delta, y, span=DELTA_YSPAN, ensure_visible=True, clamp_min=None)
        _apply_y_ticks(ax_delta, decimals=DELTA_DECIMALS)

    if len(y_ratio_all):
        y = np.concatenate([yy[np.isfinite(yy)] for yy in y_ratio_all if np.any(np.isfinite(yy))])
        _set_fixed_yspan(ax_ratio, y, span=RATIO_YSPAN, ensure_visible=True, clamp_min=None)
        _apply_y_ticks(ax_ratio)  # keeps CV_DECIMALS

    fig.suptitle(f"{cell_line} | CV vs {K_COL} (avg wells, then avg rep1+rep2)", y=1.02, fontsize=13)

    # Figure-level legend (same colors across cell types)
    # (Dynamic columns so it stays readable with many proteins.)
    n = len(legend_labels)
    ncol = min(6, max(1, int(np.ceil(n / 8))))
    fig.legend(
        legend_handles, legend_labels,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.02),
        ncol=ncol,
        frameon=True,
        fontsize=9,
        title="Protein",
    )

    fig.tight_layout(rect=[0, 0.06, 1, 1])

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out_path = os.path.join(SAVE_FOLDER, f"{cell_line}_allProteins_avgRep_normCompare.png")
        fig.savefig(out_path, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out_path}")

    plt.show()


def plot_sliding_avg_only_ax(full_dataset, cell_line, protein, *, ax, rep_colors):
    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    y_rep1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y_rep2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)

    # right axis for rep2
    ax2 = ax.twinx()
    ax2.grid(False)

    plotted1 = False
    plotted2 = False

    if not np.all(np.isnan(y_rep1)):
        ax.plot(x_grid, y_rep1, color=rep_colors["rep1"], lw=2.4, label="Biological replicate 1")
        plotted1 = True

    if not np.all(np.isnan(y_rep2)):
        ax2.plot(x_grid, y_rep2, color=rep_colors["rep2"], lw=2.4, label="Biological replicate 2")
        plotted2 = True

    if not (plotted1 or plotted2):
        ax.axis("off")
        ax2.axis("off")
        return None

    # --- Spearman vs plotted x-axis (x_grid is what you plot on x) ---
    rho1 = rho2 = np.nan
    if SHOW_SPEARMAN_TEXT and plotted1:
        rho1, p1, n1 = _spearman_rho(x_grid, y_rep1)
    if SHOW_SPEARMAN_TEXT and plotted2:
        rho2, p2, n2 = _spearman_rho(x_grid, y_rep2)

    # --- Linear fits + MSE (per replicate) ---
    fit1 = _linear_fit_mse(x_grid, y_rep1) if (SHOW_LINEAR_FIT and plotted1) else None
    fit2 = _linear_fit_mse(x_grid, y_rep2) if (SHOW_LINEAR_FIT and plotted2) else None

    yhat1 = yhat2 = None
    m1 = b1 = mse1 = None
    m2 = b2 = mse2 = None

    if fit1 is not None:
        m1, b1, mse1, yhat1 = fit1
        mask1 = np.isfinite(y_rep1) if FIT_PLOT_ONLY_WHERE_DATA else np.isfinite(yhat1)
        ax.plot(
            x_grid[mask1], yhat1[mask1],
            color=rep_colors["rep1"], ls=FIT_LS, lw=FIT_LW, alpha=FIT_ALPHA
        )

    if fit2 is not None:
        m2, b2, mse2, yhat2 = fit2
        mask2 = np.isfinite(y_rep2) if FIT_PLOT_ONLY_WHERE_DATA else np.isfinite(yhat2)
        ax2.plot(
            x_grid[mask2], yhat2[mask2],
            color=rep_colors["rep2"], ls=FIT_LS, lw=FIT_LW, alpha=FIT_ALPHA
        )

    # --- Annotations (slope/MSE + rho) ---
    if plotted1 and (fit1 is not None or SHOW_SPEARMAN_TEXT):
        lines = []
        if fit1 is not None:
            lines.append(f"rep1 m={m1:.{FIT_DECIMALS_SLOPE}f}")
            lines.append(f"MSE={mse1:.{FIT_SIGFIGS_MSE}g}")
        if SHOW_SPEARMAN_TEXT:
            lines.append(f"ρ={_fmt_rho(rho1)}")
        ax.text(
            0.02, 0.98, "\n".join(lines),
            transform=ax.transAxes,
            ha="left", va="top",
            fontsize=FIT_FONT_SIZE,
            color=rep_colors["rep1"],
        )

    if plotted2 and (fit2 is not None or SHOW_SPEARMAN_TEXT):
        lines = []
        if fit2 is not None:
            lines.append(f"rep2 m={m2:.{FIT_DECIMALS_SLOPE}f}")
            lines.append(f"MSE={mse2:.{FIT_SIGFIGS_MSE}g}")
        if SHOW_SPEARMAN_TEXT:
            lines.append(f"ρ={_fmt_rho(rho2)}")
        ax.text(
            0.98, 0.98, "\n".join(lines),
            transform=ax.transAxes,
            ha="right", va="top",
            fontsize=FIT_FONT_SIZE,
            color=rep_colors["rep2"],
        )

    # --- Fixed y-span + ticks (include fit values in span if present) ---
    if plotted1:
        if yhat1 is None:
            y_for_span1 = y_rep1
        else:
            mask1 = np.isfinite(y_rep1) if FIT_PLOT_ONLY_WHERE_DATA else np.isfinite(yhat1)
            y_for_span1 = np.concatenate([y_rep1[mask1], yhat1[mask1]])

    if plotted2:
        if yhat2 is None:
            y_for_span2 = y_rep2
        else:
            mask2 = np.isfinite(y_rep2) if FIT_PLOT_ONLY_WHERE_DATA else np.isfinite(yhat2)
            y_for_span2 = np.concatenate([y_rep2[mask2], yhat2[mask2]])

    ax.set_title(f"{protein}", fontsize=12, pad=6)
    ax.set_xlim(0.1, 0.9)

    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax.spines["top"].set_visible(False)
    ax2.spines["top"].set_visible(False)

    # clean overlap spines
    ax.spines["right"].set_visible(False)
    ax2.spines["left"].set_visible(False)

    # color-code y axes to match reps
    if plotted1:
        ax.spines["left"].set_color(rep_colors["rep1"])
        ax.tick_params(axis="y", colors=rep_colors["rep1"])
    else:
        ax.tick_params(axis="y", left=False, labelleft=False)
        ax.spines["left"].set_visible(False)

    if plotted2:
        ax2.spines["right"].set_color(rep_colors["rep2"])
        ax2.tick_params(axis="y", colors=rep_colors["rep2"])
    else:
        ax2.tick_params(axis="y", right=False, labelright=False)
        ax2.spines["right"].set_visible(False)

    return ax2


def plot_celltype_sliding_fig(full_dataset, cell_line, *, nrows, ncols, save=SAVE):
    proteins = list(full_dataset.get(cell_line, {}).keys())
    nslots = nrows * ncols
    proteins_to_plot = proteins[:nslots]

    sns.set_style("whitegrid")
    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(4.7 * ncols, 3.5 * nrows),
        dpi=SAVE_DPI,
        sharex=False,
        sharey=False,
        squeeze=False,
    )

    pal = sns.color_palette("colorblind", 2)
    rep_colors = {"rep1": pal[0], "rep2": pal[1]}

    axes2 = [[None for _ in range(ncols)] for _ in range(nrows)]

    for i, ax in enumerate(axes.ravel()):
        r, c = divmod(i, ncols)
        if i < len(proteins_to_plot):
            prot = proteins_to_plot[i]
            ax2 = plot_sliding_avg_only_ax(full_dataset, cell_line, prot, ax=ax, rep_colors=rep_colors)
            axes2[r][c] = ax2
        else:
            ax.axis("off")
            axes2[r][c] = None

    # Labels
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            ax2 = axes2[r][c]

            if not ax.axison:
                continue

            ax.set_xlabel("s", fontsize=10)

            if c == 0:
                ax.set_ylabel("CV", color="black")
            else:
                ax.set_ylabel("")

            if ax2 is not None:
                ax2.tick_params(labelright=True)
                if c == ncols - 1:
                    ax2.set_ylabel("", color=rep_colors["rep2"])
                else:
                    ax2.set_ylabel("")

    legend_handles = [
        Line2D([0], [0], color=rep_colors["rep1"], lw=2.4, label="Biological replicate 1"),
        Line2D([0], [0], color=rep_colors["rep2"], lw=2.4, label="Biological replicate 2"),
    ]

    # Legend placement
    empty_axes = []
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            ax2 = axes2[r][c]
            if not ax.axison:
                continue
            has_any_data = ax.has_data() or (ax2 is not None and ax2.has_data())
            if not has_any_data:
                empty_axes.append(ax)

    if len(empty_axes) > 0:
        ax_leg = empty_axes[0]
        ax_leg.axis("off")
        ax_leg.legend(
            handles=legend_handles,
            loc="center",
            frameon=True,
            fontsize=9,
            title=None,
        )
    else:
        fig.legend(
            handles=legend_handles,
            loc="upper right",
            frameon=True,
            fontsize=9,
            title=None,
        )

    fig.suptitle(
        f"{cell_line} | Protein CV vs Synthesis rate (s)",
        y=0.995,
        fontsize=14,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.96], pad=1.0, w_pad=0.8, h_pad=0.9)
    fig.subplots_adjust(wspace=0.35, hspace=0.35)

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out_path = os.path.join(
            SAVE_FOLDER,
            f"{cell_line}_sliding_CV_vs_kquantile_avgWells_only_dualAxis.png",
        )
        fig.savefig(out_path, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out_path}")

    plt.show()

'''
# --- Run: one figure per cell type with requested layouts ---
plot_celltype_sliding_fig(full_dataset, "3T3",  nrows=2, ncols=3, save=True)
plot_celltype_sliding_fig(full_dataset, "mESCs", nrows=2, ncols=3, save=True)
plot_celltype_sliding_fig(full_dataset, "hESCs", nrows=2, ncols=2, save=True)

PROTEIN_COLORS = make_protein_color_map(full_dataset)

plot_celltype_allproteins_avgRep_norm_compare(full_dataset, "3T3",  protein_colors=PROTEIN_COLORS, save=True)
plot_celltype_allproteins_avgRep_norm_compare(full_dataset, "mESCs", protein_colors=PROTEIN_COLORS, save=True)
plot_celltype_allproteins_avgRep_norm_compare(full_dataset, "hESCs", protein_colors=PROTEIN_COLORS, save=True)
'''

PROTEIN_COLORS = make_protein_color_map(full_dataset)
plot_all_celltypes_allproteins_deltaCV(full_dataset, protein_colors=PROTEIN_COLORS, save=True)

# Heatmaps stay as you already have them:
plot_spearman_heatmaps(full_dataset)

def _rho_for_heatmap(full_dataset, cell_line, protein, rep, *, x_mode):
    if x_mode == "q_center":
        x_grid = np.linspace(0.0, 1.0, X_GRID_N)
        y = _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid)
        rho, p, n = _spearman_rho(x_grid, y)
        return rho

    if x_mode == "k_median":
        # Use sliding-window points (combined across well1+well2) and correlate CV vs k_median (actual s)
        df_w1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
        df_w2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")
        c1 = _sliding_k_quantile_cv(df_w1)
        c2 = _sliding_k_quantile_cv(df_w2)
        comb = pd.concat([c1[["k_median", "cv"]], c2[["k_median", "cv"]]], ignore_index=True)
        rho, p, n = _spearman_rho(comb["k_median"].to_numpy(float), comb["cv"].to_numpy(float))
        return rho

    raise ValueError(f"Unknown x_mode: {x_mode!r}")


def plot_spearman_heatmaps(full_dataset, *, layouts=None, x_mode=SPEARMAN_HEATMAP_X_MODE):
    if layouts is None:
        layouts = {"3T3": (2, 3), "mESCs": (2, 3), "hESCs": (2, 2)}

    for cell_line, (nrows, ncols) in layouts.items():
        proteins = list(full_dataset.get(cell_line, {}).keys())
        proteins = proteins[: (nrows * ncols)]  # match the proteins you plot in the grid

        rho_df = pd.DataFrame(index=proteins, columns=["rep1", "rep2"], dtype=float)
        for prot in proteins:
            for rep in ("rep1", "rep2"):
                rho_df.loc[prot, rep] = _rho_for_heatmap(full_dataset, cell_line, prot, rep, x_mode=x_mode)

        plt.figure(figsize=(3.6, 0.45 * max(1, len(proteins)) + 1.2), dpi=SAVE_DPI)
        sns.heatmap(
            rho_df,
            vmin=-1, vmax=1,
            cmap=SPEARMAN_CMAP,
            annot=SPEARMAN_HEATMAP_ANNOT, fmt=f".{SPEARMAN_DECIMALS}f",
            linewidths=0.5, linecolor="0.9",
            cbar_kws={"label": "Spearman ρ"},
        )
        x_desc = "q_center (0..1 grid)" if x_mode == "q_center" else "k_median (window median s)"
        plt.title(f"{cell_line} | Spearman ρ (CV vs {x_desc})")
        plt.tight_layout()
        plt.show()


plot_spearman_heatmaps(full_dataset)
```

```python
"""
DR variability — ΔCV overlay (all proteins, per cell type) + Spearman heatmaps
================================================================================
Kept only:
  • plot_all_celltypes_allproteins_deltaCV   — 1×3 figure, one panel per cell type
  • plot_spearman_heatmaps                   — single combined figure, one panel per cell type

y-axis: DELTA_YSPAN respected, 0 always included, equal tick spacing (both sides).
Legend: above the axes, never overlapping the data.
Heatmaps: all cell types in one figure.
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from scipy.stats import spearmanr

# ── Matplotlib style ──────────────────────────────────────────────────────────
plt.rcParams.update({
    "font.family":        "DejaVu Sans",
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "axes.grid":          True,
    "axes.grid.axis":     "y",
    "grid.color":         "#e8e8e8",
    "grid.linewidth":     0.9,
    "axes.axisbelow":     True,
    "xtick.direction":    "out",
    "ytick.direction":    "out",
    "xtick.major.size":   4,
    "ytick.major.size":   4,
})

# ── Save ──────────────────────────────────────────────────────────────────────
SAVE        = True
SAVE_FOLDER = r"DR variability continuous plots\DR_vs_K2"
SAVE_DPI    = 300

# ── Sliding-window parameters ─────────────────────────────────────────────────
WINDOW_N    = None
WINDOW_FRAC = 1 / 5
STEP_N      = None
STEP_FRAC   = 0.01
MIN_WINDOW_N = 50
X_GRID_N    = 250

K_COL = "k"
Y_COL = "mean_IF_DR"

# ── ΔCV y-axis ────────────────────────────────────────────────────────────────
DELTA_YSPAN      = 0.045   # total y-range; enlarged automatically if data exceed it
DELTA_DECIMALS   = 3
DELTA_BASELINE_N = 1      # first N finite points used as baseline
CV_N_TICKS       = 5      # tick count (same on every panel)

# ── Spearman / heatmap ────────────────────────────────────────────────────────
SPEARMAN_DECIMALS  = 2
SPEARMAN_MIN_POINTS = 8
SPEARMAN_HEATMAP_X_MODE = "q_center"   # "q_center" or "k_median"
SPEARMAN_CMAP      = "vlag"


# ═══════════════════════════════════════════════════════════════════════════════
# Internal helpers
# ═══════════════════════════════════════════════════════════════════════════════

def _cv_over_cvmin(y):
    y = np.asarray(y, dtype=float)

    finite_idx = np.where(np.isfinite(y))[0]
    if finite_idx.size == 0:
        return np.full_like(y, np.nan)

    first_idx = finite_idx[0]
    baseline = y[first_idx]

    if not np.isfinite(baseline) or baseline == 0:
        return np.full_like(y, np.nan)

    return y / baseline

def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    return x if isinstance(x, pd.DataFrame) else pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []
    reps  = (rep,)  if rep  is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")
    for rep_ in reps:
        for well_ in wells:
            node = (full_dataset
                    .get(cell_line, {})
                    .get(protein, {})
                    .get(rep_, {})
                    .get(well_, {}))
            parts.append(_as_df(node.get("before_threshold", [])))
    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(df):
    out_cols = ["q_center", "k_median", "cv", "n_window"]
    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({K_COL, Y_COL} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[K_COL, Y_COL]].copy()
    work[K_COL] = pd.to_numeric(work[K_COL], errors="coerce")
    work[Y_COL] = pd.to_numeric(work[Y_COL], errors="coerce")
    work = work.dropna(subset=[K_COL, Y_COL]).sort_values(K_COL)

    n = len(work)
    if n < max(2, MIN_WINDOW_N):
        return pd.DataFrame(columns=out_cols)

    window_n = int(max(MIN_WINDOW_N, min(int(round(WINDOW_FRAC * n)) if WINDOW_N is None else WINDOW_N, n)))
    step_n   = int(max(1, int(round(STEP_FRAC * n)) if STEP_N is None else STEP_N))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows  = []
    denom = max(1, n - 1)
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y   = sub[Y_COL].to_numpy(dtype=float)
        mu  = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu
        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[K_COL].to_numpy(dtype=float)))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)



def _interp_to_grid(curve, x_grid):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)
    x = curve["q_center"].to_numpy(dtype=float)
    y = curve["cv"].to_numpy(dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)
    x, y = x[mask], y[mask]
    order = np.argsort(x)
    x, y  = x[order], y[order]
    xu, idx = np.unique(x, return_index=True)
    yu = y[idx]
    return np.interp(x_grid, xu, yu, left=np.nan, right=np.nan) if len(xu) >= 2 else np.full_like(x_grid, np.nan, dtype=float)


def _avg_two_wells(y1, y2):
    y1, y2 = np.asarray(y1, dtype=float), np.asarray(y2, dtype=float)
    out = np.full_like(y1, np.nan)
    m = np.isfinite(y1) & np.isfinite(y2)
    out[m] = 0.5 * (y1[m] + y2[m])
    return out


def _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid):
    df1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
    df2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")
    return _avg_two_wells(
        _interp_to_grid(_sliding_k_quantile_cv(df1), x_grid),
        _interp_to_grid(_sliding_k_quantile_cv(df2), x_grid),
    )


def _avg_reps_curve(full_dataset, cell_line, protein, x_grid):
    y1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)
    out = np.full_like(y1, np.nan)
    m = np.isfinite(y1) & np.isfinite(y2)
    out[m] = 0.5 * (y1[m] + y2[m])
    return out


def _delta_cv(y):
    y = np.asarray(y, dtype=float)
    finite = y[np.isfinite(y)]
    if finite.size == 0:
        return np.full_like(y, np.nan)
    n  = int(max(1, min(DELTA_BASELINE_N, finite.size)))
    cv0 = float(np.median(finite[:n]))
    return y - cv0 if np.isfinite(cv0) else np.full_like(y, np.nan)


# ── Tick helpers ───────────────────────────────────────────────────────────────
'''
def _equal_spaced_ticks_with_zero(ymin, ymax, n_ticks):
    """n_ticks evenly spaced values that include 0; covers [ymin, ymax]."""
    ymin = float(min(ymin, 0.0))
    ymax = float(max(ymax, 0.0))
    n_ticks = int(max(2, n_ticks))

    best = None
    for j in range(n_ticks):
        left  = j
        right = n_ticks - 1 - j
        if left  == 0 and ymin < 0: continue
        if right == 0 and ymax > 0: continue
        step = 1e-12
        if left  > 0: step = max(step, (-ymin) / left)
        if right > 0: step = max(step, ( ymax) / right)
        span = step * (n_ticks - 1)
        if best is None or span < best[0]:
            best = (span, step, j)

    if best is None:
        return np.linspace(ymin, ymax, n_ticks)

    _, step, j = best
    return -j * step + step * np.arange(n_ticks)
'''

def _apply_fixed_ticks(ax, ticks, *, decimals):
    ticks = np.asarray(ticks, dtype=float)
    ax.set_ylim(float(ticks[0]), float(ticks[-1]))
    ax.yaxis.set_major_locator(mticker.FixedLocator(ticks))
    eps = 0.5 * 10 ** (-decimals)
    ax.yaxis.set_major_formatter(
        mticker.FuncFormatter(
            lambda v, _: f"{0.0:.{decimals}f}" if abs(v) < eps else f"{v:.{decimals}f}"
        )
    )


def make_protein_color_map(full_dataset, *, palette="Set2"):
    all_prots = sorted({p for cl in full_dataset for p in full_dataset.get(cl, {})})
    cols = sns.color_palette(palette, n_colors=max(1, len(all_prots)))
    return {p: cols[i] for i, p in enumerate(all_prots)}


# ═══════════════════════════════════════════════════════════════════════════════
# ΔCV — all proteins, 1×3 figure
# ═══════════════════════════════════════════════════════════════════════════════

def plot_all_celltypes_allproteins_deltaCV(
    full_dataset,
    *,
    cell_lines=("3T3", "mESCs", "hESCs"),
    protein_colors=None,
    save=SAVE,
):
    if protein_colors is None:
        protein_colors = make_protein_color_map(full_dataset)

    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    ncols = len(cell_lines)
    fig, axes = plt.subplots(
        1, ncols,
        figsize=(5.0 * ncols, 4.0),
        dpi=SAVE_DPI,
        sharex=True,
        sharey=True,
    )
    if ncols == 1:
        axes = [axes]

    # ── Collect figure-level title + legend objects ───────────────────────────
    fig.suptitle(
        "ΔCV vs Decay rate (k)",
        fontsize=13, fontweight="semibold", y=1.03,
    )

    legend_handles = {}
    all_delta      = []

    for ax, cell_line in zip(axes, cell_lines):
        prots = sorted(full_dataset.get(cell_line, {}).keys())

        for prot in prots:
            y_avg = _avg_reps_curve(full_dataset, cell_line, prot, x_grid)
            if np.all(np.isnan(y_avg)):
                continue
            #y_d = _delta_cv(y_avg)
            y_d = _cv_over_cvmin(y_avg)
            if np.all(np.isnan(y_d)):
                continue

            col = protein_colors.get(prot, "0.3")
            h,  = ax.plot(x_grid, y_d, color=col, lw=1.9, alpha=0.92)
            legend_handles.setdefault(prot, h)
            all_delta.append(y_d)

        # ── zero reference ────────────────────────────────────────────────────
        #ax.axhline(0, color="#555555", lw=0.9, ls="--", alpha=0.6, zorder=0)

        # ── Appearance ────────────────────────────────────────────────────────
        ax.set_title(cell_line, fontsize=12, fontweight="semibold", pad=8)
        ax.set_xlim(0.1, 0.9)
        ax.set_xlabel("k (quantile)", fontsize=9.5)
        ax.tick_params(labelsize=8.5)
        ax.spines["left"].set_linewidth(0.8)
        ax.spines["bottom"].set_linewidth(0.8)

    axes[0].set_ylabel("ΔCV", fontsize=10.5)

    # ── Shared y ticks (sharex=True so limits are coupled) ───────────────────
    if all_delta:
        y_all   = np.concatenate([d[np.isfinite(d)] for d in all_delta if np.any(np.isfinite(d))])
        ymin_d  = float(np.min(y_all))
        ymax_d  = float(np.max(y_all))
        # enforce DELTA_YSPAN: expand centre if data already wider
        ctr     = 0.5 * (ymin_d + ymax_d)
        half    = max(DELTA_YSPAN / 2, (ymax_d - ymin_d) / 2)
        ymin_use = ctr - half
        ymax_use = ctr + half
        '''
        ticks   = _equal_spaced_ticks_with_zero(ymin_use, ymax_use, CV_N_TICKS)
        for ax in axes:
            _apply_fixed_ticks(ax, ticks, decimals=DELTA_DECIMALS)
        '''
    # ── Legend — above figure, horizontal, never overlapping data ─────────────
    prot_names = sorted(legend_handles.keys())
    handles    = [legend_handles[p] for p in prot_names]
    ncol_leg   = min(len(prot_names), 8)
    fig.legend(
        handles, prot_names,
        loc="lower center",
        bbox_to_anchor=(0.5, 1.04),
        ncol=ncol_leg,
        frameon=True,
        framealpha=0.95,
        edgecolor="#cccccc",
        fontsize=8.5,
        title="Protein",
        title_fontsize=9,
        handlelength=1.8,
        columnspacing=1.0,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.98])

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out = os.path.join(SAVE_FOLDER, "ALL_celltypes_allProteins_avgRep_deltaCV_1x3.png")
        fig.savefig(out, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out}")

    plt.show()


# ═══════════════════════════════════════════════════════════════════════════════
# Spearman heatmaps — all cell types in one figure
# ═══════════════════════════════════════════════════════════════════════════════

def _spearman_rho(x, y):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    n    = int(mask.sum())
    if n < SPEARMAN_MIN_POINTS:
        return np.nan
    rho, _ = spearmanr(x[mask], y[mask])
    return float(rho) if np.isfinite(rho) else np.nan


def _rho_for_heatmap(full_dataset, cell_line, protein, rep):
    x_grid = np.linspace(0.0, 1.0, X_GRID_N)
    if SPEARMAN_HEATMAP_X_MODE == "q_center":
        y = _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid)
        return _spearman_rho(x_grid, y)
    # k_median mode
    parts = []
    for well in ("well1", "well2"):
        df = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well=well)
        c  = _sliding_k_quantile_cv(df)
        if not c.empty:
            parts.append(c[["k_median", "cv"]])
    if not parts:
        return np.nan
    comb = pd.concat(parts, ignore_index=True)
    return _spearman_rho(comb["k_median"].to_numpy(float), comb["cv"].to_numpy(float))


def plot_spearman_heatmaps(
    full_dataset,
    *,
    layouts=None,
    x_mode=SPEARMAN_HEATMAP_X_MODE,
    save=SAVE,
):
    if layouts is None:
        layouts = {"3T3": (2, 3), "mESCs": (2, 3), "hESCs": (2, 2)}

    cell_lines = list(layouts.keys())
    ncols      = len(cell_lines)

    # ── Compute ρ tables ──────────────────────────────────────────────────────
    rho_tables = {}
    for cell_line, (nrows, ncols_grid) in layouts.items():
        proteins = list(full_dataset.get(cell_line, {}).keys())[: nrows * ncols_grid]
        rho_df   = pd.DataFrame(index=proteins, columns=["rep1", "rep2"], dtype=float)
        for prot in proteins:
            for rep in ("rep1", "rep2"):
                rho_df.loc[prot, rep] = _rho_for_heatmap(full_dataset, cell_line, prot, rep)
        rho_tables[cell_line] = rho_df

    # ── Figure layout — one column per cell type, height proportional to n prots ──
    max_prots = max(len(df) for df in rho_tables.values())
    fig_h     = max(3.2, 0.52 * max_prots + 1.6)

    fig, axes = plt.subplots(
        1, ncols,
        figsize=(3.2 * ncols, fig_h),
        dpi=SAVE_DPI,
    )
    if ncols == 1:
        axes = [axes]

    x_desc = "quantile" if x_mode == "q_center" else "window-median s"
    fig.suptitle(
        f"Spearman Correlation between CV and Decay rate",
        fontsize=12, fontweight="semibold", y=1.02,
    )

    # shared colorbar axis
    fig.subplots_adjust(wspace=0.45, right=0.88)
    cbar_ax = fig.add_axes([0.91, 0.15, 0.025, 0.7])

    for ax, cell_line in zip(axes, cell_lines):
        rho_df = rho_tables[cell_line]

        im = sns.heatmap(
            rho_df,
            ax=ax,
            vmin=-1, vmax=1,
            cmap=SPEARMAN_CMAP,
            annot=True,
            fmt=f".{SPEARMAN_DECIMALS}f",
            annot_kws={"size": 9},
            linewidths=0.5,
            linecolor="#e0e0e0",
            cbar=True,
            cbar_ax=cbar_ax,
            cbar_kws={"label": "Spearman ρ"},
        )

        ax.set_title(cell_line, fontsize=11, fontweight="semibold", pad=6)
        #ax.set_xlabel("Replicate", fontsize=9)
        #ax.set_ylabel("Protein" if cell_line == cell_lines[0] else "", fontsize=9)
        ax.tick_params(labelsize=8.5)

        # rotate x-tick labels if needed
        ax.set_xticklabels(ax.get_xticklabels(), rotation=0, ha="center")
        ax.set_yticklabels(ax.get_yticklabels(), rotation=0)

    cbar_ax.tick_params(labelsize=8.5)
    cbar_ax.set_ylabel("Spearman ρ", fontsize=9)

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out = os.path.join(SAVE_FOLDER, "ALL_celltypes_spearman_heatmap.png")
        fig.savefig(out, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out}")

    plt.show()


# ═══════════════════════════════════════════════════════════════════════════════
# Entry point — replace `full_dataset` with your actual object
# ═══════════════════════════════════════════════════════════════════════════════

if __name__ == "__main__":
    # full_dataset must already be defined in your environment
    PROTEIN_COLORS = make_protein_color_map(full_dataset)

    plot_all_celltypes_allproteins_deltaCV(
        full_dataset,
        protein_colors=PROTEIN_COLORS,
        save=True,
    )

    plot_spearman_heatmaps(full_dataset, save=True)
```

```python
"""
DR variability — Triple heatmaps (CV, SD, mean) per cell type
==============================================================
Produces 3 figures, one per cell type.
Each figure has 3 side-by-side heatmaps:
  • CV  (coefficient of variation)
  • SD  (standard deviation)
  • Mean intensity

Each heatmap cell shows:
  Line 1 — Spearman ρ  (sliding-window metric vs k-quantile)
  Line 2 — Gain %      (robust: predicted value at q=0.9 vs q=0.1
                         from an OLS linear fit → avoids endpoint noise)

Rows = proteins,  Columns = rep1 / rep2
One figure per cell type.
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from scipy.stats import spearmanr

# ── Matplotlib style ──────────────────────────────────────────────────────────
plt.rcParams.update({
    "font.family":        "DejaVu Sans",
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "axes.grid":          False,
    "xtick.direction":    "out",
    "ytick.direction":    "out",
})

# ── Save ──────────────────────────────────────────────────────────────────────
SAVE        = True
SAVE_FOLDER = r"DR variability continuous plots\DR_vs_K2"
SAVE_DPI    = 300

# ── Sliding-window parameters ─────────────────────────────────────────────────
WINDOW_N     = None
WINDOW_FRAC  = 1 / 5
STEP_N       = None
STEP_FRAC    = 0.01
MIN_WINDOW_N = 50
X_GRID_N     = 250

K_COL = "k"
Y_COL = "mean_IF_DR"

# ── Spearman / heatmap ────────────────────────────────────────────────────────
SPEARMAN_DECIMALS   = 2
SPEARMAN_MIN_POINTS = 8
SPEARMAN_CMAP       = "vlag"

# ── Gain annotation ───────────────────────────────────────────────────────────
# Gain = (predicted_value_at_q_HIGH - predicted_value_at_q_LOW)
#         / abs(predicted_value_at_q_LOW)  × 100
# Using OLS linear fit over [Q_LOW, Q_HIGH] quantile range to suppress noise.
GAIN_Q_LOW  = 0.10   # quantile position treated as "k_min"
GAIN_Q_HIGH = 0.90   # quantile position treated as "k_max"
GAIN_DECIMALS = 1


# ═══════════════════════════════════════════════════════════════════════════════
# Data helpers  (identical to the original code)
# ═══════════════════════════════════════════════════════════════════════════════

def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    return x if isinstance(x, pd.DataFrame) else pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []
    reps  = (rep,)  if rep  is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")
    for rep_ in reps:
        for well_ in wells:
            node = (full_dataset
                    .get(cell_line, {})
                    .get(protein, {})
                    .get(rep_, {})
                    .get(well_, {}))
            parts.append(_as_df(node.get("before_threshold", [])))
    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_window_stats(df):
    """
    Sliding window over k-sorted rows.
    Returns DataFrame with columns:
      q_center, k_median, cv, sd, mean_val, n_window
    """
    out_cols = ["q_center", "k_median", "cv", "sd", "mean_val", "n_window"]
    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({K_COL, Y_COL} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[K_COL, Y_COL]].copy()
    work[K_COL] = pd.to_numeric(work[K_COL], errors="coerce")
    work[Y_COL] = pd.to_numeric(work[Y_COL], errors="coerce")
    work = work.dropna(subset=[K_COL, Y_COL]).sort_values(K_COL)

    n = len(work)
    if n < max(2, MIN_WINDOW_N):
        return pd.DataFrame(columns=out_cols)

    window_n = int(max(MIN_WINDOW_N,
                       min(int(round(WINDOW_FRAC * n))
                           if WINDOW_N is None else WINDOW_N, n)))
    step_n   = int(max(1, int(round(STEP_FRAC * n))
                       if STEP_N is None else STEP_N))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows  = []
    denom = max(1, n - 1)
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start: start + window_n]
        y   = sub[Y_COL].to_numpy(dtype=float)
        mu  = float(np.mean(y))
        sd  = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            cv = sd / mu
        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[K_COL].to_numpy(dtype=float)))
        rows.append((q_center, k_median, cv, sd, mu, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _combine_wells_stats(full_dataset, cell_line, protein, rep):
    """
    Concatenate well1 + well2 raw data, then compute sliding-window stats.
    Returns DataFrame with q_center, k_median, cv, sd, mean_val columns.
    """
    parts = []
    for well in ("well1", "well2"):
        df = _concat_before_threshold(full_dataset, cell_line, protein,
                                      rep=rep, well=well)
        if not df.empty:
            parts.append(df)
    if not parts:
        return pd.DataFrame()
    combined = pd.concat(parts, ignore_index=True)
    return _sliding_window_stats(combined)


# ═══════════════════════════════════════════════════════════════════════════════
# Per-cell statistical helpers
# ═══════════════════════════════════════════════════════════════════════════════

def _spearman_rho(x, y):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    n    = int(mask.sum())
    if n < SPEARMAN_MIN_POINTS:
        return np.nan
    rho, _ = spearmanr(x[mask], y[mask])
    return float(rho) if np.isfinite(rho) else np.nan


def _robust_gain_pct(x, y):
    """
    Fit OLS line to (x, y) → predict at GAIN_Q_LOW and GAIN_Q_HIGH.
    Gain % = (y_high - y_low) / |y_low| * 100
    Returns np.nan if fit is not possible or y_low ≈ 0.
    """
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < SPEARMAN_MIN_POINTS:
        return np.nan
    xm, ym = x[mask], y[mask]
    # OLS
    p = np.polyfit(xm, ym, 1)          # p[0]=slope, p[1]=intercept
    y_low  = np.polyval(p, GAIN_Q_LOW)
    y_high = np.polyval(p, GAIN_Q_HIGH)
    if not (np.isfinite(y_low) and np.isfinite(y_high)):
        return np.nan
    if abs(y_low) < 1e-12:
        return np.nan
    return (y_high - y_low) / abs(y_low) * 100.0


def _metrics_for_cell(full_dataset, cell_line, protein, rep):
    """
    Returns dict with keys: rho_cv, gain_cv, rho_sd, gain_sd, rho_mean, gain_mean
    """
    curve = _combine_wells_stats(full_dataset, cell_line, protein, rep)
    if curve is None or curve.empty:
        return {k: np.nan for k in
                ("rho_cv", "gain_cv", "rho_sd", "gain_sd", "rho_mean", "gain_mean")}

    q  = curve["q_center"].to_numpy(float)
    cv = curve["cv"].to_numpy(float)
    sd = curve["sd"].to_numpy(float)
    mn = curve["mean_val"].to_numpy(float)

    return {
        "rho_cv":    _spearman_rho(q, cv),
        "gain_cv":   _robust_gain_pct(q, cv),
        "rho_sd":    _spearman_rho(q, sd),
        "gain_sd":   _robust_gain_pct(q, sd),
        "rho_mean":  _spearman_rho(q, mn),
        "gain_mean": _robust_gain_pct(q, mn),
    }


# ═══════════════════════════════════════════════════════════════════════════════
# Plotting
# ═══════════════════════════════════════════════════════════════════════════════

def _build_heatmap_matrix(full_dataset, cell_line, metric_key):
    """
    Returns a (n_proteins × 2) DataFrame of floats indexed by protein name.
    metric_key: 'rho_cv' | 'gain_cv' | 'rho_sd' | 'gain_sd' | 'rho_mean' | 'gain_mean'
    """
    proteins = list(full_dataset.get(cell_line, {}).keys())
    df = pd.DataFrame(index=proteins, columns=["rep1", "rep2"], dtype=float)
    for prot in proteins:
        for rep in ("rep1", "rep2"):
            m = _metrics_for_cell(full_dataset, cell_line, prot, rep)
            df.loc[prot, rep] = m[metric_key]
    return df


def _annot_matrix(rho_df, gain_df):
    """
    Build a string annotation matrix:
      ρ = 0.XX
      +YY.Y %     (or −YY.Y %)
    """
    annot = pd.DataFrame("", index=rho_df.index, columns=rho_df.columns)
    for prot in rho_df.index:
        for rep in rho_df.columns:
            rho  = rho_df.loc[prot, rep]
            gain = gain_df.loc[prot, rep]
            rho_str  = f"ρ={rho:+.{SPEARMAN_DECIMALS}f}"  if np.isfinite(rho)  else "ρ=n/a"
            gain_str = (f"{gain:+.{GAIN_DECIMALS}f}%"      if np.isfinite(gain) else "n/a")
            annot.loc[prot, rep] = f"{rho_str}\n{gain_str}"
    return annot


def plot_triple_heatmaps_per_celltype(
    full_dataset,
    *,
    cell_lines=("3T3", "mESCs", "hESCs"),
    save=SAVE,
):
    """
    One figure per cell type.
    Each figure: 3 heatmaps side by side — CV | SD | Mean.
    Each cell: Spearman ρ (top) + robust gain % (bottom).
    """
    metrics_info = [
        ("cv",   "rho_cv",   "gain_cv",   "CV"),
        ("sd",   "rho_sd",   "gain_sd",   "SD"),
        ("mean", "rho_mean", "gain_mean", "Mean intensity"),
    ]

    for cell_line in cell_lines:
        proteins = list(full_dataset.get(cell_line, {}).keys())
        if not proteins:
            print(f"  [skip] {cell_line}: no proteins found")
            continue

        n_prot = len(proteins)
        fig_h  = max(3.2, 0.65 * n_prot + 1.8)

        fig, axes = plt.subplots(
            1, 3,
            figsize=(10.5, fig_h),
            dpi=SAVE_DPI,
        )

        fig.suptitle(
            f"{cell_line} — Spearman ρ & Gain (CV / SD / Mean) vs Decay rate (k)",
            fontsize=12, fontweight="semibold", y=1.02,
        )

        # shared colorbar axis (right side)
        fig.subplots_adjust(wspace=0.55, right=0.87)
        cbar_ax = fig.add_axes([0.90, 0.15, 0.022, 0.70])

        for col_idx, (metric_key, rho_key, gain_key, title) in enumerate(metrics_info):
            ax = axes[col_idx]

            rho_df  = _build_heatmap_matrix(full_dataset, cell_line, rho_key)
            gain_df = _build_heatmap_matrix(full_dataset, cell_line, gain_key)
            annot   = _annot_matrix(rho_df, gain_df)

            # Draw heatmap (only first panel gets the shared colorbar)
            sns.heatmap(
                rho_df,
                ax=ax,
                vmin=-1, vmax=1,
                cmap=SPEARMAN_CMAP,
                annot=annot,
                fmt="",                         # annot is already a string matrix
                annot_kws={"size": 8, "va": "center"},
                linewidths=0.5,
                linecolor="#e0e0e0",
                cbar=(col_idx == 2),            # colorbar only on last panel
                cbar_ax=cbar_ax if col_idx == 2 else None,
                cbar_kws={"label": "Spearman ρ"} if col_idx == 2 else {},
            )

            ax.set_title(title, fontsize=11, fontweight="semibold", pad=6)
            ax.set_xticklabels(ax.get_xticklabels(), rotation=0, ha="center",
                               fontsize=8.5)
            ax.set_yticklabels(
                ax.get_yticklabels(),
                rotation=0,
                fontsize=8.5,
                # show y-labels only on leftmost panel
            )
            if col_idx > 0:
                ax.set_ylabel("")
            else:
                ax.set_ylabel("Protein", fontsize=9)

        if cbar_ax:
            cbar_ax.tick_params(labelsize=8)
            cbar_ax.set_ylabel("Spearman ρ", fontsize=9)

        fig.tight_layout(rect=[0, 0, 0.89, 0.98])

        if save:
            os.makedirs(SAVE_FOLDER, exist_ok=True)
            fname = f"{cell_line}_CV_SD_mean_spearman_gain_heatmap.png"
            out   = os.path.join(SAVE_FOLDER, fname)
            fig.savefig(out, dpi=SAVE_DPI, bbox_inches="tight")
            print(f"Saved: {out}")

        plt.show()


# ═══════════════════════════════════════════════════════════════════════════════
# Entry point
# ═══════════════════════════════════════════════════════════════════════════════

if __name__ == "__main__":
    # full_dataset must already be defined in your environment
    plot_triple_heatmaps_per_celltype(
        full_dataset,
        cell_lines=("3T3", "mESCs", "hESCs"),
        save=True,
    )
```

```python
"""
DR variability — ΔCV overlay (all proteins, per cell type) + Spearman heatmaps
================================================================================
Kept only:
  • plot_all_celltypes_allproteins_deltaCV   — 1×3 figure, one panel per cell type
  • plot_spearman_heatmaps                   — single combined figure, one panel per cell type

y-axis: DELTA_YSPAN respected, 0 always included, equal tick spacing (both sides).
Legend: above the axes, never overlapping the data.
Heatmaps: all cell types in one figure.
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from scipy.stats import spearmanr

# ── Matplotlib style ──────────────────────────────────────────────────────────
plt.rcParams.update({
    "font.family":        "DejaVu Sans",
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "axes.grid":          True,
    "axes.grid.axis":     "y",
    "grid.color":         "#e8e8e8",
    "grid.linewidth":     0.9,
    "axes.axisbelow":     True,
    "xtick.direction":    "out",
    "ytick.direction":    "out",
    "xtick.major.size":   4,
    "ytick.major.size":   4,
})

# ── Save ──────────────────────────────────────────────────────────────────────
SAVE        = True
SAVE_FOLDER = r"DR variability continuous plots\DR_vs_S2"
SAVE_DPI    = 300

# ── Sliding-window parameters ─────────────────────────────────────────────────
WINDOW_N    = None
WINDOW_FRAC = 1 / 5
STEP_N      = None
STEP_FRAC   = 0.01
MIN_WINDOW_N = 50
X_GRID_N    = 250

K_COL = "s"
Y_COL = "mean_IF_DR"

# ── ΔCV y-axis ────────────────────────────────────────────────────────────────
DELTA_YSPAN      = 0.08   # total y-range; enlarged automatically if data exceed it
DELTA_DECIMALS   = 3
DELTA_BASELINE_N = 1      # first N finite points used as baseline
CV_N_TICKS       = 5      # tick count (same on every panel)

# ── 95 % CI band ─────────────────────────────────────────────────────────────
CI_BAND_ALPHA = 0.18   # fill transparency (0 = invisible, 1 = opaque)

# ── Spearman / heatmap ────────────────────────────────────────────────────────
SPEARMAN_DECIMALS  = 2
SPEARMAN_MIN_POINTS = 8
SPEARMAN_HEATMAP_X_MODE = "q_center"   # "q_center" or "k_median"
SPEARMAN_CMAP      = "vlag"


# ═══════════════════════════════════════════════════════════════════════════════
# Internal helpers
# ═══════════════════════════════════════════════════════════════════════════════

def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    return x if isinstance(x, pd.DataFrame) else pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []
    reps  = (rep,)  if rep  is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")
    for rep_ in reps:
        for well_ in wells:
            node = (full_dataset
                    .get(cell_line, {})
                    .get(protein, {})
                    .get(rep_, {})
                    .get(well_, {}))
            parts.append(_as_df(node.get("before_threshold", [])))
    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(df):
    out_cols = ["q_center", "k_median", "cv", "n_window"]
    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({K_COL, Y_COL} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[K_COL, Y_COL]].copy()
    work[K_COL] = pd.to_numeric(work[K_COL], errors="coerce")
    work[Y_COL] = pd.to_numeric(work[Y_COL], errors="coerce")
    work = work.dropna(subset=[K_COL, Y_COL]).sort_values(K_COL)

    n = len(work)
    if n < max(2, MIN_WINDOW_N):
        return pd.DataFrame(columns=out_cols)

    window_n = int(max(MIN_WINDOW_N, min(int(round(WINDOW_FRAC * n)) if WINDOW_N is None else WINDOW_N, n)))
    step_n   = int(max(1, int(round(STEP_FRAC * n)) if STEP_N is None else STEP_N))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows  = []
    denom = max(1, n - 1)
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y   = sub[Y_COL].to_numpy(dtype=float)
        mu  = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu
        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[K_COL].to_numpy(dtype=float)))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)
    x = curve["q_center"].to_numpy(dtype=float)
    y = curve["cv"].to_numpy(dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)
    x, y = x[mask], y[mask]
    order = np.argsort(x)
    x, y  = x[order], y[order]
    xu, idx = np.unique(x, return_index=True)
    yu = y[idx]
    return np.interp(x_grid, xu, yu, left=np.nan, right=np.nan) if len(xu) >= 2 else np.full_like(x_grid, np.nan, dtype=float)


def _avg_two_wells(y1, y2):
    y1, y2 = np.asarray(y1, dtype=float), np.asarray(y2, dtype=float)
    out = np.full_like(y1, np.nan)
    m = np.isfinite(y1) & np.isfinite(y2)
    out[m] = 0.5 * (y1[m] + y2[m])
    return out


def _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid):
    df1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
    df2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")
    return _avg_two_wells(
        _interp_to_grid(_sliding_k_quantile_cv(df1), x_grid),
        _interp_to_grid(_sliding_k_quantile_cv(df2), x_grid),
    )


def _avg_reps_curve(full_dataset, cell_line, protein, x_grid):
    y1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)
    out = np.full_like(y1, np.nan)
    m = np.isfinite(y1) & np.isfinite(y2)
    out[m] = 0.5 * (y1[m] + y2[m])
    return out


def _rep_ci_band(full_dataset, cell_line, protein, x_grid):
    """95 % CI half-width at each grid point using t(df=1) with 2 replicates.

    With n=2 replicates the SEM = |rep1 - rep2| / sqrt(2), and the 95 % CI
    half-width = t(0.975, df=1) * SEM  ≈  12.706 * |rep1 - rep2| / sqrt(2).

    Returns (y_mean, y_lo, y_hi) on x_grid for the raw CV (before ΔCV shift).
    NaN wherever either replicate is missing.
    """
    t95_df1 = 12.706   # scipy.stats.t.ppf(0.975, df=1)
    y1 = _rep_avg_curve(full_dataset, cell_line, protein, "rep1", x_grid)
    y2 = _rep_avg_curve(full_dataset, cell_line, protein, "rep2", x_grid)
    mean = np.full_like(y1, np.nan)
    lo   = np.full_like(y1, np.nan)
    hi   = np.full_like(y1, np.nan)
    m = np.isfinite(y1) & np.isfinite(y2)
    mean[m] = 0.5 * (y1[m] + y2[m])
    half    = t95_df1 * np.abs(y1[m] - y2[m]) / np.sqrt(2)
    lo[m]   = mean[m] - half
    hi[m]   = mean[m] + half
    return mean, lo, hi


def _delta_cv(y):
    y = np.asarray(y, dtype=float)
    finite = y[np.isfinite(y)]
    if finite.size == 0:
        return np.full_like(y, np.nan)
    n  = int(max(1, min(DELTA_BASELINE_N, finite.size)))
    cv0 = float(np.median(finite[:n]))
    return y - cv0 if np.isfinite(cv0) else np.full_like(y, np.nan)


# ── Tick helpers ───────────────────────────────────────────────────────────────

def _equal_spaced_ticks_with_zero(ymin, ymax, n_ticks):
    """n_ticks evenly spaced values that include 0; covers [ymin, ymax]."""
    ymin = float(min(ymin, 0.0))
    ymax = float(max(ymax, 0.0))
    n_ticks = int(max(2, n_ticks))

    best = None
    for j in range(n_ticks):
        left  = j
        right = n_ticks - 1 - j
        if left  == 0 and ymin < 0: continue
        if right == 0 and ymax > 0: continue
        step = 1e-12
        if left  > 0: step = max(step, (-ymin) / left)
        if right > 0: step = max(step, ( ymax) / right)
        span = step * (n_ticks - 1)
        if best is None or span < best[0]:
            best = (span, step, j)

    if best is None:
        return np.linspace(ymin, ymax, n_ticks)

    _, step, j = best
    return -j * step + step * np.arange(n_ticks)


def _apply_fixed_ticks(ax, ticks, *, decimals):
    ticks = np.asarray(ticks, dtype=float)
    ax.set_ylim(float(ticks[0]), float(ticks[-1]))
    ax.yaxis.set_major_locator(mticker.FixedLocator(ticks))
    eps = 0.5 * 10 ** (-decimals)
    ax.yaxis.set_major_formatter(
        mticker.FuncFormatter(
            lambda v, _: f"{0.0:.{decimals}f}" if abs(v) < eps else f"{v:.{decimals}f}"
        )
    )


def make_protein_color_map(full_dataset, *, palette="husl"):
    all_prots = sorted({p for cl in full_dataset for p in full_dataset.get(cl, {})})
    cols = sns.color_palette(palette, n_colors=max(1, len(all_prots)))
    return {p: cols[i] for i, p in enumerate(all_prots)}


# ═══════════════════════════════════════════════════════════════════════════════
# ΔCV — all proteins, 1×3 figure
# ═══════════════════════════════════════════════════════════════════════════════

def plot_all_celltypes_allproteins_deltaCV(
    full_dataset,
    *,
    cell_lines=("3T3", "mESCs", "hESCs"),
    protein_colors=None,
    save=SAVE,
):
    if protein_colors is None:
        protein_colors = make_protein_color_map(full_dataset)

    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    ncols = len(cell_lines)
    fig, axes = plt.subplots(
        1, ncols,
        figsize=(5.0 * ncols, 4.0),
        dpi=SAVE_DPI,
        sharex=True,
        sharey=True,
    )
    if ncols == 1:
        axes = [axes]

    # ── Collect figure-level title + legend objects ───────────────────────────
    fig.suptitle(
        "ΔCV = CV − CV(baseline)  |  avg(rep1, rep2) per protein",
        fontsize=13, fontweight="semibold", y=1.03,
    )

    legend_handles = {}
    all_delta      = []

    for ax, cell_line in zip(axes, cell_lines):
        prots = sorted(full_dataset.get(cell_line, {}).keys())

        for prot in prots:
            y_avg = _avg_reps_curve(full_dataset, cell_line, prot, x_grid)
            if np.all(np.isnan(y_avg)):
                continue
            y_d = _delta_cv(y_avg)
            if np.all(np.isnan(y_d)):
                continue

            col = protein_colors.get(prot, "0.3")

            # ── 95 % CI band (same ΔCV shift applied to lo/hi) ───────────────
            _, y_lo_raw, y_hi_raw = _rep_ci_band(full_dataset, cell_line, prot, x_grid)
            # reuse the same baseline as the mean curve
            y_finite = y_avg[np.isfinite(y_avg)]
            n_base   = int(max(1, min(DELTA_BASELINE_N, y_finite.size)))
            cv0      = float(np.median(y_finite[:n_base])) if y_finite.size > 0 else np.nan
            if np.isfinite(cv0):
                y_lo_d = y_lo_raw - cv0
                y_hi_d = y_hi_raw - cv0
                ax.fill_between(
                    x_grid, y_lo_d, y_hi_d,
                    color=col, alpha=CI_BAND_ALPHA,
                    linewidth=0, zorder=1,
                )

            h, = ax.plot(x_grid, y_d, color=col, lw=1.9, alpha=0.92, zorder=2)
            legend_handles.setdefault(prot, h)
            all_delta.append(y_d)

        # ── zero reference ────────────────────────────────────────────────────
        #ax.axhline(0, color="#555555", lw=0.9, ls="--", alpha=0.6, zorder=0)

        # ── Appearance ────────────────────────────────────────────────────────
        ax.set_title(cell_line, fontsize=12, fontweight="semibold", pad=8)
        ax.set_xlim(0.1, 0.9)
        ax.set_xlabel("Synthesis rate s — quantile (low → high)", fontsize=9.5)
        ax.tick_params(labelsize=8.5)
        ax.spines["left"].set_linewidth(0.8)
        ax.spines["bottom"].set_linewidth(0.8)

    axes[0].set_ylabel("ΔCV", fontsize=10.5)

    # ── Shared y ticks (sharex=True so limits are coupled) ───────────────────
    if all_delta:
        y_all   = np.concatenate([d[np.isfinite(d)] for d in all_delta if np.any(np.isfinite(d))])
        ymin_d  = float(np.min(y_all))
        ymax_d  = float(np.max(y_all))
        # enforce DELTA_YSPAN: expand centre if data already wider
        ctr     = 0.5 * (ymin_d + ymax_d)
        half    = max(DELTA_YSPAN / 2, (ymax_d - ymin_d) / 2)
        ymin_use = ctr - half
        ymax_use = ctr + half
        ticks   = _equal_spaced_ticks_with_zero(ymin_use, ymax_use, CV_N_TICKS)
        for ax in axes:
            _apply_fixed_ticks(ax, ticks, decimals=DELTA_DECIMALS)

    # ── Legend — above figure, horizontal, never overlapping data ─────────────
    prot_names = sorted(legend_handles.keys())
    handles    = [legend_handles[p] for p in prot_names]
    ncol_leg   = min(len(prot_names), 8)
    fig.legend(
        handles, prot_names,
        loc="lower center",
        bbox_to_anchor=(0.5, 1.04),
        ncol=ncol_leg,
        frameon=True,
        framealpha=0.95,
        edgecolor="#cccccc",
        fontsize=8.5,
        title="Protein",
        title_fontsize=9,
        handlelength=1.8,
        columnspacing=1.0,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.98])

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out = os.path.join(SAVE_FOLDER, "ALL_celltypes_allProteins_avgRep_deltaCV_1x3.png")
        fig.savefig(out, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out}")

    plt.show()


# ═══════════════════════════════════════════════════════════════════════════════
# Spearman heatmaps — all cell types in one figure
# ═══════════════════════════════════════════════════════════════════════════════

def _spearman_rho(x, y):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    n    = int(mask.sum())
    if n < SPEARMAN_MIN_POINTS:
        return np.nan
    rho, _ = spearmanr(x[mask], y[mask])
    return float(rho) if np.isfinite(rho) else np.nan


def _rho_for_heatmap(full_dataset, cell_line, protein, rep):
    x_grid = np.linspace(0.0, 1.0, X_GRID_N)
    if SPEARMAN_HEATMAP_X_MODE == "q_center":
        y = _rep_avg_curve(full_dataset, cell_line, protein, rep, x_grid)
        return _spearman_rho(x_grid, y)
    # k_median mode
    parts = []
    for well in ("well1", "well2"):
        df = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well=well)
        c  = _sliding_k_quantile_cv(df)
        if not c.empty:
            parts.append(c[["k_median", "cv"]])
    if not parts:
        return np.nan
    comb = pd.concat(parts, ignore_index=True)
    return _spearman_rho(comb["k_median"].to_numpy(float), comb["cv"].to_numpy(float))


def plot_spearman_heatmaps(
    full_dataset,
    *,
    layouts=None,
    x_mode=SPEARMAN_HEATMAP_X_MODE,
    save=SAVE,
):
    if layouts is None:
        layouts = {"3T3": (2, 3), "mESCs": (2, 3), "hESCs": (2, 2)}

    cell_lines = list(layouts.keys())
    ncols      = len(cell_lines)

    # ── Compute ρ tables ──────────────────────────────────────────────────────
    rho_tables = {}
    for cell_line, (nrows, ncols_grid) in layouts.items():
        proteins = list(full_dataset.get(cell_line, {}).keys())[: nrows * ncols_grid]
        rho_df   = pd.DataFrame(index=proteins, columns=["rep1", "rep2"], dtype=float)
        for prot in proteins:
            for rep in ("rep1", "rep2"):
                rho_df.loc[prot, rep] = _rho_for_heatmap(full_dataset, cell_line, prot, rep)
        rho_tables[cell_line] = rho_df

    # ── Figure layout — one column per cell type, height proportional to n prots ──
    max_prots = max(len(df) for df in rho_tables.values())
    fig_h     = max(3.2, 0.52 * max_prots + 1.6)

    fig, axes = plt.subplots(
        1, ncols,
        figsize=(3.2 * ncols, fig_h),
        dpi=SAVE_DPI,
    )
    if ncols == 1:
        axes = [axes]

    x_desc = "quantile" if x_mode == "q_center" else "window-median s"
    fig.suptitle(
        f"Spearman ρ  (CV vs {x_desc})",
        fontsize=12, fontweight="semibold", y=1.02,
    )

    # shared colorbar axis
    fig.subplots_adjust(wspace=0.45, right=0.88)
    cbar_ax = fig.add_axes([0.91, 0.15, 0.025, 0.7])

    for ax, cell_line in zip(axes, cell_lines):
        rho_df = rho_tables[cell_line]

        im = sns.heatmap(
            rho_df,
            ax=ax,
            vmin=-1, vmax=1,
            cmap=SPEARMAN_CMAP,
            annot=True,
            fmt=f".{SPEARMAN_DECIMALS}f",
            annot_kws={"size": 9},
            linewidths=0.5,
            linecolor="#e0e0e0",
            cbar=True,
            cbar_ax=cbar_ax,
            cbar_kws={"label": "Spearman ρ"},
        )

        ax.set_title(cell_line, fontsize=11, fontweight="semibold", pad=6)
        ax.set_xlabel("Replicate", fontsize=9)
        ax.set_ylabel("Protein" if cell_line == cell_lines[0] else "", fontsize=9)
        ax.tick_params(labelsize=8.5)

        # rotate x-tick labels if needed
        ax.set_xticklabels(ax.get_xticklabels(), rotation=0, ha="center")
        ax.set_yticklabels(ax.get_yticklabels(), rotation=0)

    cbar_ax.tick_params(labelsize=8.5)
    cbar_ax.set_ylabel("Spearman ρ", fontsize=9)

    if save:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        out = os.path.join(SAVE_FOLDER, "ALL_celltypes_spearman_heatmap.png")
        fig.savefig(out, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out}")

    plt.show()


# ═══════════════════════════════════════════════════════════════════════════════
# Entry point — replace `full_dataset` with your actual object
# ═══════════════════════════════════════════════════════════════════════════════

if __name__ == "__main__":
    # full_dataset must already be defined in your environment
    PROTEIN_COLORS = make_protein_color_map(full_dataset)

    plot_all_celltypes_allproteins_deltaCV(
        full_dataset,
        protein_colors=PROTEIN_COLORS,
        save=True,
    )

    plot_spearman_heatmaps(full_dataset, save=True)
```

```python
# BINNED-k scatter (N groups) + approximation line
# One figure per cell type, same layouts as above.
# - k-range is fixed (equal-width bins in k), so each bin has variable cell count.
# - For each bin: CV(y)=std(y)/mean(y) computed within that k-bin.
# - For each protein subplot: plot ONLY the average between well1 and well2 (separately for rep1 and rep2).

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
from statsmodels.nonparametric.smoothers_lowess import lowess

# --- Controls ---
N_K_BINS = 20          # N groups (bins) along k
MIN_BIN_N = 25         # ignore bins with fewer than this many cells
LOWESS_FRAC = 0.5      # smoothing for approximation line (0..1); larger = smoother

K_COL = "k"
Y_COL = "mean_IF_DR"

SAVE = True
SAVE_FOLDER_BINNED = r"DR variability continuous plots\DR_vs_K_CHXstyle_binnedScatter"
SAVE_DPI = 300


def _binned_k_cv_array(df, edges, *, k_col=K_COL, y_col=Y_COL, min_bin_n=MIN_BIN_N):
    """Return (cv_array, n_array) for fixed k-bin edges."""
    n_bins = len(edges) - 1
    if df is None or df.empty or n_bins <= 0:
        return (np.full(n_bins, np.nan, dtype=float), np.zeros(n_bins, dtype=int))
    if not ({k_col, y_col} <= set(df.columns)):
        return (np.full(n_bins, np.nan, dtype=float), np.zeros(n_bins, dtype=int))

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col])
    if work.empty:
        return (np.full(n_bins, np.nan, dtype=float), np.zeros(n_bins, dtype=int))

    k = work[k_col].to_numpy(dtype=float)
    y = work[y_col].to_numpy(dtype=float)

    cv_out = np.full(n_bins, np.nan, dtype=float)
    n_out = np.zeros(n_bins, dtype=int)

    # Bin assignment; include rightmost edge in last bin
    bin_idx = np.digitize(k, edges, right=False) - 1
    bin_idx[bin_idx == n_bins] = n_bins - 1
    valid = (bin_idx >= 0) & (bin_idx < n_bins)
    bin_idx = bin_idx[valid]
    y = y[valid]

    for i in range(n_bins):
        mask = bin_idx == i
        if not np.any(mask):
            continue
        yi = y[mask]
        n_i = int(yi.size)
        n_out[i] = n_i
        if n_i < max(2, int(min_bin_n)):
            continue
        mu = float(np.mean(yi))
        if not np.isfinite(mu) or mu == 0:
            continue
        sd = float(np.std(yi, ddof=1)) if n_i >= 2 else 0.0
        cv_out[i] = sd / mu

    return cv_out, n_out


def _avg_two_wells(y_w1, y_w2):
    """Pointwise average, only where both wells finite."""
    stack = np.vstack([y_w1, y_w2])
    with np.errstate(invalid="ignore"):
        avg = np.nanmean(stack, axis=0)
    count = np.isfinite(stack).sum(axis=0)
    avg[count < 2] = np.nan
    return avg


def _protein_edges(full_dataset, cell_line, protein, *, n_bins=N_K_BINS, k_col=K_COL):
    """Compute fixed k-range (min..max) for this protein across reps+wells."""
    df_all = _concat_before_threshold(full_dataset, cell_line, protein, rep=None, well=None)
    if df_all is None or df_all.empty or k_col not in df_all.columns:
        return None
    k_all = pd.to_numeric(df_all[k_col], errors="coerce").dropna()
    if k_all.empty:
        return None
    k_min = float(k_all.min())
    k_max = float(k_all.max())
    if not (np.isfinite(k_min) and np.isfinite(k_max)) or k_min == k_max:
        return None
    return np.linspace(k_min, k_max, int(n_bins) + 1)


def _rep_avg_binned_curve(full_dataset, cell_line, protein, rep, edges):
    df_w1 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well1")
    df_w2 = _concat_before_threshold(full_dataset, cell_line, protein, rep=rep, well="well2")

    y1, _n1 = _binned_k_cv_array(df_w1, edges)
    y2, _n2 = _binned_k_cv_array(df_w2, edges)
    return _avg_two_wells(y1, y2)


def _lowess_line(x, y, *, frac=LOWESS_FRAC):
    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 3:
        return None
    xs = x[mask]
    ys = y[mask]
    order = np.argsort(xs)
    xs = xs[order]
    ys = ys[order]
    sm = lowess(ys, xs, frac=float(frac), return_sorted=True)
    return sm[:, 0], sm[:, 1]


def plot_binned_scatter_avg_only_ax(full_dataset, cell_line, protein, *, ax, rep_colors, n_bins=N_K_BINS):
    edges = _protein_edges(full_dataset, cell_line, protein, n_bins=n_bins)
    if edges is None:
        ax.axis("off")
        return None

    x = 0.5 * (edges[:-1] + edges[1:])  # bin centers in k-units
    y_rep1 = _rep_avg_binned_curve(full_dataset, cell_line, protein, "rep1", edges)
    y_rep2 = _rep_avg_binned_curve(full_dataset, cell_line, protein, "rep2", edges)

    any_plotted = False
    if np.isfinite(y_rep1).any():
        ax.scatter(x, y_rep1, s=22, color=rep_colors["rep1"], alpha=0.85, edgecolors="none")
        sm = _lowess_line(x, y_rep1)
        if sm is not None:
            ax.plot(sm[0], sm[1], color=rep_colors["rep1"], lw=2.2)
        any_plotted = True
    if np.isfinite(y_rep2).any():
        ax.scatter(x, y_rep2, s=22, color=rep_colors["rep2"], alpha=0.85, edgecolors="none")
        sm = _lowess_line(x, y_rep2)
        if sm is not None:
            ax.plot(sm[0], sm[1], color=rep_colors["rep2"], lw=2.2)
        any_plotted = True

    if not any_plotted:
        ax.axis("off")
        return None

    ax.set_title(f"{protein}", fontsize=12, pad=6)
    ax.grid(True, axis="y", color="0.9", linewidth=1)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    # y-lims per subplot (small padding)
    v = np.concatenate([y_rep1[np.isfinite(y_rep1)], y_rep2[np.isfinite(y_rep2)]])
    if v.size:
        y_min = float(np.min(v))
        y_max = float(np.max(v))
        rng = y_max - y_min
        pad = 0.05 * rng if rng > 0 else (0.05 * abs(y_max) if y_max != 0 else 0.1)
        return (y_min - pad, y_max + pad)
    return None


def plot_celltype_binned_scatter_fig(full_dataset, cell_line, *, nrows, ncols, save=SAVE, n_bins=N_K_BINS):
    proteins = list(full_dataset.get(cell_line, {}).keys())
    nslots = nrows * ncols
    proteins_to_plot = proteins[:nslots]

    sns.set_style("whitegrid")
    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(4.7 * ncols, 3.5 * nrows),
        dpi=SAVE_DPI,
        sharex=False,
        sharey=False,
        squeeze=False,
    )

    pal = sns.color_palette("colorblind", 2)
    rep_colors = {"rep1": pal[0], "rep2": pal[1]}

    for i, ax in enumerate(axes.ravel()):
        if i < len(proteins_to_plot):
            prot = proteins_to_plot[i]
            plot_binned_scatter_avg_only_ax(full_dataset, cell_line, prot, ax=ax, rep_colors=rep_colors, n_bins=n_bins)
        else:
            ax.axis("off")

    # labels
    for r in range(nrows):
        for c in range(ncols):
            ax = axes[r, c]
            if not ax.axison:
                continue
            if r == nrows - 1:
                ax.set_xlabel("k (bin center)")
            else:
                ax.set_xlabel("")
            if c == 0:
                ax.set_ylabel("CV")
            else:
                ax.set_ylabel("")

    legend_handles = [
        Line2D([0], [0], color=rep_colors["rep1"], lw=2.2, marker="o", markersize=5, label="rep1 avg(well1,well2)"),
        Line2D([0], [0], color=rep_colors["rep2"], lw=2.2, marker="o", markersize=5, label="rep2 avg(well1,well2)"),
    ]

    if cell_line in ("3T3", "mESCs"):
        empty_axes = [ax for ax in axes.ravel() if ax.axison and not ax.has_data()]
        if len(empty_axes) > 0:
            ax_leg = empty_axes[0]
            ax_leg.axis("off")
            ax_leg.legend(handles=legend_handles, loc="center", frameon=True, fontsize=9, title="Averages", title_fontsize=10)
        else:
            fig.legend(handles=legend_handles, loc="upper right", frameon=True, fontsize=9, title="Averages", title_fontsize=10)
    else:
        # hESCs: legend at figure level (upper right)
        fig.legend(handles=legend_handles, loc="upper right", frameon=True, fontsize=9, title="Averages", title_fontsize=10)

    fig.suptitle(
        f"{cell_line} | Binned-k CV scatter (N={int(n_bins)}) + LOWESS (avg wells only)",
        y=0.995,
        fontsize=14,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.96], pad=1.0, w_pad=0.8, h_pad=0.9)
    fig.subplots_adjust(wspace=0.25, hspace=0.35)

    if save:
        os.makedirs(SAVE_FOLDER_BINNED, exist_ok=True)
        out_path = os.path.join(
            SAVE_FOLDER_BINNED,
            f"{cell_line}_binnedScatter_CV_vs_k_avgWells_only_N{int(n_bins)}.png",
        )
        fig.savefig(out_path, dpi=SAVE_DPI, bbox_inches="tight")
        print(f"Saved: {out_path}")

    plt.show()


# --- Run: one figure per cell type with requested layouts ---
plot_celltype_binned_scatter_fig(full_dataset, "3T3",  nrows=2, ncols=3, save=True, n_bins=N_K_BINS)
plot_celltype_binned_scatter_fig(full_dataset, "mESCs", nrows=2, ncols=3, save=True, n_bins=N_K_BINS)
plot_celltype_binned_scatter_fig(full_dataset, "hESCs", nrows=2, ncols=2, save=True, n_bins=N_K_BINS)
```

```python
# SLIDING (ALL PROTS COMBINED) AREA VS k

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------
# Controls
# -----------------------
# Choose ONE window definition:
WINDOW_N = None        # e.g. 1000  -> fixed number of cells per window
WINDOW_FRAC = 1 / 6    # e.g. 0.15  -> fixed fraction of dataset per window (ignored if WINDOW_N is set)

# How far to slide the window each step:
STEP_N = None          # e.g. 100   -> fixed step in cells (ignored if None)
STEP_FRAC = 0.01       # e.g. 0.02  -> step as fraction of dataset (ignored if STEP_N is set)

MIN_WINDOW_N = 50
X_GRID_N = 300

# Column names in your pooled before_threshold DataFrames
X_COL = "area_TL"      # column to sort by (x-quantiles)
Y_COL = "k"            # column to compute statistic of within each window

# Which statistic to plot over windows: "CV", "STD", or "MEAN"
STAT = "MEAN"

# If STAT == "MEAN" and STD_STATE == True, shade +/- STD around the mean curve
STD_STATE = True

# Saving (optional)
SAVE_PLOTS = True
SAVE_FOLDER = r"DR variability continuous plots\ALL_PROTEINS_POOLED"


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold_all_proteins(full_dataset, cell_line, *, rep):
    """Pool before_threshold across ALL proteins and BOTH wells for a given cell_line + rep."""
    parts = []
    proteins = list(full_dataset.get(cell_line, {}).keys())
    for protein in proteins:
        for well in ("well1", "well2"):
            node = (
                full_dataset
                .get(cell_line, {})
                .get(protein, {})
                .get(rep, {})
                .get(well, {})
            )
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _pretty_label(col_name: str) -> str:
    """Human-friendly axis labels; extend this mapping if you want."""
    mapping = {
        "area_TL": "Area",
        "k": "k",
        "mean_IF_DR": "mean IF (DR)",
        "mean_DR": "mean DR",
        "DR": "DR",
        "s": "s",
    }
    return mapping.get(col_name, col_name)


def _sliding_quantile_stat(
    df,
    *,
    x_col,
    y_col,
    stat="CV",
    window_frac=0.20,
    step_frac=0.01,
    window_n=None,
    step_n=None,
    min_window_n=50,
):
    """
    Compute a sliding-window statistic of y_col over windows along sorted x_col.

    Output columns:
    - q_center: window center in quantile space (0..1)
    - x_median: median x in the window
    - value: requested statistic (CV/STD/MEAN) of y in the window
    - n_window: number of points in the window
    """
    stat = str(stat).upper().strip()
    out_cols = ["q_center", "x_median", "value", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({x_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[x_col, y_col]].copy()
    work[x_col] = pd.to_numeric(work[x_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[x_col, y_col]).sort_values(x_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))

    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    x_vals = work[x_col].to_numpy(dtype=float)
    y_vals = work[y_col].to_numpy(dtype=float)

    rows = []
    denom = max(1, n - 1)

    for start in range(0, n - window_n + 1, step_n):
        end = start + window_n
        x_sub = x_vals[start:end]
        y_sub = y_vals[start:end]

        mu = float(np.mean(y_sub))
        sd = float(np.std(y_sub, ddof=1)) if len(y_sub) >= 2 else 0.0

        if stat == "MEAN":
            value = mu
        elif stat == "STD":
            value = sd
        elif stat == "CV":
            if (not np.isfinite(mu)) or mu == 0.0:
                value = np.nan
            else:
                value = sd / abs(mu)
        else:
            raise ValueError("STAT must be one of: 'CV', 'STD', 'MEAN'")

        q_center = (start + 0.5 * (window_n - 1)) / denom
        x_median = float(np.median(x_sub))
        rows.append((q_center, x_median, value, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="value"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]
    order = np.argsort(x)
    x = x[order]
    y = y[order]

    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]
    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def plot_sliding_stat_reps_all_proteins(
    cell_line,
    *,
    x_col,
    y_col,
    stat,
    std_state=False,
    window_frac=0.20,
    step_frac=0.01,
    window_n=None,
    step_n=None,
):
    df_rep1 = _concat_before_threshold_all_proteins(full_dataset, cell_line, rep="rep1")
    df_rep2 = _concat_before_threshold_all_proteins(full_dataset, cell_line, rep="rep2")

    stat = str(stat).upper().strip()
    do_shade = (stat == "MEAN") and bool(std_state)

    # Main curves (requested stat)
    curve_rep1 = _sliding_quantile_stat(
        df_rep1,
        x_col=x_col,
        y_col=y_col,
        stat=stat,
        window_frac=window_frac,
        step_frac=step_frac,
        window_n=window_n,
        step_n=step_n,
        min_window_n=MIN_WINDOW_N,
    )
    curve_rep2 = _sliding_quantile_stat(
        df_rep2,
        x_col=x_col,
        y_col=y_col,
        stat=stat,
        window_frac=window_frac,
        step_frac=step_frac,
        window_n=window_n,
        step_n=step_n,
        min_window_n=MIN_WINDOW_N,
    )

    # Optional STD curves for shading (only when plotting MEAN)
    curve_rep1_std = pd.DataFrame()
    curve_rep2_std = pd.DataFrame()
    if do_shade:
        curve_rep1_std = _sliding_quantile_stat(
            df_rep1,
            x_col=x_col,
            y_col=y_col,
            stat="STD",
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
        )
        curve_rep2_std = _sliding_quantile_stat(
            df_rep2,
            x_col=x_col,
            y_col=y_col,
            stat="STD",
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
        )

    if curve_rep1.empty and curve_rep2.empty:
        return

    x_grid = np.linspace(0.0, 1.0, X_GRID_N)
    y1 = _interp_to_grid(curve_rep1, x_grid)
    y2 = _interp_to_grid(curve_rep2, x_grid)

    y1_std = _interp_to_grid(curve_rep1_std, x_grid) if do_shade else None
    y2_std = _interp_to_grid(curve_rep2_std, x_grid) if do_shade else None

    stat_label = {"CV": "CV", "STD": "STD", "MEAN": "Mean"}[stat]

    sns.set_theme(style="whitegrid")
    fig, ax = plt.subplots(figsize=(8.2, 5.0), dpi=300)

    # Shaded +/- STD around mean (only if requested)
    if do_shade:
        if y1_std is not None and (not np.all(np.isnan(y1))) and (not np.all(np.isnan(y1_std))):
            m = np.isfinite(y1) & np.isfinite(y1_std)
            ax.fill_between(
                x_grid,
                y1 - y1_std,
                y1 + y1_std,
                where=m,
                color="C0",
                alpha=0.18,
                linewidth=0,
            )
        if y2_std is not None and (not np.all(np.isnan(y2))) and (not np.all(np.isnan(y2_std))):
            m = np.isfinite(y2) & np.isfinite(y2_std)
            ax.fill_between(
                x_grid,
                y2 - y2_std,
                y2 + y2_std,
                where=m,
                color="C2",
                alpha=0.18,
                linewidth=0,
            )

    # Lines
    if not np.all(np.isnan(y1)):
        ax.plot(x_grid, y1, color="C0", lw=2.6, label="rep1")
    if not np.all(np.isnan(y2)):
        ax.plot(x_grid, y2, color="C2", lw=2.6, label="rep2")

    ax.set_title(f"{cell_line}")
    ax.set_xlabel(f"{_pretty_label(x_col)} quantile")

    if do_shade:
        ax.set_ylabel(f"{stat_label}({_pretty_label(y_col)}) ± STD")
    else:
        ax.set_ylabel(f"{stat_label}({_pretty_label(y_col)})")

    # Y-limits with padding (include shaded bounds if present)
    finite = []
    for arr in (y1, y2):
        vv = arr[np.isfinite(arr)]
        if vv.size:
            finite.append(vv)

    if do_shade:
        for mean_arr, std_arr in ((y1, y1_std), (y2, y2_std)):
            if std_arr is None:
                continue
            m = np.isfinite(mean_arr) & np.isfinite(std_arr)
            if np.any(m):
                finite.append((mean_arr[m] - std_arr[m]))
                finite.append((mean_arr[m] + std_arr[m]))

    if finite:
        all_y = np.concatenate(finite)
        y_min = float(np.min(all_y))
        y_max = float(np.max(all_y))
        y_rng = y_max - y_min
        pad = 0.06 * y_rng if y_rng > 0 else (0.06 * abs(y_max) if y_max != 0 else 0.1)
        ax.set_ylim(y_min - pad, y_max + pad)

    # Smaller legend, no title
    ax.legend(
        frameon=True,
        fontsize=10,
        borderpad=0.25,
        labelspacing=0.25,
        handlelength=1.6,
        handletextpad=0.5,
        borderaxespad=0.4,
    )

    sns.despine(ax=ax)
    plt.tight_layout()

    if SAVE_PLOTS:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        safe_title = re.sub(r"[^\w\-_. ]", "", ax.get_title()).replace(" ", "_")
        shade_tag = "__shadedSTD" if do_shade else ""
        filename = f"{safe_title}__{stat}{shade_tag}__x-{x_col}__y-{y_col}.png"
        plt.savefig(os.path.join(SAVE_FOLDER, filename))

    plt.show()


# -----------------------
# Run for all cell types
# -----------------------
for cell_line in ("3T3", "mESCs", "hESCs"):
    plot_sliding_stat_reps_all_proteins(
        cell_line,
        x_col=X_COL,
        y_col=Y_COL,
        stat=STAT,
        std_state=STD_STATE,
        window_frac=WINDOW_FRAC,
        step_frac=STEP_FRAC,
        window_n=WINDOW_N,
        step_n=STEP_N,
    )
```

```python
# SLIDING ALL VS ALL (protein combined data)
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------
# Controls
# -----------------------
CELL_LINES = ("3T3", "mESCs", "hESCs")  # 3 columns (one per cell line)
REPS = ("rep1", "rep2")

# Choose ONE window definition:
WINDOW_N = None        # e.g. 1000  -> fixed number of cells per window
WINDOW_FRAC = 1 / 6    # e.g. 0.15  -> fixed fraction of dataset per window (ignored if WINDOW_N is set)

# How far to slide the window each step:
STEP_N = None          # e.g. 100   -> fixed step in cells (ignored if None)
STEP_FRAC = 0.01       # e.g. 0.02  -> step as fraction of dataset (ignored if STEP_N is set)

MIN_WINDOW_N = 50
X_GRID_N = 300

# Columns to consider for ordered pairs (x_col, y_col):
# -> for each ordered pair x!=y, we plot STAT(y) over sliding windows along x-quantiles
COL_LIST = [
    "area_TL",
    "k",
    "s",
    "mean_IF_B",
    "sum_IF_B",
    "ratio",
    "P_tot"
    # add your columns here, e.g.:
    # "mean_IF_DR",
    # "s",
]

# Which statistic to plot for y inside each window: "CV", "STD", or "MEAN"
STAT = "MEAN"

# If STAT == "MEAN" and STD_STATE == True, shade +/- STD around the mean line
STD_STATE = True

# Plot look
SHARE_Y = False  # set True if you want same y-scale across the 3 panels

# Saving (optional)
SAVE_PLOTS = True
SAVE_FOLDER = r"DR variability continuous plots\ALL_PROTEINS_POOLED__PAIRPLOTS"


# -----------------------
# Helpers
# -----------------------
def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _pretty_label(col_name: str) -> str:
    mapping = {
        "area_TL": "Area",
        "k": "k",
        "mean_IF_DR": "mean IF (DR)",
        "mean_DR": "mean DR",
        "DR": "DR",
        "s": "s",
    }
    return mapping.get(col_name, col_name)


def _sanitize_filename(s: str) -> str:
    s = re.sub(r"[^\w\-_. ]", "", str(s))
    s = s.strip().replace(" ", "_")
    return s if s else "plot"


def _pool_before_threshold_all_proteins(full_dataset, cell_line, rep):
    """Pool before_threshold across ALL proteins and BOTH wells for a given cell_line + rep."""
    parts = []
    proteins = list(full_dataset.get(cell_line, {}).keys())
    for protein in proteins:
        for well in ("well1", "well2"):
            node = (
                full_dataset
                .get(cell_line, {})
                .get(protein, {})
                .get(rep, {})
                .get(well, {})
            )
            parts.append(_as_df(node.get("before_threshold", [])))
    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def build_pooled_data(full_dataset, cell_lines=CELL_LINES, reps=REPS):
    """Cache pooled DataFrames once (important when looping over many column pairs)."""
    pooled = {cl: {} for cl in cell_lines}
    for cl in cell_lines:
        for rep in reps:
            pooled[cl][rep] = _pool_before_threshold_all_proteins(full_dataset, cl, rep)
    return pooled


def _sliding_quantile_mean_std(
    df,
    *,
    x_col,
    y_col,
    window_frac,
    step_frac,
    window_n,
    step_n,
    min_window_n,
):
    """
    Sliding windows along sorted x_col:
    returns DataFrame with q_center and mean/std of y_col in window.
    """
    out_cols = ["q_center", "x_median", "mean", "std", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({x_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[x_col, y_col]].copy()
    work[x_col] = pd.to_numeric(work[x_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[x_col, y_col]).sort_values(x_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))

    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    x_vals = work[x_col].to_numpy(dtype=float)
    y_vals = work[y_col].to_numpy(dtype=float)

    rows = []
    denom = max(1, n - 1)

    for start in range(0, n - window_n + 1, step_n):
        end = start + window_n
        x_sub = x_vals[start:end]
        y_sub = y_vals[start:end]

        mu = float(np.mean(y_sub))
        sd = float(np.std(y_sub, ddof=1)) if len(y_sub) >= 2 else 0.0

        q_center = (start + 0.5 * (window_n - 1)) / denom
        x_median = float(np.median(x_sub))
        rows.append((q_center, x_median, mu, sd, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="mean"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]
    order = np.argsort(x)
    x = x[order]
    y = y[order]

    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]
    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def plot_triptych_for_pair(
    pooled,  # output of build_pooled_data(...)
    *,
    x_col,
    y_col,
    stat,
    std_state,
    cell_lines=CELL_LINES,
    window_frac=WINDOW_FRAC,
    step_frac=STEP_FRAC,
    window_n=WINDOW_N,
    step_n=STEP_N,
    share_y=SHARE_Y,
):
    stat = str(stat).upper().strip()
    do_shade = (stat == "MEAN") and bool(std_state)

    sns.set_theme(style="whitegrid")
    fig, axes = plt.subplots(1, 3, figsize=(14.5, 4.8), dpi=300, sharey=share_y)
    x_grid = np.linspace(0.0, 1.0, X_GRID_N)

    # Legend only once (smaller box, no title)
    legend_done = False

    for ax, cl in zip(axes, cell_lines):
        # compute sliding mean/std once; derive MEAN/STD/CV from it
        curves = {}
        for rep, color in (("rep1", "C0"), ("rep2", "C2")):
            df = pooled.get(cl, {}).get(rep, pd.DataFrame())
            curve = _sliding_quantile_mean_std(
                df,
                x_col=x_col,
                y_col=y_col,
                window_frac=window_frac,
                step_frac=step_frac,
                window_n=window_n,
                step_n=step_n,
                min_window_n=MIN_WINDOW_N,
            )
            curves[rep] = (curve, color)

        # derive y arrays
        for rep in ("rep1", "rep2"):
            curve, color = curves[rep]
            mean = _interp_to_grid(curve, x_grid, y_col="mean")
            std = _interp_to_grid(curve, x_grid, y_col="std")

            if stat == "MEAN":
                y = mean
            elif stat == "STD":
                y = std
            elif stat == "CV":
                y = np.full_like(mean, np.nan, dtype=float)
                m = np.isfinite(mean) & np.isfinite(std) & (mean != 0)
                y[m] = std[m] / np.abs(mean[m])
            else:
                raise ValueError("STAT must be one of: 'CV', 'STD', 'MEAN'")

            # shade +/- STD around MEAN if requested
            if do_shade and stat == "MEAN":
                m = np.isfinite(mean) & np.isfinite(std)
                if np.any(m):
                    ax.fill_between(
                        x_grid,
                        mean - std,
                        mean + std,
                        where=m,
                        color=color,
                        alpha=0.18,
                        linewidth=0,
                    )

            if not np.all(np.isnan(y)):
                ax.plot(x_grid, y, color=color, lw=2.6, label=rep)

        ax.set_title(cl)
        ax.set_xlabel(f"{_pretty_label(x_col)} quantile")
        stat_label = {"CV": "CV", "STD": "STD", "MEAN": "Mean"}[stat]
        ax.set_ylabel(f"{stat_label}({_pretty_label(y_col)})")

        # Smaller legend, no title (only on first axis to keep things clean)
        if not legend_done:
            ax.legend(
                frameon=True,
                fontsize=10,
                borderpad=0.25,
                labelspacing=0.25,
                handlelength=1.6,
                handletextpad=0.5,
                borderaxespad=0.4,
            )
            legend_done = True
        else:
            leg = ax.get_legend()
            if leg is not None:
                leg.remove()

    sns.despine(fig=fig)
    fig.tight_layout()

    # Optional: save per pair
    if SAVE_PLOTS:
        os.makedirs(SAVE_FOLDER, exist_ok=True)
        shade_tag = "__shadedSTD" if do_shade else ""
        filename = _sanitize_filename(f"{stat}{shade_tag}__x-{x_col}__y-{y_col}.png")
        fig.savefig(os.path.join(SAVE_FOLDER, filename))

    plt.show()


# -----------------------
# Run: build cache once, then plot all ordered pairs
# -----------------------
pooled = build_pooled_data(full_dataset, cell_lines=CELL_LINES, reps=REPS)

for x_col in COL_LIST:
    for y_col in COL_LIST:
        if x_col == y_col:
            continue  # ordered pairs, exclude self-pairs
        plot_triptych_for_pair(
            pooled,
            x_col=x_col,
            y_col=y_col,
            stat=STAT,
            std_state=STD_STATE,
        )
```

```python
# SLIDING ALL VS ALL
# Sliding-window (continuous) DR variability along k-quantiles, using the UNSPLIT data (before_threshold).
#
# Idea: sort cells by k, take a fixed-mass window (fixed number of cells == fixed area under the empirical k distribution),
# compute CV = std(mean_IF_DR) / mean(mean_IF_DR) in that window, then slide the window across k.
#
# What this version does (per your split):
# - Plots 4 curves: rep1_well1, rep1_well2, rep2_well1, rep2_well2
# - Adds 2 averages: rep1_avg = mean(rep1_well1, rep1_well2), rep2_avg = mean(rep2_well1, rep2_well2)
# - Adds y-axis padding (min - margin, max + margin)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# --- Controls ---
# Choose ONE of these window definitions:
WINDOW_N = None        # e.g. 1000  -> fixed number of cells per window (recommended if you want “same #cells”)
WINDOW_FRAC = 0.15     # e.g. 0.15  -> fixed fraction of the dataset per window (ignored if WINDOW_N is set)

# How far to slide the window each step:
STEP_N = None          # e.g. 100   -> fixed step in cells (ignored if None)
STEP_FRAC = 0.01      # e.g. 0.02  -> step as fraction of dataset (ignored if STEP_N is set)

MIN_WINDOW_N = 50

# Common x-grid for plotting/averaging (0..1 corresponds to k-quantiles)
X_GRID_N = 250


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []

    reps = (rep,) if rep is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")

    for rep_ in reps:
        for well_ in wells:
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep_, {}).get(well_, {})
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(
    df,
    *,
    k_col="k",
    y_col="mean_IF_DR",
    window_frac=0.20,
    step_frac=0.01,
    window_n=None,
    step_n=None,
    min_window_n=50,
):
    """Return a curve of CV(y)=sd(y)/mean(y) over sliding windows along sorted k.

    Output columns:
    - q_center: window center position in quantile space (0..1)
    - k_median: median k in the window
    - cv: sd(y)/mean(y)
    - n_window: number of points in each window
    """
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)

    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))
    
    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))
    #print(f"DEBUG: n={n}, window_n={window_n}, step_n={step_n}")
    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    denom = max(1, n - 1)
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy()

        mu = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd #/ mu

        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[k_col].to_numpy()))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="cv"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    # De-duplicate x (rare but can happen); keep first occurrence
    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]

    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def plot_sliding_cv_groups(
    cell_line,
    protein,
    *,
    window_frac,
    step_frac,
    window_n,
    step_n,
    ax,   # <- add axis argument
    x_col,
    y_col,
):
    curves = {
        "rep1_well1": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep1", well="well1"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
        "rep1_well2": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep1", well="well2"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
        "rep2_well1": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep2", well="well1"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
        "rep2_well2": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep2", well="well2"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
    }

    if all(c.empty for c in curves.values()):
        return

    x_grid = np.linspace(0.0, 1.0, X_GRID_N)
    y_grid = {name: _interp_to_grid(curve, x_grid) for name, curve in curves.items()}

    # averages
    for rep in ("rep1", "rep2"):

        a = y_grid[f"{rep}_well1"]
        b = y_grid[f"{rep}_well2"]

        stack = np.vstack([a, b])

        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            avg = np.nanmean(stack, axis=0)

        count = np.isfinite(stack).sum(axis=0)
        avg[count < 2] = np.nan

        y_grid[f"{rep}_avg"] = avg

    style = {
        "rep1_well1": dict(color="C0", lw=2.0),
        "rep1_well2": dict(color="C1", lw=2.0),
        "rep2_well1": dict(color="C2", lw=2.0),
        "rep2_well2": dict(color="C3", lw=2.0),
        "rep1_avg": dict(color="black", lw=2.4, ls="--"),
        "rep2_avg": dict(color="dimgray", lw=2.4, ls="--"),
    }

    for key in style.keys():
        y = y_grid.get(key)

        if y is None or np.all(np.isnan(y)):
            continue

        ax.plot(x_grid, y, label=key, **style[key])

    ax.set_title(f"{cell_line} | {protein}")
    ax.set_xlabel(x_col)
    ax.set_ylabel(y_col)
    ax.legend(fontsize=7)


# Plot for all proteins and both cell lines
lst = ["area_TL", "k", "s", "mean_TL_G", "sum_TL_G", "mean_TL_R", "sum_TL_R", "area_IF", "mean_IF_G", "sum_IF_G", "mean_IF_R", "sum_IF_R", "mean_IF_DR", "sum_IF_DR", "mean_IF_B", "sum_IF_B", "ratio", "sum_BG", "P_tot"]
for i in range(len(lst)):
    for j in range(i + 1, len(lst)):
        x = lst[i]
        y = lst[j]
        sns.set_style("whitegrid")

        proteins_3t3 = list(full_dataset["3T3"].keys())
        proteins_mescs = list(full_dataset["mESCs"].keys())
        proteins_hescs = list(full_dataset["hESCs"].keys())
        fig, axes = plt.subplots(
            nrows=3,
            ncols=5,
            figsize=(25, 10),
            dpi=300,
            sharex=True,
            sharey=True,
        )

        for col in range(5):

            plot_sliding_cv_groups(
                "3T3",
                proteins_3t3[col],
                window_frac=WINDOW_FRAC,
                step_frac=STEP_FRAC,
                window_n=WINDOW_N,
                step_n=STEP_N,
                x_col=x,
                y_col=y,
                ax=axes[0, col],
            )

            plot_sliding_cv_groups(
                "mESCs",
                proteins_mescs[col],
                window_frac=WINDOW_FRAC,
                step_frac=STEP_FRAC,
                window_n=WINDOW_N,
                step_n=STEP_N,
                ax=axes[1, col],
                x_col=x,
                y_col=y,
            ),
        for col in range(4):
            plot_sliding_cv_groups(
                "hESCs",
                proteins_hescs[col],
                window_frac=WINDOW_FRAC,
                step_frac=STEP_FRAC,
                window_n=WINDOW_N,
                step_n=STEP_N,
                ax=axes[2, col],
                x_col=x,
                y_col=y,
            )

        plt.tight_layout()
        plt.show()
```

```python
# Sliding ALL VS MEAN_IF_DR
# Sliding-window (continuous) DR variability along k-quantiles, using the UNSPLIT data (before_threshold).
#
# Idea: sort cells by k, take a fixed-mass window (fixed number of cells == fixed area under the empirical k distribution),
# compute CV = std(mean_IF_DR) / mean(mean_IF_DR) in that window, then slide the window across k.
#
# What this version does (per your split):
# - Plots 4 curves: rep1_well1, rep1_well2, rep2_well1, rep2_well2
# - Adds 2 averages: rep1_avg = mean(rep1_well1, rep1_well2), rep2_avg = mean(rep2_well1, rep2_well2)
# - Adds y-axis padding (min - margin, max + margin)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# --- Controls ---
# Choose ONE of these window definitions:
WINDOW_N = None        # e.g. 1000  -> fixed number of cells per window (recommended if you want “same #cells”)
WINDOW_FRAC = 0.15     # e.g. 0.15  -> fixed fraction of the dataset per window (ignored if WINDOW_N is set)

# How far to slide the window each step:
STEP_N = None          # e.g. 100   -> fixed step in cells (ignored if None)
STEP_FRAC = 0.01      # e.g. 0.02  -> step as fraction of dataset (ignored if STEP_N is set)

MIN_WINDOW_N = 50

# Common x-grid for plotting/averaging (0..1 corresponds to k-quantiles)
X_GRID_N = 250


def _as_df(x):
    if isinstance(x, list):
        return pd.concat(x, ignore_index=True) if len(x) else pd.DataFrame()
    if isinstance(x, pd.DataFrame):
        return x
    return pd.DataFrame()


def _concat_before_threshold(full_dataset, cell_line, protein, *, rep=None, well=None):
    parts = []

    reps = (rep,) if rep is not None else ("rep1", "rep2")
    wells = (well,) if well is not None else ("well1", "well2")

    for rep_ in reps:
        for well_ in wells:
            node = full_dataset.get(cell_line, {}).get(protein, {}).get(rep_, {}).get(well_, {})
            parts.append(_as_df(node.get("before_threshold", [])))

    parts = [p for p in parts if isinstance(p, pd.DataFrame) and not p.empty]
    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()


def _sliding_k_quantile_cv(
    df,
    *,
    k_col="k",
    y_col="mean_IF_DR",
    window_frac=0.20,
    step_frac=0.01,
    window_n=None,
    step_n=None,
    min_window_n=50,
):
    """Return a curve of CV(y)=sd(y)/mean(y) over sliding windows along sorted k.

    Output columns:
    - q_center: window center position in quantile space (0..1)
    - k_median: median k in the window
    - cv: sd(y)/mean(y)
    - n_window: number of points in each window
    """
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)

    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))
    
    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))
    #print(f"DEBUG: n={n}, window_n={window_n}, step_n={step_n}")
    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    denom = max(1, n - 1)
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy()

        mu = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd #/ mu

        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[k_col].to_numpy()))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def _interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="cv"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    # De-duplicate x (rare but can happen); keep first occurrence
    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]

    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


def plot_sliding_cv_groups(
    cell_line,
    protein,
    *,
    window_frac,
    step_frac,
    window_n,
    step_n,
    ax,   # <- add axis argument
    x_col,
    y_col,
):
    curves = {
        "rep1_well1": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep1", well="well1"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
        "rep1_well2": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep1", well="well2"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
        "rep2_well1": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep2", well="well1"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
        "rep2_well2": _sliding_k_quantile_cv(
            _concat_before_threshold(full_dataset, cell_line, protein, rep="rep2", well="well2"),
            window_frac=window_frac,
            step_frac=step_frac,
            window_n=window_n,
            step_n=step_n,
            min_window_n=MIN_WINDOW_N,
            k_col=x_col,
            y_col=y_col,
        ),
    }

    if all(c.empty for c in curves.values()):
        return

    x_grid = np.linspace(0.0, 1.0, X_GRID_N)
    y_grid = {name: _interp_to_grid(curve, x_grid) for name, curve in curves.items()}

    # averages
    for rep in ("rep1", "rep2"):

        a = y_grid[f"{rep}_well1"]
        b = y_grid[f"{rep}_well2"]

        stack = np.vstack([a, b])

        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            avg = np.nanmean(stack, axis=0)

        count = np.isfinite(stack).sum(axis=0)
        avg[count < 2] = np.nan

        y_grid[f"{rep}_avg"] = avg

    style = {
        "rep1_well1": dict(color="C0", lw=2.0),
        "rep1_well2": dict(color="C1", lw=2.0),
        "rep2_well1": dict(color="C2", lw=2.0),
        "rep2_well2": dict(color="C3", lw=2.0),
        "rep1_avg": dict(color="black", lw=2.4, ls="--"),
        "rep2_avg": dict(color="dimgray", lw=2.4, ls="--"),
    }

    for key in style.keys():
        y = y_grid.get(key)

        if y is None or np.all(np.isnan(y)):
            continue

        ax.plot(x_grid, y, label=key, **style[key])

    ax.set_title(f"{cell_line} | {protein}")
    ax.set_xlabel(x_col)
    ax.set_ylabel(y_col)
    ax.legend(fontsize=7)


# Plot for all proteins and both cell lines
lst = ["area_TL", "k", "s", "mean_TL_G", "sum_TL_G", "mean_TL_R", "sum_TL_R", "area_IF", "mean_IF_G", "sum_IF_G", "mean_IF_R", "sum_IF_R", "sum_IF_DR", "mean_IF_B", "sum_IF_B", "ratio", "sum_BG", "P_tot"]
for i in range(len(lst)):
    x = lst[i]
    y = "mean_IF_DR"
    sns.set_style("whitegrid")

    proteins_3t3 = list(full_dataset["3T3"].keys())
    proteins_mescs = list(full_dataset["mESCs"].keys())
    proteins_hescs = list(full_dataset["hESCs"].keys())
    print(proteins_hescs)
    fig, axes = plt.subplots(
        nrows=3,
        ncols=5,
        figsize=(25, 10),
        dpi=300,
        sharex=True,
        sharey=True,
    )

    for col in range(5):

        plot_sliding_cv_groups(
            "3T3",
            proteins_3t3[col],
            window_frac=WINDOW_FRAC,
            step_frac=STEP_FRAC,
            window_n=WINDOW_N,
            step_n=STEP_N,
            x_col=x,
            y_col=y,
            ax=axes[0, col],
        )

        plot_sliding_cv_groups(
            "mESCs",
            proteins_mescs[col],
            window_frac=WINDOW_FRAC,
            step_frac=STEP_FRAC,
            window_n=WINDOW_N,
            step_n=STEP_N,
            ax=axes[1, col],
            x_col=x,
            y_col=y,
        )
    for col in range(4):
        plot_sliding_cv_groups(
            "hESCs",
            proteins_hescs[col],
            window_frac=WINDOW_FRAC,
            step_frac=STEP_FRAC,
            window_n=WINDOW_N,
            step_n=STEP_N,
            ax=axes[2, col],
            x_col=x,
            y_col=y,
        )

    plt.tight_layout()
    plt.show()
```

```python
import ast
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def load_id_specs_from_txt(filepath):
    id_specs = {}

    with open(filepath, "r") as f:
        for raw in f:
            line = raw.strip()

            if not line or line.startswith("#"):
                continue

            track_id, spec = line.split(":", 1)

            track_id = int(track_id)
            ((start, end), direction) = ast.literal_eval(spec)

            start = int(start)
            end = int(end)
            direction = int(direction)

            if start > end:
                start, end = end, start

            id_specs[track_id] = ((start, end), direction)

    return id_specs


def build_aligned_dataset(df, id_specs):
    """
    Keep only frames defined in id_specs and compute aligned time.
    """

    out = []

    for tid, ((start, end), direction) in id_specs.items():

        sub = df[
            (df["display_track_id"] == tid)
            & (df["frame"] >= start)
            & (df["frame"] <= end)
        ].copy()

        if sub.empty:
            continue

        if direction == 0:
            sub["time"] = sub["frame"] - start

        else:  # direction == -1
            # event occurs at t=0
            sub["time"] = sub["frame"] - (end + 1)

        out.append(sub)

    if not out:
        return pd.DataFrame()

    return pd.concat(out, ignore_index=True)


def summarize_metric(df, metric):
    return (
        df.groupby("time")[metric]
        .agg(
            mean="mean",
            std="std",
            n="count"
        )
        .reset_index()
        .sort_values("time")
    )


# ------------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------------

path_csv = r"Z:\\EPFL\\Biology\\bachelor\\tracking-20260527T024245Z-3-001\\tracking\\R01C01\\track_traces_green_red_ratio_long.csv"
path_specs = r"Z:\\EPFL\\Biology\\bachelor\\tracking-20260527T024245Z-3-001\\tracking\\R01C01\\cells_R01C01.txt"

df = pd.read_csv(path_csv)

id_specs = load_id_specs_from_txt(path_specs)

aligned = build_aligned_dataset(df, id_specs)

# ------------------------------------------------------------------
# METRICS TO PLOT
# ------------------------------------------------------------------

metrics = [
    "mean_G",
    "mean_R",
    "k_comp",
    "ratio_G_over_R",
]

summaries = {
    m: summarize_metric(aligned, m)
    for m in metrics
}

# ------------------------------------------------------------------
# PLOT
# ------------------------------------------------------------------

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.ravel()

for ax, metric in zip(axes, metrics):

    s = summaries[metric]

    ax.scatter(
        s["time"],
        s["mean"],
        s=20
    )

    ax.set_title(metric)
    ax.set_xlabel("Aligned time")
    ax.set_ylabel("Mean")
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# CHX sliding-window CV vs k-quantile (x from 0..1), 8 lines:
# - DMSO control: R01
# - CHX: R02 (highest) -> R08 (lowest)
# Data: column C02 (c_num == 2), rows R01..R08
# Protein: Sox2 (in title)

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# -----------------------
# Controls
# -----------------------
CHX_COL = 2                      # C02
ROWS = list(range(1, 9))         # R01..R08
DMSO_ROW = 1                     # R01

CELL_LINE = "hESCs"
PROTEIN = "sox2"

K_COL = "k"
Y_COL = "mean_IF_DR"             # CV computed on this: std/mean

# Sliding window settings (same style as your earlier sliding-k code)
WINDOW_N = None                  # e.g. 1200 (if set, overrides WINDOW_FRAC)
WINDOW_FRAC = 0.15
STEP_N = None                    # e.g. 120  (if set, overrides STEP_FRAC)
STEP_FRAC = 0.02
MIN_WINDOW_N = 80

# Common x-grid in quantile space
X_GRID_N = 250

# Filtering (mirrors your initialization cell logic)
TRIM_COLS = [K_COL, Y_COL]
NONNEG_COLS = [K_COL, Y_COL]
TRIM_Q_LOW, TRIM_Q_HIGH = 0.02, 0.98

# Save
SAVE = False
OUT_FOLDER = "plots"
OUT_NAME = f"{CELL_LINE}_{PROTEIN}_CHX_C02_sliding_quantile_CV.png"
DPI = 300


# -----------------------
# Helpers (same logic as init cell)
# -----------------------
def add_rc_numbers(df, root_col="if_root"):
    root_id = df[root_col].astype(str).map(lambda s: os.path.basename(os.path.normpath(s)))
    extracted = root_id.str.extract(r"^r(\d+)c(\d+)")
    out = df.copy()
    out["r_num"] = extracted[0].astype("Int64")
    out["c_num"] = extracted[1].astype("Int64")
    return out


def drop_any_negative(df, cols):
    cols_present = [c for c in (cols or []) if c in df.columns]
    if not cols_present:
        return df.copy()
    vals = df[cols_present].apply(pd.to_numeric, errors="coerce")
    return df[~(vals.lt(0).any(axis=1))].copy()


def drop_outside_quantiles(df, cols, q_low, q_high):
    cols_present = [c for c in (cols or []) if c in df.columns]
    if not cols_present:
        return df.copy()
    vals = df[cols_present].apply(pd.to_numeric, errors="coerce")
    lows = vals.quantile(q_low)
    highs = vals.quantile(q_high)
    keep = ((vals >= lows) & (vals <= highs) | vals.isna()).all(axis=1)
    return df[keep].copy()


def get_chx_condition_df(df_chx, *, r_num, c_num):
    sub = df_chx[(df_chx["r_num"] == r_num) & (df_chx["c_num"] == c_num)].copy()
    if sub.empty:
        return sub
    sub = drop_any_negative(sub, NONNEG_COLS)
    sub = drop_outside_quantiles(sub, TRIM_COLS, TRIM_Q_LOW, TRIM_Q_HIGH)
    return sub


def sliding_k_quantile_cv(
    df,
    *,
    k_col=K_COL,
    y_col=Y_COL,
    window_frac=WINDOW_FRAC,
    step_frac=STEP_FRAC,
    window_n=WINDOW_N,
    step_n=STEP_N,
    min_window_n=MIN_WINDOW_N,
):
    """
    Sort by k, slide a fixed-mass window, compute CV = std(y)/mean(y).

    Returns DataFrame with:
      - q_center: window center in quantile space (0..1)
      - k_median: median k in the window (optional diagnostic)
      - cv: std(y)/mean(y)
      - n_window
    """
    out_cols = ["q_center", "k_median", "cv", "n_window"]

    if df is None or df.empty:
        return pd.DataFrame(columns=out_cols)
    if not ({k_col, y_col} <= set(df.columns)):
        return pd.DataFrame(columns=out_cols)

    work = df[[k_col, y_col]].copy()
    work[k_col] = pd.to_numeric(work[k_col], errors="coerce")
    work[y_col] = pd.to_numeric(work[y_col], errors="coerce")
    work = work.dropna(subset=[k_col, y_col]).sort_values(k_col)

    n = len(work)
    if n < max(2, min_window_n):
        return pd.DataFrame(columns=out_cols)

    if window_n is None:
        window_n = int(round(window_frac * n))
    window_n = int(max(min_window_n, min(window_n, n)))

    if step_n is None:
        step_n = int(round(step_frac * n))
    step_n = int(max(1, step_n))

    if window_n > n:
        return pd.DataFrame(columns=out_cols)

    rows = []
    denom = max(1, n - 1)
    for start in range(0, n - window_n + 1, step_n):
        sub = work.iloc[start : start + window_n]
        y = sub[y_col].to_numpy(dtype=float)

        mu = float(np.mean(y))
        if not np.isfinite(mu) or mu == 0:
            cv = np.nan
        else:
            sd = float(np.std(y, ddof=1)) if len(y) >= 2 else 0.0
            cv = sd / mu

        q_center = (start + 0.5 * (window_n - 1)) / denom
        k_median = float(np.median(sub[k_col].to_numpy(dtype=float)))
        rows.append((q_center, k_median, cv, window_n))

    return pd.DataFrame(rows, columns=out_cols)


def interp_to_grid(curve, x_grid, *, x_col="q_center", y_col="cv"):
    if curve is None or curve.empty:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = curve[x_col].to_numpy(dtype=float)
    y = curve[y_col].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    x = x[mask]
    y = y[mask]

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]
    if len(x_unique) < 2:
        return np.full_like(x_grid, np.nan, dtype=float)

    return np.interp(x_grid, x_unique, y_unique, left=np.nan, right=np.nan)


# -----------------------
# Main
# -----------------------
if "df_9" not in globals():
    raise NameError("df_9 not found. Run your data-loading/init cell first (the one that reads csv_path9).")

df_chx = add_rc_numbers(df_9, root_col="if_root")

x_grid = np.linspace(0.0, 1.0, X_GRID_N)

# Curves: DMSO + CHX
curves = []

# DMSO (R01)
df_dmso = get_chx_condition_df(df_chx, r_num=DMSO_ROW, c_num=CHX_COL)
curve_dmso = sliding_k_quantile_cv(df_dmso)
y_dmso = interp_to_grid(curve_dmso, x_grid)
curves.append(("DMSO (R01)", y_dmso, dict(color="0.25", lw=2.8, ls="--")))

# CHX rows (R02..R08) highest -> lowest
chx_rows = [r for r in ROWS if r != DMSO_ROW]
cmap = plt.cm.viridis
colors = cmap(np.linspace(0.10, 0.95, len(chx_rows)))  # R02 darker -> R08 brighter

for r_num, colr in zip(chx_rows, colors):
    df_r = get_chx_condition_df(df_chx, r_num=r_num, c_num=CHX_COL)
    curve = sliding_k_quantile_cv(df_r)
    y = interp_to_grid(curve, x_grid)

    label = f"CHX R{r_num:02d}"
    if r_num == 2:
        label += " (high)"
    if r_num == 8:
        label += " (low)"

    curves.append((label, y, dict(color=colr, lw=2.2)))

# Plot
sns.set_style("whitegrid")
fig, ax = plt.subplots(figsize=(9.0, 5.4), dpi=DPI)

for label, y, style in curves:
    if np.all(np.isnan(y)):
        print(f"⚠️ Curve empty after filtering/windowing: {label}")
        continue
    ax.plot(x_grid, y, label=label, **style)

ax.set_xlim(0.1, 0.9)
ax.set_xlabel("k-quantile (window center)")
ax.set_ylabel(f"CV = std({Y_COL}) / mean({Y_COL})")
ax.set_title(f"{CELL_LINE} | {PROTEIN} | CHX (C02) sliding-window CV vs k-quantile", fontsize=13, pad=10)

ax.grid(True, axis="y", color="0.9", linewidth=1)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(title="Condition (R02 high → R08 low)", fontsize=9, title_fontsize=9, ncol=2, frameon=True, loc="upper right")

ax.text(
    0.01, 0.01,
    f"window_frac={WINDOW_FRAC}, step_frac={STEP_FRAC}, min_n={MIN_WINDOW_N}",
    transform=ax.transAxes, ha="left", va="bottom", fontsize=8, color="0.35"
)

plt.tight_layout()

if SAVE:
    os.makedirs(OUT_FOLDER, exist_ok=True)
    out_path = os.path.join(OUT_FOLDER, OUT_NAME)
    fig.savefig(out_path, dpi=DPI, bbox_inches="tight")
    print(f"Saved: {out_path}")

plt.show()
```

```python
# Bar plots for CHX conditions (C02, R01..R08):
# 1) CV of mean_IF_DR per condition  (CV = std/mean)
# 2) Mean k per condition
#
# R01 = DMSO control, R02 = highest CHX, ... R08 = lowest CHX

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# -----------------------
# Controls
# -----------------------
CHX_COL = 2
ROWS = list(range(1, 9))   # R01..R08
DMSO_ROW = 1

K_COL = "k"
Y_COL = "mean_IF_DR"

TRIM_COLS = [K_COL, Y_COL]
NONNEG_COLS = [K_COL, Y_COL]
TRIM_Q_LOW, TRIM_Q_HIGH = 0.02, 0.98

DPI = 300


# -----------------------
# Helpers (same as your init cell)
# -----------------------
def add_rc_numbers(df, root_col="if_root"):
    root_id = df[root_col].astype(str).map(lambda s: os.path.basename(os.path.normpath(s)))
    extracted = root_id.str.extract(r"^r(\d+)c(\d+)")
    out = df.copy()
    out["r_num"] = extracted[0].astype("Int64")
    out["c_num"] = extracted[1].astype("Int64")
    return out


def drop_any_negative(df, cols):
    cols_present = [c for c in (cols or []) if c in df.columns]
    if not cols_present:
        return df.copy()
    vals = df[cols_present].apply(pd.to_numeric, errors="coerce")
    return df[~(vals.lt(0).any(axis=1))].copy()


def drop_outside_quantiles(df, cols, q_low, q_high):
    cols_present = [c for c in (cols or []) if c in df.columns]
    if not cols_present:
        return df.copy()
    vals = df[cols_present].apply(pd.to_numeric, errors="coerce")
    lows = vals.quantile(q_low)
    highs = vals.quantile(q_high)
    keep = ((vals >= lows) & (vals <= highs) | vals.isna()).all(axis=1)
    return df[keep].copy()


def get_chx_condition_df(df_chx, *, r_num, c_num):
    sub = df_chx[(df_chx["r_num"] == r_num) & (df_chx["c_num"] == c_num)].copy()
    if sub.empty:
        return sub
    sub = drop_any_negative(sub, NONNEG_COLS)
    sub = drop_outside_quantiles(sub, TRIM_COLS, TRIM_Q_LOW, TRIM_Q_HIGH)
    return sub


def cv_std_over_mean(series):
    v = pd.to_numeric(series, errors="coerce").dropna().to_numpy(dtype=float)
    if len(v) == 0:
        return np.nan
    mu = float(np.mean(v))
    if not np.isfinite(mu) or mu == 0:
        return np.nan
    sd = float(np.std(v, ddof=1)) if len(v) >= 2 else 0.0
    return sd / mu


# -----------------------
# Main
# -----------------------
if "df_9" not in globals():
    raise NameError("df_9 not found. Run your data-loading/init cell first (the one that reads csv_path9).")

df_chx = add_rc_numbers(df_9, root_col="if_root")

rows_out = []
for r in ROWS:
    df_r = get_chx_condition_df(df_chx, r_num=r, c_num=CHX_COL)

    label = "DMSO (R01)" if r == DMSO_ROW else f"CHX (R{r:02d})"
    if r == 2:
        label += " high"
    if r == 8:
        label += " low"

    if df_r is None or df_r.empty:
        rows_out.append((r, label, np.nan, np.nan, 0))
        continue

    k_vals = pd.to_numeric(df_r.get(K_COL, pd.Series(dtype=float)), errors="coerce")
    y_vals = pd.to_numeric(df_r.get(Y_COL, pd.Series(dtype=float)), errors="coerce")

    mean_k = float(np.nanmean(k_vals.to_numpy(dtype=float))) if np.isfinite(k_vals).any() else np.nan
    cv_y = cv_std_over_mean(y_vals)

    n = int(np.isfinite(k_vals).sum())  # just to show sample size (k non-nan)
    rows_out.append((r, label, cv_y, mean_k, n))

summary = pd.DataFrame(rows_out, columns=["r_num", "condition", "cv_mean_IF_DR", "mean_k", "n"])
summary["condition"] = pd.Categorical(summary["condition"], categories=list(summary["condition"]), ordered=True)

display(summary)

# Colors: DMSO gray + CHX sequential
chx_rows = [r for r in ROWS if r != DMSO_ROW]
cmap = plt.cm.viridis
chx_colors = cmap(np.linspace(0.10, 0.95, len(chx_rows)))  # R02 darker -> R08 brighter
color_map = {"DMSO (R01)": "0.35"}
for r, colr in zip(chx_rows, chx_colors):
    # match the exact label string used above
    if r == 2:
        key = "CHX (R02) high"
    elif r == 8:
        key = "CHX (R08) low"
    else:
        key = f"CHX (R{r:02d})"
    color_map[key] = colr

bar_colors = [color_map.get(str(cond), "0.6") for cond in summary["condition"].astype(str).tolist()]

sns.set_style("whitegrid")

# 1) Bar plot: CV(mean_IF_DR) by condition
fig1, ax1 = plt.subplots(figsize=(10.5, 4.2), dpi=DPI)
ax1.bar(summary["condition"].astype(str), summary["cv_mean_IF_DR"], color=bar_colors, edgecolor="white", linewidth=0.8)
ax1.set_title("CHX (C02) | CV of mean_IF_DR by condition", fontsize=13, pad=10)
ax1.set_ylabel("CV = std(mean_IF_DR) / mean(mean_IF_DR)")
ax1.set_xlabel("")
ax1.tick_params(axis="x", rotation=30, labelsize=9)
ax1.grid(True, axis="y", color="0.9", linewidth=1)
ax1.spines["top"].set_visible(False)
ax1.spines["right"].set_visible(False)
plt.tight_layout()
plt.show()

# 2) Bar plot: mean k by condition
fig2, ax2 = plt.subplots(figsize=(10.5, 4.2), dpi=DPI)
ax2.bar(summary["condition"].astype(str), summary["mean_k"], color=bar_colors, edgecolor="white", linewidth=0.8)
ax2.set_title("CHX (C02) | Mean k by condition", fontsize=13, pad=10)
ax2.set_ylabel("Mean k")
ax2.set_xlabel("")
ax2.tick_params(axis="x", rotation=30, labelsize=9)
ax2.grid(True, axis="y", color="0.9", linewidth=1)
ax2.spines["top"].set_visible(False)
ax2.spines["right"].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def build_filtered_distribution(dataset_dfs, protein_map, id_list, wanted_list,
                                q_low=0.02, q_high=0.98):

    rows = []

    for i, (cell_line, protein) in enumerate(protein_map):

        df1 = dataset_dfs[id_list[i][0] - 1]
        df2 = dataset_dfs[id_list[i][1] - 1]

        r1, c1, r2, c2 = wanted_list[i]

        for df, r_list, c_list, rep in [
            (df1, r1, c1, "rep1"),
            (df2, r2, c2, "rep2")
        ]:

            df = add_rc_numbers(df)

            c_list = _as_list_or_none(c_list)

            mask = df["r_num"].isin(r_list)
            if c_list is not None:
                mask &= df["c_num"].isin(c_list)

            sub = df[mask].copy()

            if len(sub) == 0:
                continue

            sub["s"] = pd.to_numeric(sub["s"], errors="coerce")
            sub["k"] = pd.to_numeric(sub["k"], errors="coerce")
            sub = sub.dropna(subset=["s", "k"])

            # -------------------------
            # OUTLIER FILTER (IMPORTANT)
            # -------------------------
            for col in ["s", "k"]:
                lo, hi = sub[col].quantile([q_low, q_high])
                sub = sub[(sub[col] >= lo) & (sub[col] <= hi)]

            sub["cell_line"] = cell_line
            sub["protein"] = protein
            sub["replicate"] = rep

            rows.append(sub[["cell_line", "protein", "replicate", "s", "k"]])

    return pd.concat(rows, ignore_index=True)


df_dist = build_filtered_distribution(dataset_dfs, protein_map, id_list, wanted_list)
```

```python
rep_colors = {
    "rep1": "#1f77b4",  # blue
    "rep2": "#ff7f0e"   # orange
}
```

```python
sns.set_style("whitegrid")

g = sns.FacetGrid(
    df_dist,
    col="cell_line",
    row="protein",
    hue="replicate",
    sharex=False,
    sharey=False,
    height=2.0,
    aspect=1.3,
    margin_titles=True
)

g.map_dataframe(
    sns.kdeplot,
    x="k",
    fill=True,
    alpha=0.35,
    linewidth=1
)

g.add_legend(title="Replicate")
g.set_axis_labels("k", "Density")
g.fig.suptitle("Distributions of k across proteins, cell lines, and replicates", y=1.02)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6,5), dpi=300)

point_colors = [get_color(c) for c in summary["CHX_conc"]]

ax.scatter(
    summary["CHX_conc"],
    summary["mean_s"],
    s=120,
    c=point_colors,
    edgecolor="black",
    linewidth=0.6
)

ax.set_xlabel("Cycloheximide concentration (μg/mL)")
ax.set_ylabel("Mean s (AU)")
ax.set_title("Effect of CHX on synthesis rate")

ax.legend(
    handles=legend_handles,
    title="CHX concentration",
    fontsize=8,
    title_fontsize=9,
    loc="best"
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.colors import LogNorm

CHX_CONC = {
    1: 0.0,
    2: 0.4,
    3: 0.2,
    4: 0.1,
    5: 0.05,
    6: 0.025,
    7: 0.0125,
    8: 0.00625
}

fig, ax = plt.subplots(figsize=(6,5), dpi=300)

positive_concs = [c for c in CHX_CONC.values() if c > 0]
norm = LogNorm(vmin=min(positive_concs), vmax=max(positive_concs))

legend_handles = []

for r in sorted(CHX_CONC.keys()):

    sub = summary[summary["r_num"] == r]
    if len(sub) == 0:
        continue

    x = sub["mean_s"].values[0]
    y = sub["cv_mean_IF_DR"].values[0]
    xerr = sub["std_s"].values[0]

    conc = CHX_CONC[r]

    if conc == 0:
        color = "black"
        label = "DMSO control"
    else:
        color = plt.cm.viridis(norm(conc))
        label = f"{conc:g} μg/mL"

    ax.errorbar(
        x, y,
        xerr=xerr,
        fmt='o',
        markersize=7,
        color=color,
        ecolor=color,
        elinewidth=1.2,
        capsize=3,
        markeredgecolor="black",
        alpha=0.9
    )

    legend_handles.append(
        Line2D(
            [0], [0],
            marker='o',
            linestyle='-',
            color=color,
            markerfacecolor=color,
            markeredgecolor='black',
            markersize=6,
            linewidth=1.5,
            label=label
        )
    )

ax.set_xlabel("Mean s (AU)")
ax.set_ylabel("Sox2 CV")
ax.set_title("Protein variability vs synthesis rate")

ax.legend(handles=legend_handles,
          title="CHX concentration",
          fontsize=7,
          title_fontsize=8,
          loc="best")

plt.tight_layout()
plt.show()
```
